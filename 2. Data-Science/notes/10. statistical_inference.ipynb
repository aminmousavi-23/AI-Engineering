# Statistical Inference — استنباط آماری

## تعریف

**Statistical Inference** فرایند استفاده از داده‌های یک نمونه برای تخمین ویژگی‌های جامعه آماری و آزمون فرضیه‌های مربوط به آن است.

## مفاهیم اصلی

- **Population:** کل جامعه مورد مطالعه.
- **Sample:** زیرمجموعه‌ای از جامعه.
- **Parameter:** ویژگی واقعی جامعه، مانند میانگین جامعه.
- **Statistic:** مقداری که از نمونه محاسبه می‌شود.
- **Estimation:** تخمین پارامتر جامعه.
- **Point Estimation:** تخمین پارامتر با یک عدد.
- **Interval Estimation:** تخمین پارامتر با یک بازه.
- **Confidence Interval:** بازه‌ای که با روش آماری مشخص، سطح پوشش معینی دارد.
- **Hypothesis Testing:** بررسی یک ادعا درباره جامعه.
- **Null Hypothesis (H0):** فرض صفر.
- **Alternative Hypothesis (H1):** فرض مقابل.
- **P-value:** احتمال مشاهده نتیجه‌ای به‌اندازه نتیجه فعلی یا شدیدتر، تحت فرض صفر و مدل آماری آزمون.
- **Significance Level (α):** سطح معناداری آزمون؛ مقدار رایج آن 0.05 است.

## دو کاربرد اصلی

### 1. Estimation

تخمین پارامترهای جامعه با استفاده از نمونه.

### 2. Hypothesis Testing

ارزیابی شواهد آماری برای رد یا رد نکردن فرض صفر.

## نکات مهم

- میانگین نمونه لزوماً با میانگین جامعه برابر نیست.
- بزرگ‌تر بودن نمونه معمولاً دقت تخمین را افزایش می‌دهد، اما نمونه‌گیری جانبدارانه را اصلاح نمی‌کند.
- اگر `p-value < α` باشد، فرض صفر را در سطح معناداری انتخاب‌شده رد می‌کنیم.
- اگر `p-value ≥ α` باشد، شواهد کافی برای رد فرض صفر نداریم.
- Confidence Interval و P-value میزان عدم‌قطعیت را از جنبه‌های متفاوت بررسی می‌کنند.
- استنباط معتبر به روش نمونه‌گیری، اندازه نمونه و فرض‌های روش آماری وابسته است.

## کاربرد در Data Science

- تخمین میانگین یا نسبت در جامعه.
- مقایسه دو گروه از کاربران.
- بررسی اثر یک تغییر در محصول.
- تحلیل نتایج آزمایش A/B.
- بررسی ادعاهای آماری درباره داده‌ها.

# تمرین ۱ — Estimation & Confidence Interval

## بخش اول: ساخت دیتاست

```python
import numpy as np
from scipy import stats

income = np.array([
    32, 35, 38, 40, 42,
    37, 45, 39, 41, 36,
    44, 43, 34, 47, 40,
    38, 42, 46, 35, 41
])
```

## بخش دوم: محاسبه تخمین‌ها

1. تعداد نمونه‌ها را محاسبه کن.
2. میانگین نمونه را با `np.mean()` محاسبه کن.
3. میانه نمونه را با `np.median()` محاسبه کن.
4. انحراف معیار نمونه را با `np.std(income, ddof=1)` محاسبه کن.

## بخش سوم: Confidence Interval

1. بازه اطمینان ۹۵٪ برای میانگین را با `stats.t.interval()` محاسبه کن.
2. مقدار `stats.sem(income)` را محاسبه کن.
3. میانگین نمونه و دو حد بازه اطمینان را چاپ کن.
4. توضیح بده چرا میانگین نمونه الزاماً با میانگین واقعی جامعه برابر نیست.
5. توضیح بده بازه اطمینان ۹۵٪ چه مفهومی دارد.

````python
import numpy as np
from scipy import stats

# داده‌های نمونه
income = np.array([
    32, 35, 38, 40, 42,
    37, 45, 39, 41, 36,
    44, 43, 34, 47, 40,
    38, 42, 46, 35, 41
])


# بخش ۱: تعداد نمونه‌ها
print("\nبخش ۱: تعداد نمونه‌ها")

sample_size = len(income)

print("Sample Size:", sample_size)


# بخش ۲: محاسبه میانگین نمونه
print("\nبخش ۲: میانگین نمونه")

sample_mean = np.mean(income)

print("Sample Mean:", sample_mean)


# بخش ۳: محاسبه میانه نمونه
print("\nبخش ۳: میانه نمونه")

sample_median = np.median(income)

print("Sample Median:", sample_median)


# بخش ۴: محاسبه انحراف معیار نمونه
print("\nبخش ۴: انحراف معیار نمونه")

sample_std = np.std(income, ddof=1)

print("Sample Standard Deviation:", sample_std)


# بخش ۵: محاسبه Standard Error
print("\nبخش ۵: خطای استاندارد میانگین")

standard_error = stats.sem(income)

print("Standard Error:", standard_error)


# بخش ۶: محاسبه بازه اطمینان ۹۵ درصد
print("\nبخش ۶: بازه اطمینان ۹۵ درصد")

confidence_interval = stats.t.interval(
    confidence=0.95,
    df=sample_size - 1,
    loc=sample_mean,
    scale=standard_error
)

print("95% Confidence Interval:", confidence_interval)
print("Lower Bound:", confidence_interval[0])
print("Upper Bound:", confidence_interval[1])


# بخش ۷: تفسیر نتایج
print("\nبخش ۷: تفسیر نتایج")

print(f"میانگین نمونه برابر با {sample_mean:.2f} است.")
print(f"میانه نمونه برابر با {sample_median:.2f} است.")
print(f"انحراف معیار نمونه برابر با {sample_std:.2f} است.")

print(
    f"بازه اطمینان ۹۵ درصد برای میانگین جامعه، "
    f"از {confidence_interval[0]:.2f} تا "
    f"{confidence_interval[1]:.2f} است."
)
````

# تمرین ۲ — Hypothesis Testing

## بخش اول: ساخت دیتاست

```python
import numpy as np
from scipy import stats

sales = np.array([
    52, 48, 55, 51, 49,
    54, 58, 47, 53, 56,
    50, 52, 57, 49, 55,
    51, 54, 46, 59, 53
])
```

فرض کن یک شرکت ادعا می‌کند میانگین فروش روزانه برابر با ۵۰ واحد است.

## بخش دوم: تعریف فرضیه‌ها

- فرض صفر (`H0`): میانگین جامعه برابر ۵۰ است.
- فرض مقابل (`H1`): میانگین جامعه با ۵۰ تفاوت دارد.

## بخش سوم: اجرای آزمون

1. میانگین نمونه را محاسبه کن.
2. با `stats.ttest_1samp()` آزمون را اجرا کن.
3. مقدار `t-statistic` و `p-value` را چاپ کن.
4. سطح معناداری را `alpha = 0.05` قرار بده.
5. اگر `p-value < alpha` بود، فرض صفر را رد کن؛ در غیر این صورت، فرض صفر را رد نکن.

## بخش چهارم: تفسیر

با توجه به نتیجه آزمون، توضیح بده:

- آیا شواهد آماری کافی برای تفاوت میانگین فروش با ۵۰ وجود دارد؟
- چرا رد نکردن فرض صفر به معنی اثبات آن نیست؟
- آیا نتیجه آزمون به‌تنهایی ثابت می‌کند تغییر مشاهده‌شده اهمیت عملی دارد؟ چرا؟

````python
import numpy as np
from scipy import stats

# داده‌های نمونه
sales = np.array([
    52, 48, 55, 51, 49,
    54, 58, 47, 53, 56,
    50, 52, 57, 49, 55,
    51, 54, 46, 59, 53
])


# بخش ۱: محاسبه میانگین نمونه
print("\nبخش ۱: میانگین نمونه")

sample_mean = np.mean(sales)

print("Sample Mean:", sample_mean)


# بخش ۲: تعریف فرضیه‌ها
print("\nبخش ۲: تعریف فرضیه‌ها")

print("H0: Population mean = 50")
print("H1: Population mean != 50")

population_mean = 50
alpha = 0.05


# بخش ۳: اجرای One-Sample T-Test
print("\nبخش ۳: اجرای آزمون فرض")

t_statistic, p_value = stats.ttest_1samp(
    sales,
    popmean=population_mean
)

print("T-statistic:", t_statistic)
print("P-value:", p_value)


# بخش ۴: تصمیم‌گیری آماری
print("\nبخش ۴: تصمیم‌گیری آماری")

print("Significance Level (alpha):", alpha)

if p_value < alpha:
    print("نتیجه: فرض صفر (H0) رد می‌شود.")
    print("شواهد آماری کافی برای متفاوت بودن میانگین از 50 وجود دارد.")
else:
    print("نتیجه: فرض صفر (H0) رد نمی‌شود.")
    print("شواهد آماری کافی برای متفاوت بودن میانگین از 50 وجود ندارد.")


# بخش ۵: بررسی نتیجه با Confidence Interval
print("\nبخش ۵: بازه اطمینان ۹۵ درصد")

confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(sales) - 1,
    loc=sample_mean,
    scale=stats.sem(sales)
)

print("95% Confidence Interval:", confidence_interval)

if confidence_interval[0] <= population_mean <= confidence_interval[1]:
    print("عدد 50 داخل بازه اطمینان قرار دارد.")
else:
    print("عدد 50 خارج از بازه اطمینان قرار دارد.")


# بخش ۶: تفسیر نهایی
print("\nبخش ۶: تفسیر نهایی")

print(f"میانگین فروش نمونه برابر با {sample_mean:.2f} است.")
print(f"P-value برابر با {p_value:.4f} است.")

if p_value < alpha:
    print("نتیجه آزمون از تفاوت آماری معنادار با میانگین 50 حمایت می‌کند.")
else:
    print("نتیجه آزمون شواهد کافی برای تفاوت آماری معنادار ارائه نمی‌کند.")

print("معناداری آماری لزوماً به معنی اهمیت عملی تفاوت نیست.")
````
