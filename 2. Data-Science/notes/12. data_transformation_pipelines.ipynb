# Data Transformation Pipelines — پایپ‌لاین‌های تبدیل داده

## ۱. تعریف

**Data Transformation Pipeline** زنجیره‌ای از مراحل پردازش داده است که به ترتیب اجرا می‌شوند تا داده برای تحلیل یا آموزش مدل آماده شود.

## ۲. مراحل رایج

- **Imputation:** مدیریت مقادیر گمشده.
- **Encoding:** تبدیل داده‌های دسته‌ای یا متنی به نمایش عددی.
- **Scaling:** تغییر مقیاس ویژگی‌های عددی.
- **Feature Selection:** انتخاب ویژگی‌های مناسب، در صورت نیاز.

## ۳. ابزارهای مهم در Scikit-learn

- `SimpleImputer`: مدیریت مقادیر گمشده.
- `StandardScaler`: استانداردسازی ویژگی‌های عددی.
- `OneHotEncoder`: تبدیل ویژگی‌های دسته‌ای به ستون‌های عددی.
- `Pipeline`: اتصال مراحل پردازش و مدل‌سازی.
- `ColumnTransformer`: اجرای تبدیل‌های متفاوت روی گروه‌های مختلف ستون‌ها.

## ۴. متدهای مهم

- `fit()`: یادگیری پارامترها از داده.
- `transform()`: اعمال تبدیل‌های یادگرفته‌شده.
- `fit_transform()`: یادگیری پارامترها و اعمال تبدیل روی همان داده.

## ۵. نمونه ساده

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])

X_train_transformed = pipeline.fit_transform(X_train)
X_test_transformed = pipeline.transform(X_test)
```

در این مثال، ابتدا مقادیر گمشده با میانگین ستون پر می‌شوند و سپس ویژگی‌ها استانداردسازی می‌شوند.

## ۶. جلوگیری از Data Leakage

پارامترهای مراحل پردازش باید فقط از داده آموزش یاد گرفته شوند.

روش صحیح:

```python
pipeline.fit(X_train)

X_train_transformed = pipeline.transform(X_train)
X_test_transformed = pipeline.transform(X_test)
```

روش نادرست، یادگیری پارامترهای پردازش از کل دیتاست پیش از جداسازی داده‌هاست؛ زیرا اطلاعات داده آزمون ممکن است به فرایند آموزش نشت کند.

## ۷. نکات مهم

- ترتیب مراحل Pipeline اهمیت دارد.
- همه ستون‌ها لزوماً به یک نوع تبدیل نیاز ندارند.
- برای ستون‌های عددی و دسته‌ای می‌توان از `ColumnTransformer` استفاده کرد.
- Pipeline پردازش را منظم، قابل‌تکرار و کم‌خطاتر می‌کند.
- استفاده از Pipeline به‌تنهایی تضمین نمی‌کند Data Leakage رخ ندهد؛ نحوه تقسیم داده و مراحل انتخاب ویژگی نیز مهم‌اند.

# تمرین ۱ — ساخت یک Pipeline ساده

## صورت مسئله

دیتاست زیر را در نظر بگیر:

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "age": [20, 25, np.nan, 35, 40],
    "income": [2000, 3000, 4000, np.nan, 6000]
})

print(df)
```

## وظایف

۱. با استفاده از `SimpleImputer` مقادیر گمشده را با میانگین هر ستون پر کن.

۲. با استفاده از `StandardScaler` ویژگی‌ها را استانداردسازی کن.

۳. این دو مرحله را با `Pipeline` به یکدیگر متصل کن.

۴. نتیجه نهایی را چاپ کن.

۵. توضیح بده چرا بهتر است این مراحل را در یک Pipeline قرار دهیم.

**محدودیت:** از حلقه یا جایگزینی دستی مقادیر گمشده استفاده نکن؛ هدف تمرین کار با ابزارهای `scikit-learn` است.

</div>

````python
import pandas as pd
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "age": [20, 25, np.nan, 35, 40],
    "income": [2000, 3000, 4000, np.nan, 6000]
})

print("بخش 1: داده‌های اولیه")
print(df)

# ساخت Pipeline
pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])

# اجرای مراحل پردازش
X_transformed = pipeline.fit_transform(df)

print("\nبخش 2: داده‌های پردازش‌شده")
print(X_transformed)

# تبدیل خروجی به DataFrame برای خوانایی بهتر
df_transformed = pd.DataFrame(
    X_transformed,
    columns=df.columns
)

print("\nبخش 3: خروجی نهایی به‌صورت DataFrame")
print(df_transformed)

print("\nبخش 4: بررسی میانگین و انحراف معیار")
print("Means:")
print(df_transformed.mean().round(6))

print("\nStandard Deviations:")
print(df_transformed.std(ddof=0).round(6))
````

# تمرین ۲ — پردازش ستون‌های عددی و دسته‌ای

## صورت مسئله

این دیتاست اطلاعات چند مشتری را نگهداری می‌کند:

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "age": [22, 35, np.nan, 28, 40],
    "income": [2500, 5000, 3500, np.nan, 7000],
    "city": ["Tehran", "Shiraz", "Tehran", "Tabriz", "Shiraz"]
})

print(df)
```

## وظایف

۱. ستون‌های عددی (`age` و `income`) را با `SimpleImputer` و `StandardScaler` پردازش کن.

۲. ستون `city` را با `SimpleImputer` و `OneHotEncoder` پردازش کن. برای مقدار گمشده احتمالی شهر از پرتکرارترین مقدار استفاده کن.

۳. با `ColumnTransformer` این دو نوع پردازش را روی ستون‌های مربوطه اعمال کن.

۴. خروجی تبدیل‌شده را چاپ کن و ابعاد آن را با `shape` نمایش بده.

۵. توضیح بده چرا برای ستون `city` از `StandardScaler` استفاده نمی‌کنیم.

**راهنمای ابزارها:**

- `SimpleImputer`
- `StandardScaler`
- `OneHotEncoder`
- `ColumnTransformer`

</div>

````python
import pandas as pd
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# بخش 1: ایجاد دیتاست
df = pd.DataFrame({
    "age": [22, 35, np.nan, 28, 40],
    "income": [2500, 5000, 3500, np.nan, 7000],
    "city": ["Tehran", "Shiraz", "Tehran", "Tabriz", "Shiraz"]
})

print("بخش 1: داده‌های اولیه")
print(df)

# بخش 2: تعریف Pipeline برای ستون‌های عددی
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])

# بخش 3: تعریف Pipeline برای ستون دسته‌ای
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# بخش 4: ترکیب پردازش‌ها با ColumnTransformer
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, ["age", "income"]),
    ("categorical", categorical_pipeline, ["city"])
])

# بخش 5: اجرای Pipeline
X_transformed = preprocessor.fit_transform(df)

print("\nبخش 5: داده‌های پردازش‌شده")
print(X_transformed)

# بخش 6: نام ویژگی‌های خروجی
feature_names = preprocessor.get_feature_names_out()

print("\nبخش 6: نام ویژگی‌ها")
print(feature_names)

# بخش 7: تبدیل خروجی به DataFrame
df_transformed = pd.DataFrame(
    X_transformed,
    columns=feature_names
)

print("\nبخش 7: خروجی نهایی")
print(df_transformed)

# بخش 8: بررسی ابعاد
print("\nبخش 8: ابعاد داده")
print("Original shape:", df.shape)
print("Transformed shape:", X_transformed.shape)
````
