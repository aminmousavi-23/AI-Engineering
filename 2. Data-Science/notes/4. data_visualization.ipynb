# Data Visualization

## تعریف

**Data Visualization** یا بصری‌سازی داده، فرایند نمایش داده‌ها با استفاده از **Chart**ها و **Graph**ها است تا بتوانیم **Pattern**ها، **Trend**ها، **Distribution**ها، **Relationship**ها و تفاوت‌های موجود در داده را راحت‌تر درک کنیم.

**Visualization** یکی از بخش‌های مهم **Exploratory Data Analysis (EDA)** است.

---

## چرا از Data Visualization استفاده می‌کنیم؟

**Visualization** به ما کمک می‌کند موارد زیر را شناسایی کنیم:

* **Pattern**ها
* **Trend**ها
* **Distribution**ها
* **Relationship**ها
* تفاوت بین گروه‌های مختلف
* **Potential Outlier**ها
* **Anomaly**ها

---

## انواع رایج Chart

| Chart            | کاربرد اصلی                                       |
| ---------------- | ------------------------------------------------- |
| **Bar Chart**    | مقایسه‌ی دسته‌ها                                  |
| **Count Plot**   | نمایش تعداد نمونه‌های هر دسته                     |
| **Histogram**    | بررسی **Distribution** داده‌های عددی              |
| **Box Plot**     | بررسی **Distribution** و **Outlier**ها            |
| **Scatter Plot** | بررسی رابطه بین متغیرهای عددی                     |
| **Line Plot**    | بررسی روند داده‌های دارای ترتیب یا داده‌های زمانی |
| **Pie Chart**    | نمایش سهم هر بخش از یک کل                         |

---

## Matplotlib

**Matplotlib** یکی از کتابخانه‌های اصلی Python برای ایجاد **Visualization** است.

```python
import matplotlib.pyplot as plt
```

### Basic Plot

برای ایجاد یک نمودار ساده:

```python
plt.plot([1, 2, 3, 4], [10, 20, 15, 30])

plt.title("Example Plot")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

---

## Bar Chart

از **Bar Chart** برای مقایسه‌ی مقادیر بین دسته‌های مختلف استفاده می‌کنیم.

مثلاً می‌توانیم میزان فروش در شهرهای مختلف را مقایسه کنیم:

```python
import matplotlib.pyplot as plt

cities = ["Paris", "London", "Berlin"]
sales = [500, 700, 450]

plt.bar(cities, sales)

plt.title("Sales by City")
plt.xlabel("City")
plt.ylabel("Sales")

plt.show()
```

---

## Histogram

از **Histogram** برای بررسی **Distribution** یک متغیر عددی استفاده می‌کنیم.

```python
plt.hist(df["age"], bins=10)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

### bins

پارامتر `bins` مشخص می‌کند که محدوده‌ی مقادیر عددی به چند **Interval** تقسیم شود.

```python
bins=10
```

برای مثال:

```python
plt.hist(df["age"], bins=10)
```

یعنی مقادیر `age` در ۱۰ بازه نمایش داده شوند.

---

## Box Plot

**Box Plot** برای بررسی **Distribution** داده و شناسایی **Potential Outlier**ها بسیار کاربردی است.

در Box Plot مفاهیمی مانند موارد زیر قابل مشاهده هستند:

* **Median**
* **Quartiles**
* **IQR**
* **Potential Outlier**

```python
plt.boxplot(df["salary"])

plt.title("Salary Distribution")
plt.ylabel("Salary")

plt.show()
```

---

## Scatter Plot

از **Scatter Plot** برای بررسی **Relationship** بین دو متغیر عددی استفاده می‌کنیم.

مثلاً می‌توانیم رابطه‌ی بین `age` و `salary` را بررسی کنیم:

```python
plt.scatter(
    df["age"],
    df["salary"]
)

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```

اگر نقاط یک الگوی مشخص داشته باشند، ممکن است بین دو متغیر یک **Relationship** وجود داشته باشد.

---

## Line Plot

از **Line Plot** برای داده‌های دارای ترتیب استفاده می‌کنیم و یکی از مهم‌ترین کاربردهای آن نمایش **Time Series** است.

مثلاً برای نمایش درآمد ماهانه:

```python
plt.plot(
    df["month"],
    df["revenue"]
)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.show()
```

با **Line Plot** می‌توانیم **Trend** داده را بهتر مشاهده کنیم.

---

## Pandas Plotting

کتابخانه‌ی **Pandas** نیز یک رابط ساده برای ایجاد نمودارها در اختیار ما قرار می‌دهد.

### Histogram

```python
df["salary"].plot(kind="hist")
```

### Bar Chart

مثلاً میانگین حقوق هر شهر:

```python
df.groupby("city")["salary"].mean().plot(
    kind="bar"
)
```

### Scatter Plot

```python
df.plot(
    kind="scatter",
    x="age",
    y="salary"
)
```

### Line Plot

```python
df.plot(
    kind="line",
    x="month",
    y="revenue"
)
```

---

## Important Visualization Concepts

### Axis

**X-axis** و **Y-axis** ابعاد مختلف داده‌ای که در حال Visualization هستیم را نمایش می‌دهند.

```python
plt.xlabel("Age")
plt.ylabel("Salary")
```

---

### Title

**Title** مشخص می‌کند که نمودار چه چیزی را نمایش می‌دهد.

```python
plt.title("Age vs Salary")
```

استفاده از Title مناسب باعث می‌شود هدف نمودار سریع‌تر قابل فهم باشد.

---

### Legend

از **Legend** زمانی استفاده می‌کنیم که چند **Data Series** در یک نمودار داشته باشیم.

```python
plt.legend()
```

Legend مشخص می‌کند هر خط، نقطه یا بخش از نمودار مربوط به چه داده‌ای است.

---

### Grid

از **Grid** می‌توان برای خوانایی بهتر مقادیر روی نمودار استفاده کرد.

```python
plt.grid()
```

---

### Bins

در **Histogram**، پارامتر `bins` مشخص می‌کند داده‌ها به چند **Interval** تقسیم شوند.

```python
plt.hist(df["age"], bins=10)
```

---

## انتخاب Chart مناسب

انتخاب نوع نمودار باید بر اساس سؤالی باشد که می‌خواهیم با داده پاسخ دهیم.

```text
Category comparison
        ↓
    Bar Chart
```

برای مقایسه‌ی دسته‌های مختلف از **Bar Chart** استفاده می‌کنیم.

```text
Numeric distribution
        ↓
    Histogram / Box Plot
```

برای بررسی **Distribution** متغیرهای عددی از **Histogram** یا **Box Plot** استفاده می‌کنیم.

```text
Two numeric variables
        ↓
    Scatter Plot
```

برای بررسی **Relationship** بین دو متغیر عددی از **Scatter Plot** استفاده می‌کنیم.

```text
Time or ordered trend
        ↓
    Line Plot
```

برای بررسی **Trend** در داده‌های زمانی یا دارای ترتیب از **Line Plot** استفاده می‌کنیم.

```text
Part of a whole
        ↓
    Pie Chart
```

برای نمایش سهم بخش‌های مختلف از یک کل می‌توان از **Pie Chart** استفاده کرد.

---

## Data Visualization Workflow

یک Workflow ساده برای **Data Visualization**:

```text
Understand the Question
        ↓
Identify the Data Type
        ↓
Choose the Appropriate Chart
        ↓
Create the Visualization
        ↓
Add Labels and Title
        ↓
Interpret the Result
```

یعنی ابتدا سؤال را مشخص می‌کنیم، سپس نوع داده را بررسی می‌کنیم، **Chart** مناسب را انتخاب می‌کنیم، نمودار را ایجاد می‌کنیم، اطلاعات لازم مانند **Title** و **Labels** را اضافه می‌کنیم و در نهایت نمودار را **Interpret** می‌کنیم.

# تمرین ۱ — Data Visualization

از **Dataset** زیر استفاده کنید:

```python
import pandas as pd

df = pd.DataFrame({
    "age": [21, 25, 31, 25, 40, 52, 29, 35],
    "salary": [3000, 4200, 5500, 4200, 8000, 12000, 4500, 6500],
    "city": [
        "Paris",
        "Lyon",
        "Paris",
        "Lyon",
        "Lille",
        "Paris",
        "Lyon",
        "Lille"
    ]
})
```
## وظایف
### 1. Histogram

یک **Histogram** برای ستون `age` ایجاد کنید تا **Distribution** سن افراد را مشاهده کنید.

### 2. Box Plot

یک **Box Plot** برای ستون `salary` ایجاد کنید و **Potential Outlier**ها را بررسی کنید.

### 3. Bar Chart

یک **Bar Chart** ایجاد کنید که تعداد مشتریان هر `city` را نمایش دهد.

### 4. Scatter Plot

یک **Scatter Plot** ایجاد کنید تا رابطه‌ی بین `age` و `salary` را بررسی کنید.

### 5. Labels و Title

برای تمام نمودارها موارد زیر را به‌درستی مشخص کنید:

* **Title**
* **X-axis Label**
* **Y-axis Label**

### 6. تحلیل

بررسی کنید که آیا در ستون `salary` یک **Potential Outlier** وجود دارد یا خیر.

</div>


````python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "age": [21, 25, 31, 25, 40, 52, 29, 35],
    "salary": [3000, 4200, 5500, 4200, 8000, 12000, 4500, 6500],
    "city": [
        "Paris", "Lyon", "Paris", "Lyon",
        "Lille", "Paris", "Lyon", "Lille"
    ]
})

plt.hist(df["age"], bins=5)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
````

````python
plt.boxplot(df["salary"])

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Salary")

plt.show()
````

````python
city_counts = df["city"].value_counts()

plt.bar(city_counts.index, city_counts.values)

plt.title("Number of Customers by City")
plt.xlabel("City")
plt.ylabel("Number of Customers")

plt.show()
````

````python
plt.scatter(df["age"], df["salary"])

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
````

````python
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["salary"] < lower_bound) |
    (df["salary"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("\nPotential Outliers:")
print(outliers)
````

# تمرین ۲ — Data Visualization

از **Dataset** زیر استفاده کنید:

```python
import pandas as pd

df = pd.DataFrame({
    "month": [
        "Jan", "Feb", "Mar", "Apr",
        "May", "Jun", "Jul", "Aug"
    ],
    "revenue": [
        1200, 1500, 1400, 1800,
        2100, 2300, 2200, 2700
    ],
    "orders": [
        20, 25, 23, 30,
        35, 38, 36, 45
    ],
    "marketing_spend": [
        300, 350, 340, 400,
        450, 500, 480, 550
    ]
})
```

## وظایف

### 1. Revenue Line Plot

یک **Line Plot** برای `revenue` ایجاد کنید تا **Trend** درآمد ماهانه را مشاهده کنید.

### 2. Orders Line Plot

یک **Line Plot** برای `orders` ایجاد کنید تا روند تعداد سفارش‌ها در طول ماه‌ها مشخص شود.

### 3. Scatter Plot

یک **Scatter Plot** برای بررسی رابطه‌ی بین `marketing_spend` و `revenue` ایجاد کنید.

### 4. Bar Chart

یک **Bar Chart** برای نمایش `revenue` هر ماه ایجاد کنید.

### 5. Labels و Title

برای تمام نمودارها موارد زیر را مشخص کنید:

* **Title**
* **X-axis Label**
* **Y-axis Label**

### 6. تحلیل Visualization

بر اساس نمودارهایی که ایجاد کرده‌اید، به سؤالات زیر پاسخ دهید:

* روند کلی **Revenue** چگونه است؟
* آیا بین **Marketing Spend** و **Revenue** رابطه‌ای مشاهده می‌شود؟
* کدام ماه بیشترین **Revenue** را دارد؟

</div>


````python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "month": [
        "Jan", "Feb", "Mar", "Apr",
        "May", "Jun", "Jul", "Aug"
    ],
    "revenue": [
        1200, 1500, 1400, 1800,
        2100, 2300, 2200, 2700
    ],
    "orders": [
        20, 25, 23, 30,
        35, 38, 36, 45
    ],
    "marketing_spend": [
        300, 350, 340, 400,
        450, 500, 480, 550
    ]
})

plt.plot(df["month"], df["revenue"], marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.show()
````

````python
plt.plot(df["month"], df["orders"], marker="o")

plt.title("Monthly Orders Trend")
plt.xlabel("Month")
plt.ylabel("Orders")

plt.show()
````

````python
plt.scatter(df["marketing_spend"], df["revenue"])

plt.title("Marketing Spend vs Revenue")
plt.xlabel("Marketing Spend")
plt.ylabel("Revenue")

plt.show()
````

````python
plt.bar(df["month"], df["revenue"])

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.show()
````
