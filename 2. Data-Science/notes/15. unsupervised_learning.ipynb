# Unsupervised Learning Fundamentals

## تعریف
**Unsupervised Learning** نوعی یادگیری ماشین است که در آن داده‌ها برچسب هدف مشخصی ندارند و مدل سعی می‌کند ساختارها و الگوهای داده را کشف کند.

## تفاوت با Supervised Learning
- **Supervised Learning:** داده‌ها همراه با برچسب `y` هستند.
- **Unsupervised Learning:** معمولاً فقط داده‌های `X` در اختیار مدل قرار می‌گیرند.

## کاربردهای اصلی
- **Clustering:** گروه‌بندی نمونه‌های مشابه
- **Dimensionality Reduction:** کاهش تعداد ابعاد داده
- **Anomaly Detection:** شناسایی نمونه‌های غیرعادی

## K-Means
الگوریتمی برای تقسیم داده‌ها به `K` خوشه بر اساس نزدیکی نمونه‌ها به مراکز خوشه‌ها.

پارامترهای مهم:
- `n_clusters`: تعداد خوشه‌ها
- `random_state`: کنترل تصادفی‌بودن برای تکرارپذیری
- `n_init`: تعداد اجراهای اولیه با مراکز متفاوت

ویژگی‌های مهم:
- `labels_`: خوشهٔ اختصاص‌یافته به هر نمونه
- `cluster_centers_`: مراکز خوشه‌ها

## نکات
- شمارهٔ خوشه‌ها به‌خودی‌خود معنای خاصی ندارد.
- انتخاب تعداد خوشه‌ها اهمیت دارد.
- K-Means به مقیاس ویژگی‌ها حساس است؛ در بسیاری از کاربردها باید ابتدا از `StandardScaler` استفاده کرد.
- خوشه‌های پیدا‌شده لزوماً معادل گروه‌های واقعی و از پیش تعریف‌شده نیستند.

# تمرین ۱ — اجرای K-Means

با استفاده از `make_blobs` و `KMeans`:

1. دیتاستی با ۲۰۰ نمونه و ۴ مرکز تولید کن.
2. الگوریتم K-Means را با `n_clusters=4` اجرا کن.
3. شکل دادهٔ ورودی را چاپ کن.
4. تعداد نمونه‌های اختصاص‌یافته به هر خوشه را با `np.unique(..., return_counts=True)` نمایش بده.
5. مراکز خوشه‌ها را چاپ کن.

**هدف:** درک نحوهٔ اجرای Clustering و بررسی خروجی الگوریتم.

</div>

```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# 1. Generate dataset
X, _ = make_blobs(
    n_samples=200,
    centers=4,
    random_state=42
)

print("Dataset Information:")
print("Data shape:", X.shape)

# 2. Create and train K-Means
model = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

model.fit(X)

# 3. Get cluster labels
labels = model.labels_

print("\nCluster Distribution:")

cluster_ids, counts = np.unique(
    labels,
    return_counts=True
)

for cluster_id, count in zip(cluster_ids, counts):
    print(f"Cluster {cluster_id}: {count} samples")

# 4. Display cluster centers
print("\nCluster Centers:")
print(model.cluster_centers_)
```
