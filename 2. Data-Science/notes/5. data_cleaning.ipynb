# Data Cleaning

## تعریف

**Data Cleaning** فرایند شناسایی و اصلاح مشکلات کیفیت داده‌ها است تا Dataset برای تحلیل و مراحل بعدی Data Science آماده شود.

## مشکلات رایج داده‌ها

| مشکل | توضیح |
|---|---|
| **Missing Values** | مقادیر گمشده |
| **Duplicate Rows** | ردیف‌های تکراری |
| **Invalid Values** | مقادیر نامعتبر |
| **Inconsistent Data** | داده‌های ناسازگار |
| **Incorrect Data Types** | نوع داده‌ی نامناسب |
| **Outliers** | مقادیر بسیار دور از سایر داده‌ها که باید بررسی شوند |

## 1. بررسی Missing Values

شمارش مقادیر گمشده در هر ستون:

```python
df.isna().sum()
```

حذف ردیف‌هایی که حداقل یک مقدار گمشده دارند:

```python
df.dropna()
```

جایگزینی مقادیر گمشده:

```python
df["age"] = df["age"].fillna(
    df["age"].median()
)
```

انتخاب روش جایگزینی به نوع داده، میزان Missing Values و هدف تحلیل بستگی دارد.

## 2. مدیریت Duplicate Rows

شمارش ردیف‌های تکراری:

```python
df.duplicated().sum()
```

نمایش ردیف‌های تکراری:

```python
df[df.duplicated()]
```

حذف ردیف‌های تکراری:

```python
df = df.drop_duplicates()
```

برای بررسی تکرار بر اساس یک ستون مشخص:

```python
df.duplicated(subset=["customer_id"])
```

## 3. اصلاح داده‌های متنی

یکسان‌سازی فاصله‌ها و حروف:

```python
df["city"] = (
    df["city"]
    .str.strip()
    .str.title()
)
```

برای مثال، مقادیر `" paris "` و `"PARIS"` به `"Paris"` تبدیل می‌شوند.

## 4. مدیریت Invalid Values

فرض کنید حقوق منفی در Dataset معتبر نیست:

```python
import numpy as np

df.loc[
    df["salary"] < 0,
    "salary"
] = np.nan
```

در این مثال، مقادیر منفی ابتدا به Missing Values تبدیل می‌شوند تا بعداً با روش مناسب مدیریت شوند.

## 5. اصلاح Data Types

بررسی نوع داده‌ها:

```python
df.dtypes
```

تبدیل یک ستون به نوع عددی:

```python
df["age"] = pd.to_numeric(
    df["age"],
    errors="coerce"
)
```

با `errors="coerce"`، مقادیر غیرقابل‌تبدیل به `NaN` تبدیل می‌شوند.

## 6. بررسی نهایی

پس از پاک‌سازی، دوباره کیفیت داده‌ها را بررسی می‌کنیم:

```python
df.info()

df.isna().sum()

df.duplicated().sum()

df.describe()
```

## Data Cleaning Workflow

```text
Inspect the Dataset
        ↓
Identify Data Quality Issues
        ↓
Choose a Cleaning Strategy
        ↓
Apply the Changes
        ↓
Validate the Results
```

## نکات مهم

- قبل از حذف داده‌ها، علت مشکل را بررسی کنید.
- هر Outlier الزاماً اشتباه نیست.
- جایگزینی Missing Values باید با توجه به ماهیت داده انجام شود.
- حذف Duplicate Rows باید با توجه به معنای ردیف‌ها انجام شود.
- در پروژه‌های Machine Learning، اطلاعات مجموعه‌ی Test نباید برای یادگیری روش‌های جایگزینی یا سایر مراحل آماده‌سازی استفاده شود.

## اصطلاحات مهم

- **Data Cleaning:** پاک‌سازی داده
- **Data Quality:** کیفیت داده

- **Invalid Value:** مقدار نامعتبر
- **Inconsistent Data:** داده‌ی ناسازگار
- **Data Type:** نوع داده
- **Outlier:** مقدار پرت
- **Imputation:** جایگزینی مقادیر گمشده

## وظایف

### بخش اول — Data Inspection

1. تعداد ردیف‌ها و ستون‌ها را بررسی کنید.
2. نوع داده‌ی هر ستون را نمایش دهید.
3. تعداد **Missing Values** هر ستون را محاسبه کنید.
4. ردیف‌های تکراری کامل را پیدا کنید.

### بخش دوم — Data Cleaning

5. فاصله‌های اضافی ابتدا و انتهای مقادیر `city` را حذف کنید و حروف را یکسان‌سازی کنید.
6. مقادیر ستون `membership` را یکسان‌سازی کنید؛ مثلاً `basic` و `Basic` باید یکسان شوند.
7. مقادیر نامعتبر `age`، یعنی سن کمتر از صفر یا بیشتر از 120 را به `NaN` تبدیل کنید.
8. مقادیر منفی `total_spent` را به `NaN` تبدیل کنید.
9. مقدار گمشده‌ی `age` را با **Median** ستون جایگزین کنید.
10. مقدار گمشده‌ی `total_spent` را با **Median** همان ستون جایگزین کنید.
11. ردیف‌های تکراری کامل را حذف کنید.

### بخش سوم — Data Validation

12. بررسی کنید آیا هنوز **Missing Values** وجود دارد.
13. بررسی کنید آیا هنوز ردیف تکراری وجود دارد.
14. مقادیر یکتای `city` و `membership` را نمایش دهید.
15. با `describe()` آمار توصیفی ستون‌های عددی را بررسی کنید.

### بخش چهارم — تحلیل نهایی

16. میانگین `total_spent` را برای هر نوع `membership` محاسبه کنید.
17. تعداد مشتریان هر شهر را به دست آورید.
18. توضیح دهید چرا نباید صرفاً به دلیل تکراری بودن `customer_id`، همه‌ی ردیف‌های دارای شناسه‌ی تکراری را حذف کنیم.

**نکته:** ترتیب مراحل مهم است. ابتدا داده‌ها را بررسی کنید، سپس پاک‌سازی را انجام دهید و در پایان نتیجه را اعتبارسنجی کنید.

````python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 104, 106, 107, 108, 109, 110],
    "age": [25, 32, None, 45, 45, -3, 29, 150, 38, None],
    "city": [
        " Paris ", "LONDON", "paris", "Berlin", "Berlin",
        "London", " PARIS", "Berlin", None, "london"
    ],
    "membership": [
        "Basic", "Premium", "basic", "Gold", "Gold",
        "Premium", None, "GOLD", "Basic", "Premium"
    ],
    "total_spent": [250, 1200, 450, 3200, 3200, -100, 800, 5000, None, 1500],
    "orders": [2, 8, 3, 15, 15, 1, 5, 20, 4, 10]
})

# ==================================================
print("\nبخش ۱: بررسی اولیه داده‌ها (Data Inspection)")
# ==================================================

print("Shape:")
print(df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isna().sum())

print("\nDuplicate Rows:")
print(df[df.duplicated()])

print("\nDuplicate Customer IDs:")
print(df[df["customer_id"].duplicated(keep=False)].sort_values("customer_id"))


# ==================================================
print("\nبخش ۲: پاک‌سازی مقادیر متنی (Text Cleaning)")
# ==================================================

df["city"] = df["city"].str.strip().str.title()
df["membership"] = df["membership"].str.strip().str.title()

print(df[["city", "membership"]])


# ==================================================
print("\nبخش ۳: اصلاح مقادیر نامعتبر (Invalid Values)")
# ==================================================

# سن کمتر از صفر یا بیشتر از ۱۲۰ نامعتبر فرض می‌شود
df.loc[(df["age"] < 0) | (df["age"] > 120), "age"] = np.nan

# هزینه منفی نامعتبر فرض می‌شود
df.loc[df["total_spent"] < 0, "total_spent"] = np.nan

print(df[["customer_id", "age", "total_spent"]])


# ==================================================
print("\nبخش ۴: مدیریت مقادیر گمشده (Missing Values)")
# ==================================================

# جایگزینی مقادیر گمشده با میانه (Median)
df["age"] = df["age"].fillna(df["age"].median())
df["total_spent"] = df["total_spent"].fillna(
    df["total_spent"].median()
)

print("Missing Values After Imputation:")
print(df.isna().sum())


# ==================================================
print("\nبخش ۵: حذف ردیف‌های تکراری (Duplicate Removal)")
# ==================================================

df = df.drop_duplicates()

print("Shape After Removing Duplicates:")
print(df.shape)

print("\nDuplicate Rows After Cleaning:")
print(df.duplicated().sum())


# ==================================================
print("\nبخش ۶: اعتبارسنجی نهایی (Final Validation)")
# ==================================================

print("DataFrame Info:")
df.info()

print("\nMissing Values:")
print(df.isna().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nUnique Cities:")
print(df["city"].unique())

print("\nUnique Memberships:")
print(df["membership"].unique())


# ==================================================
print("\nبخش ۷: آمار توصیفی (Descriptive Statistics)")
# ==================================================

print(df.describe())


# ==================================================
print("\nبخش ۸: میانگین هزینه بر اساس نوع عضویت")
# ==================================================

membership_spending = df.groupby("membership")["total_spent"].mean()

print(membership_spending)


# ==================================================
print("\nبخش ۹: تعداد مشتریان بر اساس شهر")
# ==================================================

customers_per_city = df.groupby("city")["customer_id"].count()

print(customers_per_city)


# ==================================================
print("\nبخش ۱۰: نمایش دیتاست نهایی (Cleaned Dataset)")
# ==================================================

print(df.to_string(index=False))
````
