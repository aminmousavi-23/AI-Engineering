# Data Understanding

## Definition

Data Understanding is the process of inspecting and understanding a dataset before performing deeper analysis or building machine learning models.

The goal is to understand:

* Dataset structure
* Variables and their meanings
* Data types
* Dataset size
* Missing values
* Duplicate records
* Unique values
* Basic statistics
* Feature and target variables
* Potential data quality issues

---

## Dataset Shape

Use `shape` to get the number of rows and columns:

```python
df.shape
```

Example:

```text
(1000, 8)
```

This means:

* 1000 rows / observations
* 8 columns / variables

---

## Dataset Columns

Use `columns` to inspect column names:

```python
df.columns
```

---

## Data Types

Use `dtypes` to inspect the data type of each column:

```python
df.dtypes
```

Common Pandas data types include:

* `int64`
* `float64`
* `object`
* `bool`
* `datetime64`
* `category`

---

## DataFrame Information

`info()` provides a high-level overview of a DataFrame:

```python
df.info()
```

It can show:

* Number of rows
* Number of columns
* Column names
* Non-null counts
* Data types
* Memory usage

---

## Missing Values

Missing values can be checked with:

```python
df.isna().sum()
```

This returns the number of missing values in each column.

---

## Duplicate Rows

Check for duplicate rows:

```python
df.duplicated().sum()
```

Select duplicate rows:

```python
df[df.duplicated()]
```

---

## Descriptive Overview

Use `describe()` to get basic descriptive statistics for numerical columns:

```python
df.describe()
```

It can provide values such as:

* count
* mean
* standard deviation
* minimum
* quartiles
* maximum

---

## Categorical Variables

For categorical columns:

```python
df["city"].describe()
```

Common outputs include:

* `count`
* `unique`
* `top`
* `freq`

---

## Unique Values

Get unique values:

```python
df["city"].unique()
```

Get the number of unique values:

```python
df["city"].nunique()
```

---

## Value Counts

Count the occurrences of each value:

```python
df["city"].value_counts()
```

This is especially useful for understanding categorical variables.

---

## Inspecting Rows

First rows:

```python
df.head()
```

Last rows:

```python
df.tail()
```

Random rows:

```python
df.sample(5)
```

---

## Memory Usage

Inspect memory usage:

```python
df.memory_usage(deep=True)
```

Total memory usage:

```python
df.memory_usage(deep=True).sum()
```

# Data Understanding — Exercise 1

## Dataset Inspection

Consider the following dataset:

```python
import pandas as pd

df = pd.DataFrame({
    "age": [21, 25, 31, 25, 40, None],
    "salary": [3000, 4200, 5500, 4200, 8000, 7000],
    "city": ["Paris", "Lyon", "Paris", "Lyon", "Lille", "Paris"],
    "orders": [2, 5, 8, 5, 12, 7]
})
```

### Tasks

Write Python/Pandas code to:

1. Display the shape of the dataset.
2. Display all column names.
3. Display the data types.
4. Display the DataFrame information.
5. Count missing values in each column.
6. Count duplicate rows.
7. Display the first 3 rows.
8. Display 3 random rows.
9. Display the unique values of `city`.
10. Count the occurrences of each city.


````python
import pandas as pd

df = pd.DataFrame({
    "age": [21, 25, 31, 25, 40, None],
    "salary": [3000, 4200, 5500, 4200, 8000, 7000],
    "city": ["Paris", "Lyon", "Paris", "Lyon", "Lille", "Paris"],
    "orders": [2, 5, 8, 5, 12, 7]
})


# 1. Display the shape of the dataset
print("Shape:")
print(df.shape)


# 2. Display all column names
print("\nColumns:")
print(df.columns.tolist())


# 3. Display the data types
print("\nData Types:")
print(df.dtypes)


# 4. Display the DataFrame information
print("\nDataFrame Information:")
df.info()


# 5. Count missing values in each column
print("\nMissing Values:")
print(df.isna().sum())


# 6. Count duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())


# 7. Display the first 3 rows
print("\nFirst 3 Rows:")
print(df.head(3))


# 8. Display 3 random rows
print("\n3 Random Rows:")
print(df.sample(3))


# 9. Display the unique values of city
print("\nUnique Cities:")
print(df["city"].unique())


# 10. Count the occurrences of each city
print("\nCity Value Counts:")
print(df["city"].value_counts())
````

# Data Understanding — Exercise 2

## E-Commerce Dataset

Consider the following dataset:

```python
import pandas as pd

df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106, 107],
    "age": [22, 35, None, 41, 29, 35, 50],
    "membership": [
        "Basic",
        "Premium",
        "Basic",
        "Premium",
        "Basic",
        "Premium",
        "Basic"
    ],
    "orders": [2, 8, 3, 12, 5, 8, 15],
    "total_spent": [120, 850, 250, 1400, 450, 850, 1700]
})
```

### Tasks

Write Python/Pandas code to:

1. Determine the number of observations and columns.
2. Inspect the dataset structure using `info()`.
3. Identify missing values.
4. Check for duplicate rows.
5. Find the number of unique membership types.
6. Display the frequency of each membership type.
7. Calculate basic descriptive statistics for numerical columns.
8. Display 5 random observations.
9. Identify which columns are numerical.
10. Identify which columns are categorical or identifiers.


````python
import pandas as pd

df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106, 107],
    "age": [22, 35, None, 41, 29, 35, 50],
    "membership": [
        "Basic",
        "Premium",
        "Basic",
        "Premium",
        "Basic",
        "Premium",
        "Basic"
    ],
    "orders": [2, 8, 3, 12, 5, 8, 15],
    "total_spent": [120, 850, 250, 1400, 450, 850, 1700]
})

# 1. Number of observations and columns
print("Shape:", df.shape)
print("Number of observations:", df.shape[0])
print("Number of columns:", df.shape[1])

# 2. Display DataFrame information
print("\nDataFrame Information:")
df.info()

# 3. Count missing values
print("\nMissing Values:")
print(df.isna().sum())

# 4. Count duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# 5. Count unique membership values
print("\nUnique Membership Count:")
print(df["membership"].nunique())

# 6. Count occurrences of each membership type
print("\nMembership Value Counts:")
print(df["membership"].value_counts())

# 7. Display basic descriptive statistics
print("\nDescriptive Statistics:")
print(df.describe())

# 8. Display 5 random observations
print("\n5 Random Observations:")
print(df.sample(5))

# 9. Identify numerical columns
numerical_columns = df.select_dtypes(include="number").columns.tolist()

print("\nNumerical Columns:")
print(numerical_columns)

# 10. Identify categorical and identifier columns
categorical_columns = ["membership"]
identifier_columns = ["customer_id"]

print("\nCategorical Columns:")
print(categorical_columns)

print("\nIdentifier Columns:")
print(identifier_columns)
````
