# Descriptive Statistics

## Definition

**Descriptive Statistics** is the process of summarizing and describing the main characteristics of a dataset using statistical measures.

It helps us understand:

* Central tendency
* Data spread
* Position of values
* Distribution shape
* Potential outliers

---

# 1. Mean

**Mean** is the arithmetic average of a variable.

$$
Mean = \frac{\sum x_i}{n}
$$

Example:

```python
import pandas as pd

df = pd.DataFrame({
    "salary": [3000, 4000, 5000, 6000, 7000]
})

mean_salary = df["salary"].mean()

print(mean_salary)
```

Output:

```text
5000.0
```

> Mean is sensitive to outliers.

---

# 2. Median

**Median** is the middle value after sorting the data.

Example:

```python
df = pd.DataFrame({
    "salary": [3000, 4000, 5000, 6000, 7000]
})

median_salary = df["salary"].median()

print(median_salary)
```

Output:

```text
5000.0
```

For an even number of observations, the median is the average of the two middle values.

> Median is more resistant to outliers than Mean.

---

# 3. Mode

**Mode** is the most frequently occurring value.

Example:

```python
df = pd.DataFrame({
    "city": ["Paris", "Lyon", "Paris", "Lille", "Paris"]
})

mode_city = df["city"].mode()

print(mode_city)
```

Output:

```text
0    Paris
```

Mode can be used for both **numerical** and **categorical** variables.

---

# 4. Range

**Range** measures the difference between the maximum and minimum values.

$$
Range = Maximum - Minimum
$$

Example:

```python
df = pd.DataFrame({
    "age": [20, 25, 30, 35, 50]
})

data_range = df["age"].max() - df["age"].min()

print(data_range)
```

Output:

```text
30
```

---

# 5. Variance

**Variance** measures how far values are spread from the Mean.

A higher Variance generally indicates greater dispersion.

```python
variance = df["age"].var()

print(variance)
```

Variance is expressed in squared units.

---

# 6. Standard Deviation

**Standard Deviation (Std)** is the square root of Variance.

$$
Std = \sqrt{Variance}
$$

It measures the typical spread of values around the Mean.

Example:

```python
std_age = df["age"].std()

print(std_age)
```

Unlike Variance, Standard Deviation uses the same unit as the original variable.

---

# 7. Quartiles

Quartiles divide ordered data into four parts.

| Quartile | Percentile |
| -------- | ---------: |
| Q1       |       25th |
| Q2       |       50th |
| Q3       |       75th |

Q2 is the same as the **Median**.

Example:

```python
q1 = df["age"].quantile(0.25)
q2 = df["age"].quantile(0.50)
q3 = df["age"].quantile(0.75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

---

# 8. Percentile

A **Percentile** indicates the value below which a certain percentage of observations fall.

Example:

```python
p90 = df["age"].quantile(0.90)

print(p90)
```

This represents the **90th Percentile**.

Important relationships:

```text
25th Percentile = Q1
50th Percentile = Q2 = Median
75th Percentile = Q3
```

---

# 9. Interquartile Range (IQR)

**IQR** measures the spread of the middle 50% of the data.

$$
IQR = Q3 - Q1
$$

Example:

```python
q1 = df["age"].quantile(0.25)
q3 = df["age"].quantile(0.75)

iqr = q3 - q1

print(iqr)
```

IQR is also commonly used for **Outlier Detection**.

---

# 10. Five-Number Summary

The **Five-Number Summary** consists of:

1. Minimum
2. Q1
3. Median
4. Q3
5. Maximum

Example:

```python
five_number_summary = {
    "min": df["age"].min(),
    "q1": df["age"].quantile(0.25),
    "median": df["age"].median(),
    "q3": df["age"].quantile(0.75),
    "max": df["age"].max()
}

print(five_number_summary)
```

---

# 11. Skewness

**Skewness** measures the asymmetry of a distribution.

### Symmetric Distribution

Mean and Median are usually close:

```text
Mean ≈ Median
```

### Right-Skewed Distribution

The distribution has a longer tail on the right:

```text
Mean > Median
```

### Left-Skewed Distribution

The distribution has a longer tail on the left:

```text
Mean < Median
```

Example:

```python
skewness = df["age"].skew()

print(skewness)
```

---

# 12. `describe()`

Pandas provides `describe()` to calculate several descriptive statistics at once.

```python
df["age"].describe()
```

Typical output:

```text
count     5.000000
mean     32.000000
std      11.726039
min      20.000000
25%      25.000000
50%      30.000000
75%      35.000000
max      50.000000
```

For all numerical columns:

```python
df.describe()
```

# 13. Summary of Important Statistics

| Statistic          | Purpose                         |
| ------------------ | ------------------------------- |
| Mean               | Average value                   |
| Median             | Middle value                    |
| Mode               | Most frequent value             |
| Minimum            | Smallest value                  |
| Maximum            | Largest value                   |
| Range              | Maximum − Minimum               |
| Variance           | Measures dispersion             |
| Standard Deviation | Measures typical spread         |
| Q1                 | 25th percentile                 |
| Q2                 | 50th percentile / Median        |
| Q3                 | 75th percentile                 |
| IQR                | Q3 − Q1                         |
| Skewness           | Measures distribution asymmetry |

---

# Key Idea

Descriptive Statistics helps us summarize a dataset using a small number of meaningful statistics.

The most important concepts to remember are:

```text
Central Tendency
→ Mean, Median, Mode

Spread
→ Range, Variance, Standard Deviation, IQR

Position
→ Percentiles, Quartiles

Distribution Shape
→ Skewness
```

These statistics are commonly used during **EDA** to understand the data before statistical modeling or Machine Learning.
