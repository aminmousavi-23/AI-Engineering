# Exploratory Data Analysis (EDA)

## Definition

**Exploratory Data Analysis (EDA)** is the process of exploring, analyzing, and understanding a dataset to discover patterns, relationships, distributions, anomalies, and potential data quality issues.

The main goal of EDA is to understand **what is happening inside the data** before applying statistical analysis or building Machine Learning models.

---

## Main Questions in EDA

EDA helps answer questions such as:

* How are the values distributed?
* What are the typical values?
* Are there outliers?
* Are there unusual or suspicious values?
* Are variables related to each other?
* Do different groups behave differently?
* Are there patterns or trends?
* Are there potential data quality problems?

---

## EDA vs Data Understanding

### Data Understanding

Focuses on understanding the structure and meaning of the dataset.

Examples:

* Dataset shape
* Columns
* Data types
* Missing values
* Duplicate rows
* Unique values
* Dataset schema

### EDA

Focuses on discovering patterns and relationships within the data.

Examples:

* Distributions
* Relationships
* Outliers
* Group comparisons
* Correlations
* Patterns and trends

In simple terms:

> **Data Understanding = What is the data?**

> **EDA = What is happening in the data?**

---

# Types of EDA

## 1. Univariate Analysis

**Univariate Analysis** means analyzing one variable at a time.

Example:

```python
df["age"].describe()
```

For categorical variables:

```python
df["membership"].value_counts()
```

Typical questions:

* What are the minimum and maximum values?
* What is the mean or median?
* How many unique values exist?
* How is the variable distributed?

---

## 2. Bivariate Analysis

**Bivariate Analysis** studies the relationship between two variables.

Example:

```python
df[["age", "total_spent"]].corr()
```

Or comparing groups:

```python
df.groupby("membership")["total_spent"].mean()
```

Typical questions:

* Are two variables related?
* Does one variable change when another changes?
* Do different groups have different values?

---

## 3. Multivariate Analysis

**Multivariate Analysis** examines multiple variables simultaneously.

Example:

```python
df.corr(numeric_only=True)
```

This can help identify relationships between multiple numerical variables.

---

# Distribution

A **Distribution** describes how values of a variable are spread across the dataset.

Important characteristics include:

* Center
* Spread
* Shape
* Skewness
* Outliers

Example:

```python
df["age"].describe()
```

---

# Outliers

An **Outlier** is a value that is unusually far from the general pattern of the data.

Example:

```text
[20, 21, 22, 23, 24, 25, 150]
```

Here, `150` may be an outlier for an age variable.

Important:

> An outlier is not necessarily an error.

It may represent:

* Data entry error
* Measurement error
* System error
* A rare but valid observation

EDA identifies potential outliers. Decisions about removing or modifying them are generally part of **Data Cleaning**.

---

# Group Analysis

**Group Analysis** compares different groups within a dataset.

Example:

```python
df.groupby("membership")["total_spent"].mean()
```

This can help identify differences between groups.

Example groups:

* Basic
* Premium
* Enterprise

# Visualization in EDA

Visualization is an important part of EDA.

Common plots include:

* Histogram → Distribution
* Box Plot → Outliers and spread
* Scatter Plot → Relationship between numerical variables
* Bar Plot → Category comparison
* Line Plot → Trends over time
* Heatmap → Correlation and relationships

Example:

```python
df["age"].hist()
```

```python
df.boxplot(column="total_spent")
```

```python
df.plot.scatter(
    x="age",
    y="total_spent"
)
```

Visualization will be studied in more detail in the **Data Visualization** topic.

# Typical EDA Workflow

```text
Dataset
   ↓
Inspect Structure
   ↓
Analyze Variables
   ↓
Analyze Distributions
   ↓
Check Missing Values
   ↓
Check Duplicates
   ↓
Detect Outliers
   ↓
Analyze Relationships
   ↓
Compare Groups
   ↓
Generate Insights
```

EDA is usually an **iterative process**. Discovering something interesting may lead to additional analysis.
