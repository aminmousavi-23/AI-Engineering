# Data Preprocessing

## 1. تعریف

**Data Preprocessing** فرایند آماده‌سازی داده‌ها برای استفاده در الگوریتم‌های **Machine Learning** است.

هدف آن تبدیل داده‌های خام به داده‌هایی با ساختار و مقیاس مناسب برای مدل است.

## 2. مراحل اصلی

| مرحله | توضیح |
|---|---|
| Missing Value Imputation | جایگزینی مقادیر گمشده |
| Encoding | تبدیل ویژگی‌های دسته‌ای به مقادیر عددی |
| Feature Scaling | هم‌مقیاس کردن ویژگی‌های عددی |
| Data Transformation | تبدیل مقادیر یا ساختار داده |
| Train-Test Separation | جداسازی داده‌های آموزش و ارزیابی |

## 3. Encoding

### One-Hot Encoding

برای ویژگی‌های دسته‌ای بدون ترتیب مشخص استفاده می‌شود.

```python
import pandas as pd

df = pd.DataFrame({
    "city": ["Tehran", "Shiraz", "Tehran", "Tabriz"]
})

encoded_df = pd.get_dummies(
    df,
    columns=["city"],
    dtype=int
)

print(encoded_df)
```

### Ordinal Encoding

برای ویژگی‌های دارای ترتیب طبیعی استفاده می‌شود.

```python
df = pd.DataFrame({
    "level": ["Low", "Medium", "High"]
})

mapping = {
    "Low": 1,
    "Medium": 2,
    "High": 3
}

df["level_encoded"] = df["level"].map(mapping)

print(df)
```

## 4. Feature Scaling

### StandardScaler

ویژگی‌ها را با استفاده از میانگین و انحراف معیار داده‌های آموزش مقیاس‌بندی می‌کند.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

### MinMaxScaler

مقادیر را معمولاً به بازه صفر تا یک منتقل می‌کند.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## 5. Data Leakage

**Data Leakage** زمانی رخ می‌دهد که اطلاعاتی که نباید در فرایند آموزش استفاده شوند، به‌طور ناخواسته وارد آن شوند.

نکته مهم: پارامترهای تبدیل باید فقط از داده‌های `Train` یاد گرفته شوند. سپس همان پارامترها روی `Test` اعمال می‌شوند.

- `fit`: یادگیری پارامترها از داده‌ها
- `transform`: اعمال تبدیل
- `fit_transform`: یادگیری پارامترها و اعمال تبدیل

## 6. تفاوت Data Cleaning و Data Preprocessing

- **Data Cleaning:** اصلاح داده‌های نامعتبر، تکراری و ناسازگار.
- **Data Preprocessing:** آماده‌سازی داده برای الگوریتم‌های یادگیری ماشین، مانند Encoding و Scaling.

این دو مرحله می‌توانند هم‌پوشانی داشته باشند.

## 7. اصطلاحات مهم

- **Categorical Feature:** ویژگی دسته‌ای
- **Numerical Feature:** ویژگی عددی
- **Encoding:** تبدیل داده‌های دسته‌ای به نمایش عددی
- **Scaling:** تغییر مقیاس ویژگی‌ها
- **Imputation:** جایگزینی مقادیر گمشده
- **Data Leakage:** نشت اطلاعات
- **Scaler:** ابزاری برای مقیاس‌بندی ویژگی‌ها

# تمرین ۱ — Categorical Encoding

دیتاست زیر را در نظر بگیر:

```python
import pandas as pd

df = pd.DataFrame({
    "age": [22, 30, 25, 40, 35, 28],
    "city": [
        "Tehran", "Shiraz", "Tabriz",
        "Tehran", "Shiraz", "Tabriz"
    ],
    "membership": [
        "Basic", "Premium", "Standard",
        "Premium", "Basic", "Standard"
    ],
    "satisfaction": [
        "Low", "High", "Medium",
        "High", "Low", "Medium"
    ]
})
```

## وظایف

۱. دیتاست را چاپ کن.

۲. با استفاده از `pd.get_dummies()` ویژگی `city` را به روش **One-Hot Encoding** تبدیل کن.

۳. ویژگی `membership` را نیز با One-Hot Encoding تبدیل کن؛ چون در این تمرین ترتیب مشخصی برای آن تعریف نشده است.

۴. ویژگی `satisfaction` را با **Ordinal Encoding** و نگاشت زیر تبدیل کن:

- `Low = 1`
- `Medium = 2`
- `High = 3`

۵. دیتاست نهایی را چاپ کن و توضیح بده کدام ستون‌ها حذف یا اضافه شده‌اند.

</div>

````python
import pandas as pd

df = pd.DataFrame({
    "age": [22, 30, 25, 40, 35, 28],
    "city": [
        "Tehran", "Shiraz", "Tabriz",
        "Tehran", "Shiraz", "Tabriz"
    ],
    "membership": [
        "Basic", "Premium", "Standard",
        "Premium", "Basic", "Standard"
    ],
    "satisfaction": [
        "Low", "High", "Medium",
        "High", "Low", "Medium"
    ]
})

# ==================================================
print("\nبخش ۱: نمایش دیتاست اولیه")
# ==================================================

print(df)


# ==================================================
print("\nبخش ۲: اعمال One-Hot Encoding")
# ==================================================

df_encoded = pd.get_dummies(
    df,
    columns=["city", "membership"],
    dtype=int
)

print(df_encoded)


# ==================================================
print("\nبخش ۳: اعمال Ordinal Encoding")
# ==================================================

mapping = {
    "Low": 1,
    "Medium": 2,
    "High": 3
}

df_encoded["satisfaction_encoded"] = (
    df_encoded["satisfaction"].map(mapping)
)

print(df_encoded)


# ==================================================
print("\nبخش ۴: حذف ستون متنی اصلی")
# ==================================================

df_encoded = df_encoded.drop(columns=["satisfaction"])

print(df_encoded)


# ==================================================
print("\nبخش ۵: بررسی ستون‌های نهایی")
# ==================================================

print("Columns:")
print(df_encoded.columns.tolist())

print("\nData Types:")
print(df_encoded.dtypes)
````

# تمرین ۲ — Feature Scaling

دیتاست زیر را در نظر بگیر:

```python
import pandas as pd

df = pd.DataFrame({
    "age": [20, 25, 30, 35, 40, 45],
    "salary": [2000, 3000, 4000, 5000, 6000, 7000],
    "experience": [1, 2, 4, 6, 8, 10]
})
```

## وظایف

۱. میانگین و انحراف معیار هر ستون را با `describe()` بررسی کن.

۲. با استفاده از `StandardScaler` تمام ویژگی‌های عددی را مقیاس‌بندی کن.

۳. خروجی را به یک DataFrame تبدیل کن و نام ستون‌ها را حفظ کن.

۴. میانگین و انحراف معیار داده‌های مقیاس‌بندی‌شده را بررسی کن.

۵. همین کار را با `MinMaxScaler` تکرار کن.

۶. بررسی کن آیا مقادیر خروجی `MinMaxScaler` در بازه صفر تا یک قرار دارند یا خیر.

۷. در چند خط توضیح بده تفاوت خروجی این دو روش چیست.

**نکته:** در این تمرین هنوز مجموعه Train و Test نداریم. در پروژه واقعی، ابتدا داده‌ها را تقسیم می‌کنیم و Scaler را فقط روی `Train` با `fit` آموزش می‌دهیم.

</div>

````python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, StandardScaler

df = pd.DataFrame({
    "age": [20, 25, 30, 35, 40, 45],
    "salary": [2000, 3000, 4000, 5000, 6000, 7000],
    "experience": [1, 2, 4, 6, 8, 10]
})

# ==================================================
print("\nبخش ۱: بررسی دیتاست اولیه")
# ==================================================

print(df)
print("\nDescriptive Statistics:")
print(df.describe())


# ==================================================
print("\nبخش ۲: اعمال StandardScaler")
# ==================================================

standard_scaler = StandardScaler()

standard_scaled = standard_scaler.fit_transform(df)

df_standard = pd.DataFrame(
    standard_scaled,
    columns=df.columns
)

print(df_standard)


# ==================================================
print("\nبخش ۳: بررسی میانگین و انحراف معیار")
# ==================================================

print("Means:")
print(df_standard.mean())

print("\nStandard Deviations:")
print(df_standard.std(ddof=0))


# ==================================================
print("\nبخش ۴: اعمال MinMaxScaler")
# ==================================================

minmax_scaler = MinMaxScaler()

minmax_scaled = minmax_scaler.fit_transform(df)

df_minmax = pd.DataFrame(
    minmax_scaled,
    columns=df.columns
)

print(df_minmax)


# ==================================================
print("\nبخش ۵: بررسی بازه مقادیر")
# ==================================================

print("Minimum values:")
print(df_minmax.min())

print("\nMaximum values:")
print(df_minmax.max())

all_values_between_zero_and_one = (
    (df_minmax >= 0) & (df_minmax <= 1)
).all().all()

print("\nAll values between 0 and 1:")
print(all_values_between_zero_and_one)


# ==================================================
print("\nبخش ۶: مقایسه دو روش")
# ==================================================

print("StandardScaler:")
print(df_standard.describe())

print("\nMinMaxScaler:")
print(df_minmax.describe())


# ==================================================
print("\nبخش ۷: جلوگیری از Data Leakage")
# ==================================================

X_train, X_test = train_test_split(
    df,
    test_size=0.2,
    random_state=42
)

scaler = StandardScaler()

# یادگیری پارامترها فقط از داده‌های آموزش
X_train_scaled = scaler.fit_transform(X_train)

# استفاده از همان پارامترها برای داده‌های آزمون
X_test_scaled = scaler.transform(X_test)

print("Training Data Shape:", X_train_scaled.shape)
print("Testing Data Shape:", X_test_scaled.shape)
````
