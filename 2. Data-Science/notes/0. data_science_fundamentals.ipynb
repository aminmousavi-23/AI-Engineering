# Data Science Fundamentals

## What is Data Science?

Data Science is the field of using data, statistics, programming, and analytical methods to extract useful information, discover patterns, generate insights, and support decision-making or prediction.

Data Science is broader than Machine Learning. Machine Learning is one component of a typical Data Science workflow.

---

## Data Science Workflow

A typical Data Science workflow can include:

```text
Raw Data
    ↓
Data Understanding
    ↓
Exploratory Data Analysis (EDA)
    ↓
Data Cleaning
    ↓
Data Preprocessing
    ↓
Feature Engineering
    ↓
Train / Validation / Test
    ↓
Modeling
    ↓
Model Evaluation
    ↓
Interpretation
```

The workflow is usually iterative rather than strictly linear. A Data Scientist may return to previous steps after discovering problems or new insights.

---

## Dataset

A **Dataset** is a collection of data organized for analysis.

A tabular dataset is commonly represented as rows and columns.

---

## Observation

An **Observation** represents one individual record or instance in a dataset.

In a tabular dataset, an observation is usually represented by one row.

---

## Feature

A **Feature** is an input variable that describes an observation and can be used for analysis or prediction.

Example:

```text
age
salary
experience
```

---

## Target

The **Target** is the variable that we want to predict or explain.

Example:

```text
Features:
age
salary
experience

Target:
purchased
```

Other terms that may be encountered depending on the context include:

* Target Variable
* Output Variable
* Label
* Dependent Variable

These terms are related but are not always perfectly interchangeable.

---

## Variable

A **Variable** is a characteristic or attribute that can take different values.

### Numerical Variable

A variable whose values represent numerical quantities.

Examples:

```text
age = 25
salary = 5000
height = 183.5
```

### Categorical Variable

A variable whose values represent categories or groups.

Examples:

```text
city = "Paris"
plan = "Premium"
gender = "Male"
```

---

## Population

A **Population** is the complete group that we are interested in studying or making conclusions about.

Example:

```text
All software engineers in a country
```

---

## Sample

A **Sample** is a subset of the population selected for analysis.

```text
Population
    ↓
  Sample
```

Sampling is important because collecting data from the entire population is often impractical or impossible.

---

## Structured Data

**Structured Data** has a predefined and organized structure, often represented as rows and columns.

Examples:

* CSV files
* Excel tables
* SQL tables
* Relational databases

---

## Unstructured Data

**Unstructured Data** does not follow a fixed tabular structure.

Examples:

* Images
* Audio
* Video
* Free-form text

---

## Data Science vs Machine Learning

Machine Learning is a subset of the broader Data Science workflow.

```text
Data Science
│
├── Data Understanding
├── EDA
├── Statistics
├── Probability
├── Data Cleaning
├── Data Preprocessing
├── Feature Engineering
├── Visualization
├── Machine Learning
└── Evaluation
```

Therefore:

> Data Science is not simply training a Machine Learning model.

---

## Important Terminology

| Term                 | Definition                                             |
| -------------------- | ------------------------------------------------------ |
| Dataset              | A collection of data                                   |
| Observation          | A single record or instance                            |
| Feature              | An input variable                                      |
| Target               | The variable to predict or explain                     |
| Variable             | A characteristic that can take different values        |
| Population           | The complete group of interest                         |
| Sample               | A subset of a population                               |
| Numerical Variable   | A variable containing numerical quantities             |
| Categorical Variable | A variable representing categories                     |
| Structured Data      | Data with a predefined structure                       |
| Unstructured Data    | Data without a fixed tabular structure                 |
| Data Understanding   | Understanding the structure and meaning of data        |
| EDA                  | Exploratory Data Analysis                              |
| Data Cleaning        | Identifying and handling data quality problems         |
| Data Preprocessing   | Preparing data for analysis or modeling                |
| Feature Engineering  | Creating or transforming useful features               |
| Data Pipeline        | A sequence of data-processing steps                    |
| Iterative Process    | A process that may repeatedly return to previous steps |

---

## Key Takeaways

* Data Science uses data to discover information, patterns, and insights.
* Machine Learning is only one part of Data Science.
* A dataset contains observations and variables.
* Features are commonly used as model inputs.
* The target is the variable we want to predict or explain.
* Variables can be numerical or categorical.
* A population is the complete group of interest.
* A sample is a subset of the population.
* Data Science workflows are usually iterative.

## Customer Orders

Consider the following dataset:

```python
import pandas as pd

orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "customer_age": [21, 35, 29, 42, 31],
    "order_value": [120, 450, 230, 800, 310],
    "city": ["Paris", "Lyon", "Paris", "Lille", "Lyon"],
    "returned": [0, 0, 1, 0, 1]
})
```

### Tasks

1. How many observations are in the dataset?
2. What are the features?
3. What could be the target if the goal is to predict returned orders?
4. Which variables are numerical?
5. Which variables are categorical?
6. What could the population be in this scenario?
7. What is the sample?
8. Use Pandas to print the number of observations.


````python
import pandas as pd

orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "customer_age": [21, 35, 29, 42, 31],
    "order_value": [120, 450, 230, 800, 310],
    "city": ["Paris", "Lyon", "Paris", "Lille", "Lyon"],
    "returned": [0, 0, 1, 0, 1]
})


# 1. Number of observations
number_of_observations = len(orders)


# 2. Features
features = [
    "customer_age",
    "order_value",
    "city"
]


# 3. Target
target = "returned"


# 4. Numerical variables
numerical_variables = [
    "customer_age",
    "order_value"
]


# 5. Categorical variables
categorical_variables = [
    "city"
]


# 6. Population
population = "All orders placed in the store"


# 7. Sample
sample = orders["order_id"].tolist()


# 8. Number of observations using Pandas
number_of_observations_from_shape = orders.shape[0]


print("Number of observations:", number_of_observations)
print("Features:", features)
print("Target:", target)
print("Numerical variables:", numerical_variables)
print("Categorical variables:", categorical_variables)
print("Population:", population)
print("Sample:", sample)
print("Number of observations (shape):", number_of_observations_from_shape)
````

## Student Performance

Consider the following dataset:

```python
import pandas as pd

students = pd.DataFrame({
    "student_id": [1, 2, 3, 4, 5, 6],
    "age": [19, 21, 20, 22, 19, 23],
    "study_hours": [2, 5, 3, 8, 4, 6],
    "major": ["CS", "Math", "CS", "Physics", "Math", "CS"],
    "passed": [0, 1, 1, 1, 0, 1]
})
```

### Tasks

1. What does one observation represent?
2. Identify the features.
3. Identify the target if the goal is to predict whether a student passes.
4. Which variables are numerical?
5. Which variables are categorical?
6. Which columns could potentially be used as model inputs?
7. How many observations are there?
8. Write Python code that prints the dataset shape.


````python
import pandas as pd

students = pd.DataFrame({
    "student_id": [1, 2, 3, 4, 5, 6],
    "age": [19, 21, 20, 22, 19, 23],
    "study_hours": [2, 5, 3, 8, 4, 6],
    "major": ["CS", "Math", "CS", "Physics", "Math", "CS"],
    "passed": [0, 1, 1, 1, 0, 1]
})


# 1. What does one observation represent?
# One observation represents one student.


# 2. Identify the features.
features = [
    "age",
    "study_hours",
    "major"
]

print("Features:", features)


# 3. Identify the target.
target = "passed"

print("Target:", target)


# 4. Which variables are numerical?
numerical_variables = [
    "age",
    "study_hours"
]

print("Numerical Variables:", numerical_variables)


# 5. Which variables are categorical?
categorical_variables = [
    "major"
]

print("Categorical Variables:", categorical_variables)


# 6. Which columns could potentially be used as model inputs?
model_inputs = [
    "age",
    "study_hours",
    "major"
]

print("Model Inputs:", model_inputs)


# 7. How many observations are there?
number_of_observations = len(students)

print("Number of Observations:", number_of_observations)


# 8. Print the dataset shape.
print("Dataset Shape:", students.shape)
````
