# Feature Engineering

## 1. تعریف

**Feature Engineering** فرایند ساخت، تغییر و انتخاب ویژگی‌ها (**Features**) برای بهبود عملکرد مدل‌های **Machine Learning** است.

هدف این است که اطلاعات موجود در داده‌ها به شکلی نمایش داده شوند که مدل بتواند الگوهای مفید را بهتر یاد بگیرد.

## 2. روش‌های اصلی

| روش | توضیح |
|---|---|
| Feature Creation | ساخت ویژگی جدید |
| Date-Time Extraction | استخراج اطلاعات زمانی |
| Mathematical Transformation | تبدیل ریاضی ویژگی‌ها |
| Feature Selection | انتخاب ویژگی‌های مفید |
| Interaction Features | ترکیب چند ویژگی |
| Aggregation Features | محاسبه مجموع، میانگین و سایر آمارها |

## 3. ساخت ویژگی جدید

```python
df["total_price"] = df["price"] * df["quantity"]
```

## 4. استخراج ویژگی‌های زمانی

```python
df["date"] = pd.to_datetime(df["date"])

df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["weekday"] = df["date"].dt.dayofweek
```

## 5. تبدیل ریاضی

برای مثال، `log1p` لگاریتم یک به‌علاوه مقدار را محاسبه می‌کند و برای داده‌های نامنفی، از جمله صفر، کاربرد دارد.

```python
import numpy as np

df["log_income"] = np.log1p(df["income"])
```

## 6. Feature Selection

انتخاب ویژگی‌های مرتبط و حذف ویژگی‌های کم‌فایده می‌تواند پیچیدگی مدل را کاهش دهد.

این کار باید با روش مناسب و بدون استفاده ناخواسته از اطلاعات مجموعه `Test` انجام شود.

## 7. Data Leakage

**Data Leakage** زمانی رخ می‌دهد که اطلاعاتی از خارج از داده‌های مجاز آموزش، به فرایند یادگیری وارد شود.

ویژگی‌ها باید فقط از اطلاعاتی ساخته شوند که در زمان پیش‌بینی واقعاً در دسترس هستند.

## 8. اصطلاحات مهم

- **Feature:** ویژگی ورودی مدل
- **Target:** متغیری که مدل باید پیش‌بینی کند
- **Feature Creation:** ساخت ویژگی
- **Feature Selection:** انتخاب ویژگی
- **Transformation:** تبدیل داده
- **Aggregation:** تجمیع داده
- **Interaction Feature:** ویژگی حاصل از ترکیب چند ویژگی
- **Data Leakage:** نشت اطلاعات

# تمرین ۱ — Feature Creation

```python
import pandas as pd

df = pd.DataFrame({
    "product": ["A", "B", "C", "D", "E"],
    "price": [100, 250, 150, 400, 300],
    "quantity": [2, 3, 5, 1, 4],
    "discount": [0.10, 0.20, 0.00, 0.15, 0.05],
    "order_date": [
        "2026-01-10",
        "2026-02-15",
        "2026-02-20",
        "2026-03-05",
        "2026-03-25"
    ]
})
```
## وظایف

۱. دیتاست را بررسی کن.

۲. ویژگی `total_price` را از حاصل‌ضرب `price` در `quantity` بساز.

۳. ویژگی `discount_amount` را با ضرب `total_price` در `discount` محاسبه کن.

۴. ویژگی `final_price` را پس از کسر تخفیف بساز.

۵. ستون `order_date` را به `datetime` تبدیل کن.

۶. ستون‌های `order_month` و `order_weekday` را استخراج کن.

۷. مجموع فروش نهایی را برای هر ماه محاسبه کن.

۸. دیتاست نهایی و نتیجه تجمیع ماهانه را چاپ کن.

</div>

````python
import pandas as pd

df = pd.DataFrame({
    "product": ["A", "B", "C", "D", "E"],
    "price": [100, 250, 150, 400, 300],
    "quantity": [2, 3, 5, 1, 4],
    "discount": [0.10, 0.20, 0.00, 0.15, 0.05],
    "order_date": [
        "2026-01-10",
        "2026-02-15",
        "2026-02-20",
        "2026-03-05",
        "2026-03-25"
    ]
})

df["total_price"] = df["price"] * df["quantity"]
df["discount_amount"] = df["total_price"] * df["discount"]
df["final_price"] = df["total_price"] - df["discount_amount"]
df["order_date"] = pd.to_datetime(df["order_date"])
df["order_month"] = df["order_date"].dt.month
df["order_weekday"] = df["order_date"].dt.dayofweek
monthly_sales = df.groupby("order_month")["final_price"].sum()

print(monthly_sales)

df
````

# تمرین ۲ — Feature Engineering and Analysis

```python
import pandas as pd

df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106, 107],
    "age": [22, 35, 28, 45, 31, 52, 26],
    "monthly_income": [2000, 4500, 3000, 7000, 3800, 9000, 2500],
    "monthly_spending": [1200, 2500, 1800, 4000, 2200, 5000, 1600],
    "visits_per_month": [3, 8, 5, 12, 6, 15, 4]
})
```

## وظایف

۱. ویژگی `annual_income` را از `monthly_income` بساز.

۲. ویژگی `spending_ratio` را از تقسیم `monthly_spending` بر `monthly_income` محاسبه کن.

۳. ویژگی `income_per_visit` را با تقسیم درآمد ماهانه بر تعداد مراجعه‌ها بساز.

۴. مشتریان را بر اساس نسبت هزینه به درآمد دسته‌بندی کن:

- کمتر از `0.4`: `Low`
- از `0.4` تا کمتر از `0.6`: `Medium`
- برابر یا بیشتر از `0.6`: `High`

۵. میانگین درآمد و هزینه ماهانه را محاسبه کن.

۶. مشتریان را بر اساس `annual_income` از بیشترین به کمترین مرتب کن.

۷. توضیح بده کدام ویژگی‌ها جدید هستند و هرکدام چه اطلاعاتی به مدل اضافه می‌کنند.

````python
import pandas as pd

df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106, 107],
    "age": [22, 35, 28, 45, 31, 52, 26],
    "monthly_income": [2000, 4500, 3000, 7000, 3800, 9000, 2500],
    "monthly_spending": [1200, 2500, 1800, 4000, 2200, 5000, 1600],
    "visits_per_month": [3, 8, 5, 12, 6, 15, 4]
})

# بخش ۱: ساخت ویژگی‌های جدید
df["annual_income"] = df["monthly_income"] * 12

df["spending_ratio"] = (
    df["monthly_spending"] / df["monthly_income"]
)

df["income_per_visit"] = (
    df["monthly_income"] / df["visits_per_month"]
)

# بخش ۲: دسته‌بندی نسبت هزینه به درآمد
df["spending_category"] = pd.cut(
    df["spending_ratio"],
    bins=[-float("inf"), 0.4, 0.6, float("inf")],
    labels=["Low", "Medium", "High"],
    right=False
)

print(df)

# بخش ۳: مرتب‌سازی مشتریان بر اساس درآمد سالانه
df_sorted = df.sort_values(
    by="annual_income",
    ascending=False
)

print("\nCustomers Sorted by Annual Income:")
print(df_sorted)

# بخش ۴: میانگین درآمد و هزینه ماهانه
print("\nAverage Monthly Income:")
print(df["monthly_income"].mean())

print("\nAverage Monthly Spending:")
print(df["monthly_spending"].mean())
````
