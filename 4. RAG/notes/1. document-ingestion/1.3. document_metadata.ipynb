# Document Metadata در RAG

## ۱. تعریف

**Metadata** اطلاعاتی درباره‌ی یک سند یا قطعه‌ی متن است که به شناسایی، دسته‌بندی، فیلتر کردن، مدیریت و ردیابی آن کمک می‌کند.

مثال:

```python
metadata = {
    "source": "hr_policy_2026.pdf",
    "department": "HR",
    "version": "2.0",
    "page": 3,
    "status": "active"
}
```

در این مثال، متن سند محتوای اصلی را مشخص می‌کند و Metadata اطلاعات توصیفی و مدیریتی مربوط به آن را نگه می‌دارد.

## ۲. کاربردهای Metadata

- **Source Attribution:** شناسایی منبعی که پاسخ از آن استخراج شده است.
- **Metadata Filtering:** محدود کردن جست‌وجو به اسناد دارای ویژگی‌های مشخص.
- **Version Management:** تشخیص نسخه‌ی سند و تاریخ اعتبار آن.
- **Document Organization:** دسته‌بندی اسناد بر اساس واحد، نوع یا مالک.
- **Access Control Support:** کمک به اعمال سیاست‌های دسترسی.
- **Traceability:** ردیابی Chunk تا سند اصلی و در صورت امکان صفحه‌ی مربوطه.

## ۳. انواع Metadata

### Document-Level Metadata

اطلاعات مربوط به کل سند:

- `document_id`
- `source`
- `version`
- `department`
- `effective_date`
- `status`

### Chunk-Level Metadata

اطلاعات مربوط به یک Chunk:

- `chunk_id`
- `document_id`
- `page`
- `chunk_index`
- `start_offset`
- `end_offset`

این فیلدها نمونه‌اند و لازم نیست همه‌ی آن‌ها در هر پروژه استفاده شوند.

## ۴. Metadata در معماری RAG

```text
Source Document
      |
      v
Document Metadata
      |
      v
Text Extraction
      |
      v
Text Cleaning
      |
      v
Chunking
      |
      v
Attach Chunk Metadata
      |
      v
Embeddings and Indexing
      |
      v
Metadata Filtering + Retrieval
      |
      v
Answer with Source References
```

Metadata باید در مراحل پردازش حفظ شود تا سیستم بتواند Chunkهای بازیابی‌شده را به منبع اصلی مرتبط کند.

## ۵. نمونه ساختار یک Chunk

```python
chunk = {
    "id": "HR-001-C2",
    "text": "Employees can request annual leave after six months.",
    "metadata": {
        "document_id": "HR-001",
        "source": "hr_policy_2026.pdf",
        "department": "HR",
        "version": "2.0",
        "page": 3
    }
}
```

## ۶. اصول طراحی در Production

۱. برای Metadata یک Schema مشخص تعریف کن.

۲. نام فیلدها و نوع داده‌ها را در تمام اسناد سازگار نگه دار.

۳. از شناسه‌های پایدار برای اسناد استفاده کن.

۴. اطلاعات منبع و شماره صفحه را در صورت نیاز حفظ کن.

۵. نسخه‌ی سند و تاریخ اعتبار آن را مدیریت کن.

۶. Metadata را قبل از ذخیره‌سازی اعتبارسنجی کن.

۷. اطلاعات حساس را بدون سیاست دسترسی مناسب در معرض کاربران قرار نده.

۸. به یاد داشته باش که وجود فیلدهای دسترسی در Metadata به‌تنهایی امنیت ایجاد نمی‌کند؛ سیستم باید مجوز کاربر را واقعاً بررسی کند.

۹. فقط Metadata موردنیاز را ذخیره کن تا پیچیدگی و هزینه‌ی نگه‌داری افزایش پیدا نکند.

## ۷. تفاوت Metadata با متن سند

- `Text`: محتوای واقعی که باید جست‌وجو یا به آن استناد شود.
- `Metadata`: اطلاعاتی که مشخص می‌کند این محتوا متعلق به کدام سند، صفحه، نسخه یا دسته‌بندی است.

Metadata معمولاً به‌صورت فیلدهای ساختاریافته ذخیره می‌شود. اضافه کردن بعضی اطلاعات مانند عنوان سند به متن Embedding ممکن است مفید باشد، اما باید بر اساس نیاز جست‌وجو تصمیم‌گیری شود.

**اصل کلیدی:** هر Chunk باید تا حد لازم اطلاعات کافی برای فیلتر کردن، مدیریت و ردیابی منبع خود را داشته باشد.

# پاسخ تمرین نظری — Document Metadata

## پاسخ سؤال ۱

فیلدهای پیشنهادی:

- `document_id`: شناسه‌ی پایدار سند.
- `source`: نام یا مسیر مرجع سند.
- `page`: شماره‌ی صفحه‌ی متن.
- `effective_date`: تاریخ شروع اعتبار سیاست.
- `updated_at`: زمان آخرین به‌روزرسانی.
- `document_type`: نوع سند، مانند `leave_policy`.
- `language`: زبان سند.

انتخاب فیلدها به نیازهای واقعی سیستم بستگی دارد.

## پاسخ سؤال ۲

می‌توان هنگام بازیابی، فیلترهایی مانند این اعمال کرد:

```python
filters = {
    "employment_type": "full_time",
    "status": "active"
}
```

در این حالت، جست‌وجو به اسناد فعال مربوط به کارکنان تمام‌وقت محدود می‌شود؛ البته به شرطی که Vector Database و لایه‌ی Retrieval از چنین فیلترهایی پشتیبانی کنند و Metadata به‌درستی ذخیره شده باشد.

## پاسخ سؤال ۳

ممکن است چند سند هم‌زمان فعال باشند، یا تاریخ اعتبار آن‌ها متفاوت باشد. همچنین وضعیت یک سند ممکن است به‌روز نشده باشد.

بنابراین، منطق انتخاب نسخه باید معیارهای مشخصی داشته باشد؛ برای مثال، بررسی `effective_date`، نسخه‌ی معتبر و وضعیت انتشار.

## پاسخ سؤال ۴

اگر اطلاعات منبع و شناسه‌ی سند حفظ نشوند:

- تشخیص منبع اصلی دشوار می‌شود.
- ارائه‌ی ارجاع دقیق به صفحه‌ی سند ممکن نیست.
- فیلتر کردن اسناد بر اساس واحد یا نوع کارمند دشوار می‌شود.
- مدیریت نسخه‌ها و حذف یا به‌روزرسانی Chunkها پیچیده‌تر می‌شود.

به همین دلیل، هر Chunk باید Metadata کافی برای ردیابی به سند اصلی را داشته باشد.

# تمرین عملی — مدیریت Metadata با Python

یک فهرست از اسناد زیر داری:

```python
documents = [
    {
        "document_id": "HR-001",
        "text": "Full-time employees receive annual leave.",
        "metadata": {
            "department": "HR",
            "employment_type": "full_time",
            "version": "1.0",
            "status": "archived",
            "page": 2
        }
    },
    {
        "document_id": "HR-002",
        "text": "Full-time employees can request annual leave after six months.",
        "metadata": {
            "department": "HR",
            "employment_type": "full_time",
            "version": "2.0",
            "status": "active",
            "page": 3
        }
    },
    {
        "document_id": "HR-003",
        "text": "Contract employees follow a separate leave policy.",
        "metadata": {
            "department": "HR",
            "employment_type": "contract",
            "version": "1.0",
            "status": "active",
            "page": 1
        }
    }
]
```

## وظایف

۱. فقط اسنادی را انتخاب کن که `status` آن‌ها `active` است.

۲. از میان اسناد فعال، فقط اسناد مربوط به `full_time` را نگه دار.

۳. برای هر سند انتخاب‌شده، `document_id`، متن، نام واحد و شماره‌ی صفحه را چاپ کن.

۴. تابعی به نام `filter_documents()` بنویس که نوع کارمند موردنظر را به‌عنوان ورودی بگیرد و اسناد فعال مربوط به آن نوع را برگرداند.

</div>

````python
documents = [
    {
        "document_id": "HR-001",
        "text": "Full-time employees receive annual leave.",
        "metadata": {
            "department": "HR",
            "employment_type": "full_time",
            "version": "1.0",
            "status": "archived",
            "page": 2
        }
    },
    {
        "document_id": "HR-002",
        "text": "Full-time employees can request annual leave after six months.",
        "metadata": {
            "department": "HR",
            "employment_type": "full_time",
            "version": "2.0",
            "status": "active",
            "page": 3
        }
    },
    {
        "document_id": "HR-003",
        "text": "Contract employees follow a separate leave policy.",
        "metadata": {
            "department": "HR",
            "employment_type": "contract",
            "version": "1.0",
            "status": "active",
            "page": 1
        }
    }
]


def filter_documents(documents, employment_type):
    results = []

    for document in documents:
        metadata = document["metadata"]

        if (
            metadata["status"] == "active"
            and metadata["employment_type"] == employment_type
        ):
            results.append(document)

    return results


filtered_documents = filter_documents(
    documents,
    employment_type="full_time"
)

for document in filtered_documents:
    print("Document ID:", document["document_id"])
    print("Text:", document["text"])
    print("Department:", document["metadata"]["department"])
    print("Page:", document["metadata"]["page"])
    print("-" * 40)
````
