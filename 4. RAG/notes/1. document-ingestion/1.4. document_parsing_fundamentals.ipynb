# Document Parsing Fundamentals

## ۱. تعریف

**Document Parsing** فرایند تحلیل محتوای یک سند و تبدیل آن به ساختاری معنادار و قابل‌پردازش است.

هدف فقط استخراج متن نیست؛ بلکه شناسایی اجزایی مانند عنوان‌ها، بخش‌ها، پاراگراف‌ها، فهرست‌ها و جدول‌ها و حفظ ارتباط بین آن‌هاست.

## ۲. تفاوت مفاهیم

- **Document Loading:** باز کردن فایل و آماده‌سازی آن برای پردازش.
- **Text Extraction:** استخراج متن یا محتوای قابل‌خواندن از فایل.
- **Document Parsing:** تشخیص ساختار و نوع اجزای محتوا.
- **Document Metadata:** نگهداری اطلاعات توصیفی مانند شناسه سند، نسخه و شماره صفحه.
- **Chunking:** تقسیم محتوای آماده‌شده به قطعه‌های مناسب برای Embedding و Retrieval.

## ۳. اجزای قابل Parse

- `Headings`: عنوان‌ها و سلسله‌مراتب بخش‌ها
- `Paragraphs`: پاراگراف‌ها
- `Lists`: فهرست‌های شماره‌گذاری‌شده و Bullet Listها
- `Tables`: ردیف‌ها، ستون‌ها و سرستون‌ها
- `Layout`: موقعیت و چیدمان اجزای صفحه
- `Scanned Content`: محتوای اسکن‌شده که ممکن است به OCR نیاز داشته باشد

## ۴. اهمیت در RAG

Parsing ضعیف می‌تواند ارتباط بین اطلاعات را از بین ببرد. برای مثال، اگر یک جدول به متن نامنظم تبدیل شود، مشخص نیست هر مقدار متعلق به کدام ستون است.

Parsing مناسب به حفظ ساختار معنایی سند، بهبود Chunking، افزایش کیفیت Retrieval و ارائه Source Citation دقیق‌تر کمک می‌کند.

## ۵. ابزارها

- `markdown-it-py`: تحلیل Markdown
- `BeautifulSoup`: تحلیل HTML
- `PyMuPDF` و `pdfplumber`: کار با PDF
- `unstructured`: پردازش ساختاریافته انواع اسناد
- ابزارهای `OCR`: استخراج متن از تصاویر و اسناد اسکن‌شده

انتخاب ابزار به فرمت فایل، پیچیدگی ساختار و نیازهای پروژه بستگی دارد.

## ۶. نکات مهم Production

1. ساختار و ترتیب منطقی محتوا را حفظ کن.
2. جدول‌ها را بدون حفظ ارتباط ردیف‌ها و ستون‌ها به متن نامنظم تبدیل نکن.
3. اطلاعاتی مانند `document_id`، `page_number` و `section_title` را در صورت امکان نگه دار.
4. خطاهای Parsing و اسناد بدون محتوای قابل‌استفاده را شناسایی کن.
5. Parsing را از Chunking جدا نگه دار تا هر مرحله قابل‌آزمایش و بهینه‌سازی باشد.
6. برای فرمت‌های پیچیده از Parser مناسب همان فرمت استفاده کن؛ Regex ساده جایگزین کامل Parser نیست.
7. در صورت نیاز، محتوای اصلی را برای بررسی خطاها و پردازش مجدد حفظ کن.

## ۷. مسیر کلی Document Ingestion

`Loading → Extraction → Cleaning → Parsing → Metadata → Chunking`

# تمرین عملی — Markdown Parser

این تمرین به کدنویسی با **Python** نیاز دارد و فقط از کتابخانه‌های استاندارد Python استفاده می‌کند.

متن زیر را در نظر بگیر:

```python
document = """
# Company Policy

All employees must follow company policies.

## Annual Leave

Full-time employees receive 20 days of annual leave.

## Sick Leave

Employees must notify their manager when taking sick leave.
"""
```

تابعی به نام `parse_sections(document)` بنویس که:

1. عنوان‌های Markdown را تشخیص دهد؛ عنوان `#` سطح ۱ و عنوان `##` سطح ۲ دارد.
2. محتوای هر بخش را تا عنوان بعدی جمع‌آوری کند.
3. خروجی را به شکل یک `list` از `dictionary`ها برگرداند.
4. هر دیکشنری شامل کلیدهای `heading`، `level` و `content` باشد.
5. بخش اول باید عنوان `Company Policy` و بخش‌های بعدی باید عنوان‌های `Annual Leave` و `Sick Leave` داشته باشند.

در پایان، عنوان و سطح هر بخش را چاپ کن.

</div>

````python
import re

document = """
# Company Policy

All employees must follow company policies.

## Annual Leave

Full-time employees receive 20 days of annual leave.

## Sick Leave

Employees must notify their manager when taking sick leave.
"""


def parse_sections(document):
    sections = []
    current_section = None

    for line in document.splitlines():
        line = line.strip()

        if not line:
            continue

        match = re.match(r"^(#{1,6})\s+(.+)$", line)

        if match:
            if current_section is not None:
                current_section["content"] = "\n".join(
                    current_section["content"]
                ).strip()

                sections.append(current_section)

            current_section = {
                "heading": match.group(2),
                "level": len(match.group(1)),
                "content": []
            }

        elif current_section is not None:
            current_section["content"].append(line)

    if current_section is not None:
        current_section["content"] = "\n".join(
            current_section["content"]
        ).strip()

        sections.append(current_section)

    return sections


sections = parse_sections(document)

for section in sections:
    print("Heading:", section["heading"])
    print("Level:", section["level"])
    print("Content:", section["content"])
    print("-" * 40)
````
