```python

# Expected relevant documents from the evaluation dataset
expected_docs = {
    "HR-001",
    "HR-002",
    "HR-003"
}

# Documents returned by the retriever
retrieved_docs = {
    "HR-001",
    "HR-004",
    "HR-005"
}

def evaluate_retrieval(expected, retrieved):

    # Correctly retrieved documents
    true_positives = expected & retrieved

    # Retrieved documents that are not relevant
    false_positives = retrieved - expected

    # Relevant documents that were missed
    false_negatives = expected - retrieved

    precision = (
        len(true_positives) / len(retrieved)
        if retrieved else 0.0
    )

    recall = (
        len(true_positives) / len(expected)
        if expected else 0.0
    )

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0.0
    )

    return {
        "true_positives": true_positives,
        "false_positives": false_positives,
        "false_negatives": false_negatives,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }


# Evaluate retrieval
results = evaluate_retrieval(
    expected_docs,
    retrieved_docs
)

# Display results
print("True Positives:", results["true_positives"])
print("False Positives:", results["false_positives"])
print("False Negatives:", results["false_negatives"])

print(f"\nPrecision: {results['precision']:.2f}")
print(f"Recall:    {results['recall']:.2f}")
print(f"F1 Score:  {results['f1']:.2f}")
```
