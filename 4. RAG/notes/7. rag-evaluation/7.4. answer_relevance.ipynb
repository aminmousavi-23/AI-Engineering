```python
import json
from openai import OpenAI

# 1. Connect to Ollama
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)

MODEL_ID = "qwen3:4b"

# 2. User question
question = "How many annual leave days do employees receive?"

# 3. Retrieved context
retrieved_contexts = [
    "Employees receive 26 days of annual leave per year.",
    "All annual leave requests require manager approval."
]

# 4. Generated answer from the RAG system
generated_answer = (
    "Employees receive 26 days of annual leave per year. "
    "All leave requests must be approved by their manager."
)

# 5. Evaluation prompt
prompt = f"""
Evaluate the relevance of this answer to the question.

Question: {question}
Answer: {generated_answer}

Score:
0 = Irrelevant
1 = Partially relevant
2 = Fully relevant

Return JSON only:
{{"score": 2, "reason": "Short explanation"}}
"""

# 6. Ask Qwen3 to evaluate
response = client.chat.completions.create(
    model=MODEL_ID,
    messages=[{"role": "user", "content": prompt}],
    temperature=0,
    extra_body={"think": False}
)

# 7. Parse evaluation result
result_text = response.choices[0].message.content

try:
    result = json.loads(result_text)

    score = result["score"]
    reason = result["reason"]

    if not isinstance(score, int) or score not in (0, 1, 2):
        raise ValueError("Invalid relevance score.")

    if not isinstance(reason, str):
        raise ValueError("Invalid evaluation reason.")

    # 8. Normalize score to [0, 1]
    answer_relevance = score / 2

    print("User Question:")
    print(question)

    print("\nGenerated Answer:")
    print(generated_answer)

    print("\nEvaluation:")
    print("Score:", score)
    print("Reason:", reason)

    print(f"\nAnswer Relevance: {answer_relevance:.2f}")

except (json.JSONDecodeError, KeyError, TypeError, ValueError) as error:
    print("Could not parse a valid evaluation.")
    print("Error:", error)
    print("Model response:", result_text)
```
