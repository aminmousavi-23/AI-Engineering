# Phase 3 — Chunking
## Topic 1: What Is Text Chunking?

### تعریف

**Text Chunking** فرایند تقسیم یک سند به قطعه‌های کوچک‌تر و معنادار است که بتوان آن‌ها را جداگانه ذخیره، Embedding و بازیابی کرد.

### چرا Chunking مهم است؟

1. مدیریت محدودیت `Context Window`
2. بهبود کیفیت `Retrieval`
3. کاهش حجم متن ارسالی به LLM
4. کاهش احتمالی هزینه و زمان پاسخ‌گویی
5. حفظ ارتباط معنایی محتوا

### ویژگی‌های Chunk مناسب

- محتوای آن تا حد امکان متمرکز و معنادار باشد.
- اطلاعات ضروری برای درک مفهوم را حفظ کند.
- اندازه آن متناسب با نوع محتوا و کاربرد باشد.
- در صورت نیاز، Metadata مربوط به سند و بخش را داشته باشد.
- امکان ردیابی آن تا منبع اصلی وجود داشته باشد.

### جایگاه Chunking در RAG

`Document Ingestion → Chunking → Embedding → Vector Database → Retrieval → Generation`

### نکته مهم

Chunk کوچک‌تر همیشه بهتر نیست و Chunk بزرگ‌تر هم الزاماً کیفیت بیشتری ندارد. اندازه و روش Chunking باید با استفاده از داده‌های واقعی ارزیابی شوند.

### روش‌های آینده این فاز

- `Fixed-Size Chunking`
- `Chunk Size and Chunk Overlap`
- `Sentence-Based Chunking`
- `Recursive Chunking`
- `Metadata-Aware Chunking`
- `Chunking Quality Evaluation`

# تمرین عملی — Fixed-Size Chunking

این تمرین به کدنویسی Python نیاز دارد و به کتابخانه اضافی احتیاج ندارد.

تابعی به نام `fixed_size_chunking(text, chunk_size)` بنویس که:

1. متن را به قطعه‌هایی با حداکثر `chunk_size` کاراکتر تقسیم کند.
2. قطعه‌ها را به ترتیب اصلی در یک `list` قرار دهد.
3. هیچ بخشی از متن را حذف نکند.
4. برای متن خالی، یک لیست خالی برگرداند.

متن آزمایشی:

```python
text = (
    "RAG retrieves relevant information. "
    "The LLM uses that information to answer questions."
)
```

تابع را با `chunk_size=20` اجرا کن و طول هر Chunk را نیز چاپ کن.

</div>

````python
def fixed_size_chunking(text, chunk_size):
    if chunk_size <= 0:
        raise ValueError("chunk_size must be greater than zero")

    if not text:
        return []

    chunks = []

    for start in range(0, len(text), chunk_size):
        chunk = text[start : start + chunk_size]
        chunks.append(chunk)

    return chunks


text = (
    "RAG retrieves relevant information. "
    "The LLM uses that information to answer questions."
)

chunks = fixed_size_chunking(text, chunk_size=20)

for index, chunk in enumerate(chunks, start=1):
    print(f"Chunk {index}: {chunk!r}")
    print("Length:", len(chunk))
    print("-" * 30)
````
