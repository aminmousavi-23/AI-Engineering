# Phase 3 — Chunking
## Topic 2: Fixed-Size Chunking

### تعریف

**Fixed-Size Chunking** روشی برای تقسیم متن به قطعه‌هایی با اندازه مشخص است. اندازه می‌تواند بر اساس Character یا Token تعیین شود.

### دو نوع رایج

- `Character-Based`: تقسیم بر اساس تعداد کاراکترها
- `Token-Based`: تقسیم بر اساس تعداد Tokenها

### فرمول تعداد Chunkها

برای متن غیرخالی و تقسیم بدون Overlap:

\[
N=\left\lceil\frac{L}{S}\right\rceil
\]

- \(N\): تعداد Chunkها
- \(L\): طول متن
- \(S\): اندازه هر Chunk

### مزایا

- پیاده‌سازی ساده
- سرعت بالا
- اندازه قابل‌کنترل
- مناسب برای ساخت Baseline

### محدودیت‌ها

- احتمال نصف شدن جمله‌ها
- احتمال جدا شدن عنوان از متن
- احتمال از دست رفتن زمینه معنایی
- مناسب نبودن برای برخی جدول‌ها و اسناد پیچیده

### نکات فنی

- `len()` در Python طول رشته را بر حسب عناصر رشته می‌دهد، نه Tokenهای مدل.
- تعداد کاراکتر و Token یکسان نیست.
- قطعه آخر ممکن است کوتاه‌تر از اندازه تعیین‌شده باشد.
- Fixed-Size Chunking بدون Overlap می‌تواند اطلاعات مرزی را از هم جدا کند.
- اندازه مناسب باید با داده و پرسش‌های واقعی ارزیابی شود.

### جایگاه در RAG

`Document Ingestion → Fixed-Size Chunking → Embedding → Vector Database`

### صورت سؤال

می‌خواهیم تابعی با نام `fixed_size_chunking` با استفاده از Python پیاده‌سازی کنیم که متن را برای استفاده در یک سیستم RAG به چند Chunk تقسیم کند.

تابع باید شرایط زیر را داشته باشد:

1. متن و دو پارامتر `chunk_size` و `overlap` را دریافت کند.
2. هر Chunk حداکثر `chunk_size` کاراکتر داشته باشد.
3. مقدار `overlap` بین دو Chunk متوالی رعایت شود.
4. اگر متن خالی باشد، فهرست خالی برگرداند.
5. اگر `chunk_size` نامعتبر یا `overlap` خارج از محدوده مجاز باشد، `ValueError` ایجاد کند.

متن زیر را با `chunk_size=25` و `overlap=5` پردازش کن و تمام Chunkها را همراه با شماره و طولشان چاپ کن.

```python
text = (
    "RAG divides documents into chunks. "
    "These chunks are embedded for semantic search."
)
```

</div>

````python
def fixed_size_chunking(
    text: str,
    chunk_size: int,
    overlap: int = 0
) -> list[str]:

    if chunk_size <= 0:
        raise ValueError("chunk_size must be positive")

    if overlap < 0 or overlap >= chunk_size:
        raise ValueError(
            "overlap must be >= 0 and < chunk_size"
        )

    if not text:
        return []

    step = chunk_size - overlap
    chunks = []

    for start in range(0, len(text), step):
        chunk = text[start:start + chunk_size]
        chunks.append(chunk)

        if start + chunk_size >= len(text):
            break

    return chunks


text = (
    "RAG divides documents into chunks. "
    "These chunks are embedded for semantic search."
)

chunks = fixed_size_chunking(
    text=text,
    chunk_size=25,
    overlap=5
)

for index, chunk in enumerate(chunks, start=1):
    print(f"Chunk {index}: {chunk!r}")
    print(f"Length: {len(chunk)}")
    print("-" * 40)
````
