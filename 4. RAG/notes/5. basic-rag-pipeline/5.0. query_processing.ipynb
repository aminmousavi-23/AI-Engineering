```python
import re
from sentence_transformers import SentenceTransformer

# Load embedding model
model = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)


def process_query(query: str):
    # 1. Cleaning
    query = query.strip()
    query = re.sub(r"\s+", " ", query)

    # 2. Normalization
    query = query.replace("ي", "ی")
    query = query.replace("ك", "ک")

    # 3. Query Embedding
    embedding = model.encode(query, convert_to_numpy=True, normalize_embeddings=True)

    return {"processed_query": query, "embedding": embedding}


# Test
query = "  شرایط بيمه تكميلي چيه؟  "

result = process_query(query)

print("Processed Query:", result["processed_query"])
print("Embedding Shape:", result["embedding"].shape)
```
