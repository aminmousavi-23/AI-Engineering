**GroupBy Practices**

```
# One group + one aggregation
df.groupby("department")["salary"].mean()

# Multiple aggregations
df.groupby("department")["salary"].agg(
    ["mean", "min", "max"]
)

# Multiple columns
df.groupby("department")[
    ["salary", "performance"]
].mean()

# Multiple grouping columns
df.groupby(
    ["department", "level"]
)["salary"].mean()

# Keep group column as normal column
df.groupby(
    "department",
    as_index=False
)["salary"].mean()

# Named aggregation
df.groupby("department").agg(
    avg_salary=("salary", "mean"),
    max_salary=("salary", "max")
)

# Number of rows per group
df.groupby("department").size()
```

##### Exercise 1 — Employee Analytics
```text
Create:

df = pd.DataFrame({
    "employee": [
        "Alice", "Bob", "Charlie",
        "Diana", "Ethan", "Frank",
        "Grace", "Henry"
    ],
    "department": [
        "AI", "Backend", "AI", "Backend",
        "AI", "Backend", "AI", "Backend"
    ],
    "level": [
        "Senior", "Junior", "Junior", "Senior",
        "Senior", "Junior", "Junior", "Senior"
    ],
    "salary": [
        9000, 5500, 6500, 8500,
        9500, 6000, 7000, 8800
    ],
    "performance": [
        95, 82, 88, 91,

    ]
})

Perform the following:

1. Calculate the average salary for each department.
2. Calculate the maximum performance for each department.
3. Calculate the number of employees in each department.
4. Calculate the average salary for each department + level combination.
5. Calculate mean, min, and max salary for each department.
6. Create a summary using named aggregation containing:
    average_salary
    max_salary
    average_performance
    employee_count
10. Return the final result with department as a normal column rather than the index.

````python
import pandas as pd

df = pd.DataFrame({
    "employee": [
        "Alice", "Bob", "Charlie",
        "Diana", "Ethan", "Frank",
        "Grace", "Henry"
    ],
    "department": [
        "AI", "Backend", "AI", "Backend",
        "AI", "Backend", "AI", "Backend"
    ],
    "level": [
        "Senior", "Junior", "Junior", "Senior",
        "Senior", "Junior", "Junior", "Senior"
    ],
    "salary": [
        9000, 5500, 6500, 8500,
        9500, 6000, 7000, 8800
    ],
    "performance": [
        95, 82, 88, 91,
        97, 79, 85, 93
    ]
})


# 1. Average salary for each department
print(
    df.groupby("department")["salary"].mean()
)


# 2. Maximum performance for each department
print(
    df.groupby("department")["performance"].max()
)


# 3. Number of employees in each department
print(
    df.groupby("department")["employee"].count()
)


# 4. Average salary for each department + level
print(
    df.groupby(["department", "level"])["salary"].mean()
)


# 5. Mean, min, and max salary for each department
print(
    df.groupby("department")["salary"].agg(
        ["mean", "min", "max"]
    )
)


# 6. Named aggregation
result = df.groupby("department").agg(
    average_salary=("salary", "mean"),
    max_salary=("salary", "max"),
    average_performance=("performance", "mean"),
    employee_count=("employee", "count")
)


# 7. Return department as a normal column
result = result.reset_index()

print(result)
````
