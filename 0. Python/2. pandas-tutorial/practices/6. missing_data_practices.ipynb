**Missing Data Practices**

```
# Detect
df.isna()
df.notna()

# Count
df.isna().sum()

# Filter missing
df[df["age"].isna()]

# Filter non-missing
df[df["age"].notna()]

# Drop rows
df.dropna()

# Drop columns
df.dropna(axis=1)

# Fill with a value
df["age"].fillna(0)

# Fill with mean
df["age"].fillna(df["age"].mean())

# Forward fill
df["age"].ffill()

# Backward fill
df["age"].bfill()
```

##### Exercise 1 — Handling Missing Employee Data
```text
Create:

df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Frank"],
    "age": [25, np.nan, 30, 28, np.nan, 35],
    "salary": [5500, 7200, np.nan, 6800, 6400, np.nan],
    "experience": [2, 6, np.nan, 4, 3, 10]
})

Perform the following:

1. Display the DataFrame.
2. Find all missing values using isna().
3. Count missing values in each column.
4. Calculate the total number of missing values.
5. Display only rows where salary is missing.
6. Display only rows where age is not missing.
7. Create a new DataFrame with all rows containing at least one missing value removed.
8. Fill missing age values with the mean age.
9. Fill missing salary values with the median salary.
10. Print the final cleaned DataFrame.

````python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Frank"],
    "age": [25, np.nan, 30, 28, np.nan, 35],
    "salary": [5500, 7200, np.nan, 6800, 6400, np.nan],
    "experience": [2, 6, np.nan, 4, 3, 10]
})

# 1. Display the DataFrame
print(df)

# 2. Find all missing values
print(df.isna())

# 3. Count missing values in each column
print(df.isna().sum())

# 4. Calculate the total number of missing values
print(df.isna().sum().sum())

# 5. Display only rows where salary is missing
print(df[df["salary"].isna()])

# 6. Display only rows where age is NOT missing
print(df[df["age"].notna()])

# 7. Create a new DataFrame with all rows
#    containing at least one missing value removed
cleaned_df = df.dropna()

print(cleaned_df)

# 8. Fill missing age values with the mean age
df["age"] = df["age"].fillna(df["age"].mean())

# 9. Fill missing salary values with the median salary
df["salary"] = df["salary"].fillna(df["salary"].median())

# 10. Print the final cleaned DataFrame
print(df)
````
