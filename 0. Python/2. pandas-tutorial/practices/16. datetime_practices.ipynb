````python
logs = pd.DataFrame({
    "timestamp": [
        "2026-09-01 09:15:00",
        "2026-09-01 14:30:00",
        "2026-09-01 18:45:00",
        "2026-09-02 10:00:00",
        "2026-09-02 16:20:00",
        "2026-09-03 11:10:00",
        "2026-09-03 15:40:00",
        "invalid-date"
    ],
    "model": [
        "CNN", "CNN", "ResNet",
        "ResNet", "CNN", "CNN",
        "ResNet", "CNN"
    ],
    "confidence": [
        0.91, 0.87, 0.95,
        0.82, 0.89, 0.93,
        0.96, 0.75
    ]
})

print(logs)
````

````python
logs["timestamp"] = pd.to_datetime(
    logs["timestamp"],
    errors="coerce"
)

print(logs)
````

````python
logs["year"] = logs["timestamp"].dt.year
logs["month"] = logs["timestamp"].dt.month
logs["day"] = logs["timestamp"].dt.day
logs["hour"] = logs["timestamp"].dt.hour
logs["day_of_week"] = logs["timestamp"].dt.dayofweek

logs["date"] = logs["timestamp"].dt.date

print(logs)
````

````python
after_12 = logs[
    logs["timestamp"].dt.hour > 12
]

print(after_12)
````

````python
target_date = pd.Timestamp("2026-09-02").date()

predictions_on_date = logs[
    logs["timestamp"].dt.date == target_date
]

print(predictions_on_date)
````

````python
earliest = logs["timestamp"].min()
latest = logs["timestamp"].max()

print("Earliest:", earliest)
print("Latest:", latest)
````

````python
time_difference = latest - earliest

print(time_difference)
````

````python
six_hours = pd.Timedelta(hours=6)

within_6_hours = logs[
    (logs["timestamp"] >= earliest) &
    (logs["timestamp"] <= earliest + six_hours)
]

print(within_6_hours)
````

````python
one_day = pd.DateOffset(days=1)

one_day_after_earliest = earliest + one_day

print(one_day_after_earliest)
````

````python
logs_indexed = logs.set_index("timestamp")

print(logs_indexed)
````

````python
daily_mean_confidence = (
    logs_indexed["confidence"]
    .resample("D")
    .mean()
)

print(daily_mean_confidence)
````

````python
daily_max_confidence = (
    logs_indexed["confidence"]
    .resample("D")
    .max()
)

print(daily_max_confidence)
````

````python
daily_prediction_count = (
    logs_indexed["confidence"]
    .resample("D")
    .count()
)

print(daily_prediction_count)
````

````python
dates = pd.date_range(
    start="2026-09-01",
    end="2026-09-03"
)

print(dates)
````

````python
formatted_timestamps = (
    logs["timestamp"]
    .dropna()
    .dt.strftime("%Y-%m-%d %H:%M")
)

print(formatted_timestamps)
````

````python
# 1. Why do we use pd.to_datetime() instead of keeping timestamps as strings?
# pd.to_datetime() converts strings into datetime values,
# allowing us to compare, subtract, filter, and analyze dates and times.


# 2. What is the difference between NaT and NaN?
# NaT means "Not a Time" and represents a missing datetime value.
# NaN means "Not a Number" and is generally used for missing numerical values.


# 3. What does the .dt accessor allow us to do?
# The .dt accessor allows us to extract and manipulate
# datetime components such as year, month, day, hour, etc.


# 4. What is the difference between Timedelta and DateOffset?
# Timedelta represents a fixed duration of time.
# DateOffset represents a calendar-based date offset.


# 5. What does resample("D") do?
# resample("D") groups time-series data into daily periods.
# "D" means Day.


# 6. What is the purpose of date_range()?
# date_range() generates a sequence of datetime values
# between a start date and an end date.
````
