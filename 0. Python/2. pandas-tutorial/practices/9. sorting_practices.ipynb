**Sorting Practices**

```
# One column
df.sort_values("score")

# Descending
df.sort_values("score", ascending=False)

# Multiple columns
df.sort_values(["label", "score"])

# Different direction for each column
df.sort_values(
    ["label", "score"],
    ascending=[True, False]
)

# NaN first
df.sort_values("score", na_position="first")

# Sort by index
df.sort_index()
```
