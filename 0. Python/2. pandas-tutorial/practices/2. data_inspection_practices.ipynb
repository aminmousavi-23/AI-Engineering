**Data Inspection Practices**

| Method / Attribute | کاربرد              |
| ------------------ | ------------------- |
| `head()`           | چند Row اول         |
| `tail()`           | چند Row آخر         |
| `shape`            | تعداد Row و Column  |
| `columns`          | نام Columnها        |
| `index`            | Indexهای DataFrame  |
| `dtypes`           | نوع داده Columnها   |
| `info()`           | اطلاعات کلی Dataset |
| `describe()`       | آمار توصیفی         |
| `len(df)`          | تعداد Rowها         |

Perform the following tasks:

1. Display the first 5 rows.
2. Display the first 3 rows.
3. Display the last 3 rows.
4. Print the DataFrame's shape.
5. Print the number of rows and columns separately.
6. Print all column names.
7. Print the DataFrame's index.

````python
df = pd.DataFrame({
    "image_id": [
        "IMG001", "IMG002", "IMG003", "IMG004",
        "IMG005", "IMG006", "IMG007", "IMG008"
    ],
    "label": [
        "cat", "dog", "cat", "bird",
        "dog", "cat", "bird", "dog"
    ],
    "confidence": [
        0.95, 0.82, 0.91, 0.76,
        0.88, 0.97, 0.69, 0.85
    ],
    "width": [
        224, 224, 224, 128,
        224, 256, 128, 224
    ],
    "height": [
        224, 224, 224, 128,
        224, 256, 128, 224
    ]
})

# 1. First 5 rows
print(df.head())

# 2. First 3 rows
print(df.head(3))

# 3. Last 3 rows
print(df.tail(3))

# 4. Shape
print(df.shape)

# 5. Number of rows and columns separately
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# 6. Column names
print(df.columns)

# 7. DataFrame index
print(df.index)

# 8. Data types
print(df.dtypes)

# 9. General DataFrame information
df.info()

# 10. Statistical information
print(df.describe())

# 11. Compare len(df) and df.shape[0]
print("len(df):", len(df))
print("df.shape[0]:", df.shape[0])
print("Are they equal?", len(df) == df.shape[0])
````
