##### Exercise 1 — Feature Engineering
```text
Create:

df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "age": [24, 31, 28, 35, 26],
    "salary": [5500, 7200, 6800, 9500, 6400],
    "experience": [1, 6, 4, 10, 3]
})

Perform the following:

1. Create an annual_salary column from salary.
2. Create a salary_per_year_of_experience column.
3. Create an is_senior boolean column where experience is at least 5.
4. Create an age_after_5_years column.
5. Increase every employee's salary by 10%.
6. Create a salary_category column:
    "high" if salary is at least 8000
    "medium" if salary is at least 6000
    "low" otherwise
7. Remove the age_after_5_years column.
8. Print the final DataFrame.

````python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "age": [24, 31, 28, 35, 26],
    "salary": [5500, 7200, 6800, 9500, 6400],
    "experience": [1, 6, 4, 10, 3]
})

# 1. Create annual salary
df["annual_salary"] = df["salary"] * 12

# 2. Salary per year of experience
df["salary_per_year_of_experience"] = (
    df["salary"] / df["experience"]
)

# 3. Senior employees
df["is_senior"] = df["experience"] >= 5

# 4. Age after 5 years
df["age_after_5_years"] = df["age"] + 5

# 5. Increase salary by 10%
df["salary"] = df["salary"] * 1.10

# 6. Salary category
df["salary_category"] = np.where(
    df["salary"] >= 8000,
    "high",
    np.where(
        df["salary"] >= 6000,
        "medium",
        "low"
    )
)

# 7. Remove age_after_5_years
df = df.drop(columns=["age_after_5_years"])

# 8. Print final DataFrame
print(df)
````

##### Exercise 2 — ML Prediction Feature Engineering
```text
Create:

predictions = pd.DataFrame({
    "image_id": [
        "img_001", "img_002", "img_003",
        "img_004", "img_005", "img_006"
    ],
    "confidence": [0.95, 0.62, 0.81, 0.45, 0.91, 0.73],
    "inference_time_ms": [12, 18, 10, 25, 14, 11],
    "width": [128, 256, 128, 512, 256, 128],
    "height": [128, 256, 128, 512, 256, 128]
})

Perform the following:

1. Create confidence_percent from confidence.
2. Create pixel_count using width × height.
3. Create is_high_confidence where confidence is at least 0.80.
4. Create is_fast where inference time is below 20 ms.
5. Create image_area_k containing pixel count divided by 1000.
6. Modify confidence so that all values are rounded to two decimal places.
7. Create a new DataFrame using assign() containing:
    confidence_percent
    pixel_count
    Print the final DataFrame and its columns.

````python
import pandas as pd

predictions = pd.DataFrame({
    "image_id": [
        "img_001", "img_002", "img_003",
        "img_004", "img_005", "img_006"
    ],
    "confidence": [0.95, 0.62, 0.81, 0.45, 0.91, 0.73],
    "inference_time_ms": [12, 18, 10, 25, 14, 11],
    "width": [128, 256, 128, 512, 256, 128],
    "height": [128, 256, 128, 512, 256, 128]
})

# 1. Confidence percentage
predictions["confidence_percent"] = predictions["confidence"] * 100

# 2. Pixel count
predictions["pixel_count"] = (
    predictions["width"] * predictions["height"]
)

# 3. High-confidence predictions
predictions["is_high_confidence"] = (
    predictions["confidence"] >= 0.80
)

# 4. Fast predictions
predictions["is_fast"] = (
    predictions["inference_time_ms"] < 20
)

# 5. Image area in thousands of pixels
predictions["image_area_k"] = (
    predictions["pixel_count"] / 1000
)

# 6. Round confidence to two decimal places
predictions["confidence"] = predictions["confidence"].round(2)

# 7. Create a new DataFrame using assign()
new_df = predictions.assign(
    confidence_percent=predictions["confidence_percent"],
    pixel_count=predictions["pixel_count"]
)

print(new_df)

print(new_df.columns)
````
