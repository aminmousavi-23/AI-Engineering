**Merge Practices**

```
# Join by index
df1.join(df2)

# Inner join
df1.join(df2, how="inner")

# Outer join
df1.join(df2, how="outer")

# Match left column with right index
df1.join(
    df2.set_index("key"),
    on="key"
)

# Join multiple DataFrames
df1.join([df2, df3])

# Handle duplicate column names
df1.join(
    df2,
    lsuffix="_left",
    rsuffix="_right"
)
```

Perform the following:

1. Use join() to combine the image metadata with model predictions.
2. Perform an inner join.
3. Perform a left join.
4. Perform an outer join.
5. Identify the image that has no prediction.
6. Identify the prediction that has no corresponding image metadata.
7. Create an is_correct column for the matched records.
8. Calculate the average confidence of the matched predictions.

````python
images = pd.DataFrame({
    "label": ["cat", "dog", "bird", "cat", "dog"],
    "width": [224, 224, 128, 224, 128]
}, index=[
    "img_001",
    "img_002",
    "img_003",
    "img_004",
    "img_005"
])

predictions = pd.DataFrame({
    "prediction": ["cat", "cat", "bird", "dog", "dog"],
    "confidence": [0.95, 0.72, 0.91, 0.83, 0.96]
}, index=[
    "img_001",
    "img_002",
    "img_003",
    "img_004",
    "img_006"
])


# 1. Join image metadata with predictions
joined = images.join(predictions)

print(joined)


# 2. Inner join
inner_df = images.join(
    predictions,
    how="inner"
)

print(inner_df)


# 3. Left join
left_df = images.join(
    predictions,
    how="left"
)

print(left_df)


# 4. Outer join
outer_df = images.join(
    predictions,
    how="outer"
)

print(outer_df)


# 5. Identify the image that has no prediction
no_prediction = left_df[
    left_df["prediction"].isna()
]

print(no_prediction)


# 6. Identify the prediction that has no corresponding image metadata
no_metadata = outer_df[
    outer_df["label"].isna()
]

print(no_metadata)


# 7. Create is_correct column for matched records
matched = inner_df.copy()

matched["is_correct"] = (
    matched["label"] == matched["prediction"]
)

print(matched)


# 8. Calculate average confidence of matched predictions
average_confidence = matched["confidence"].mean()

print(average_confidence)
````
