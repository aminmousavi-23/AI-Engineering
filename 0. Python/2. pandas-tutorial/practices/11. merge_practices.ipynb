**Merge Practices**

```
# Basic
pd.merge(df1, df2, on="id")

# Inner
pd.merge(df1, df2, on="id", how="inner")

# Left
pd.merge(df1, df2, on="id", how="left")

# Right
pd.merge(df1, df2, on="id", how="right")

# Outer
pd.merge(df1, df2, on="id", how="outer")

# Different column names
pd.merge(
    df1,
    df2,
    left_on="user_id",
    right_on="customer_id"
)

# Multiple keys
pd.merge(
    df1,
    df2,
    on=["store", "product"]
)

# Custom suffixes
pd.merge(
    df1,
    df2,
    on="id",
    suffixes=("_left", "_right")
)

# Validate relationship
pd.merge(
    df1,
    df2,
    on="id",
    validate="one_to_one"
)

# Track source
pd.merge(
    df1,
    df2,
    on="id",
    indicator=True
)
```

```text
Create:

Perform the following:

1. Perform an inner merge using department_id.
2. Perform a left merge.
3. Perform an outer merge.
4. Find employees whose department does not exist in departments.
5. Use indicator=True to identify unmatched records.
6. Create the final DataFrame containing:
    employee_id
    name
    department
    manager
7. Explain why employee Ethan behaves differently between the inner and left merge.

````python
employees = pd.DataFrame({
    "employee_id": [101, 102, 103, 104, 105],
    "name": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "department_id": [1, 2, 1, 3, 4]
})

departments = pd.DataFrame({
    "department_id": [1, 2, 3],
    "department": ["AI", "Backend", "DevOps"],
    "manager": ["Sarah", "John", "Mike"]
})


# 1. Inner merge
inner_df = pd.merge(
    employees,
    departments,
    how="inner",
    on="department_id"
)

print(inner_df)


# 2. Left merge
left_df = pd.merge(
    employees,
    departments,
    how="left",
    on="department_id"
)

print(left_df)


# 3. Outer merge
outer_df = pd.merge(
    employees,
    departments,
    how="outer",
    on="department_id"
)

print(outer_df)


# 4. Find employees whose department does not exist
unmatched_employees = left_df[
    left_df["department"].isna()
]

print(unmatched_employees)


# 5. Use indicator=True
merged_with_indicator = pd.merge(
    employees,
    departments,
    how="outer",
    on="department_id",
    indicator=True
)

print(merged_with_indicator)


# 6. Create final DataFrame
final_df = left_df[
    ["employee_id", "name", "department", "manager"]
]

print(final_df)
````
