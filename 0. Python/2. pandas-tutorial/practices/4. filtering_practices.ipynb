**Filtering Practices**

```
# Greater than
df[df["salary"] > 7000]

# Equal
df[df["age"] == 30]

# Not equal
df[df["age"] != 30]

# AND
df[(df["age"] > 25) & (df["salary"] > 7000)]

# OR
df[(df["age"] < 25) | (df["salary"] > 9000)]

# NOT
df[~(df["age"] > 25)]

# Multiple values
df[df["department"].isin(["AI", "Backend"])]

# String filtering
df[df["name"].str.contains("a", case=False)]
```

##### Exercise 1 — Employee Filtering
```text
Create:

df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Frank", "Grace"],
    "department": ["AI", "Backend", "AI", "DevOps", "Backend", "AI", "DevOps"],
    "age": [24, 31, 28, 35, 26, 42, 29],
    "salary": [5500, 7200, 6800, 9500, 6400, 8800, 7600],
    "experience": [1, 6, 4, 10, 3, 15, 5]
})

Perform the following:

1. Select employees with a salary greater than 7000.
2. Select employees with at least 5 years of experience.
3. Select employees who are in the "AI" department.
4. Select employees who are in "AI" and have a salary greater than 6000.
5. Select employees who are in "Backend" or "DevOps".
6. Select employees whose age is between 25 and 35 inclusive.
7. Select employees whose names contain the letter "a" regardless of case.
8. Select employees whose department is either "AI" or "DevOps" using .isin().

````python
import pandas as pd

df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Frank", "Grace"],
    "department": ["AI", "Backend", "AI", "DevOps", "Backend", "AI", "DevOps"],
    "age": [24, 31, 28, 35, 26, 42, 29],
    "salary": [5500, 7200, 6800, 9500, 6400, 8800, 7600],
    "experience": [1, 6, 4, 10, 3, 15, 5]
})

# 1. Salary greater than 7000
print(df[df["salary"] > 7000])

# 2. At least 5 years of experience
print(df[df["experience"] >= 5])

# 3. AI department
print(df[df["department"] == "AI"])

# 4. AI department AND salary greater than 6000
print(
    df[
        (df["department"] == "AI") &
        (df["salary"] > 6000)
    ]
)

# 5. Backend OR DevOps
print(
    df[
        (df["department"] == "Backend") |
        (df["department"] == "DevOps")
    ]
)

# 6. Age between 25 and 35 inclusive
print(
    df[
        (df["age"] >= 25) &
        (df["age"] <= 35)
    ]
)

# 7. Names containing "a", case-insensitive
print(
    df[
        df["name"].str.contains("a", case=False)
    ]
)

# 8. Department is AI or DevOps
print(
    df[
        df["department"].isin(["AI", "DevOps"])
    ]
)
````
