##### <mark>Exercise 1 — Dot Product & Vector Norm</mark>
```text
Create two vectors:

a = np.array([2, 4, 6])
b = np.array([1, 3, 5])

Perform:

1. Calculate their dot product using np.dot().
2. Calculate their dot product using @.
3. Calculate the norm of a.
4. Calculate the norm of b.
5. Verify that both dot-product methods produce the same result.

````python
a = np.array([2, 4, 6])
b = np.array([1, 3, 5])

dot_1 = np.dot(a, b)
dot_2 = a @ b

norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

print("Dot product using np.dot():", dot_1)
print("Dot product using @:", dot_2)

print("Norm of a:", norm_a)
print("Norm of b:", norm_b)

# 5. Verify both dot products are equal
print("Both dot products are equal:", dot_1 == dot_2)
````

##### <mark>Exercise 2 — Neural Network Matrix Multiplication</mark>
```text
Simulate a simple neural network layer:

32 samples
128 input features
64 neurons

Create:

X
W

with appropriate shapes.

Then:

1. Calculate output = X @ W.
2. Print the shape of X.
3. Print the shape of W.
4. Print the shape of output.
5. Explain why the multiplication is valid.
6. Verify that the output shape is (32, 64).

````python
import numpy as np

x = np.random.random((32, 128))

w = np.random.random((128, 64))

output = x @ w

print("X shape:", x.shape)
print("W shape:", w.shape)
print("Output shape:", output.shape)
print("Output shape is (32, 64):", output.shape == (32, 64))
````

##### <mark>Exercise 3 — Linear System</mark>
```text
Solve the following system:

2x + y = 5
x + 3y = 6

Requirements:

1. Represent the equations using matrices a and b.
2. Solve the system using np.linalg.solve().
3. Print the values of x and y.
4. Verify the solution by calculating a @ solution.
5. Compare the result with b.

````python
a = np.array([
    [2, 1],
    [1, 3]
])

b = np.array([5, 6])

# Solve a @ solution = b
solution = np.linalg.solve(a, b)

print("Solution:", solution)

# Extract x and y
x = solution[0]
y = solution[1]

print("x =", x)
print("y =", y)

# Verify the solution
result = a @ solution

print("a @ solution =", result)
print("b =", b)

# Compare the result with b
print("Solution is correct:", np.allclose(result, b))
````
