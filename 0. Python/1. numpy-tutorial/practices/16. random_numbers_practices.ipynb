##### <mark>Exercise 1 — Synthetic ML Dataset</mark>
```text
Create a synthetic dataset with:

1000 samples
5 features

Requirements:

1. Generate the features using a normal distribution.
2. Use mean=0 and standard deviation=1.
3. Use a fixed random seed.
4. Print the shape.
5. Calculate the mean of each feature.
6. Calculate the standard deviation of each feature.
7. Verify that the generated statistics are approximately close to 0 and 1.

````python
np.random.seed(42)

array = np.random.normal(
    loc=0,
    scale=1,
    size=(1000, 5)
)

print("Shape:", array.shape)

feature_means = np.mean(array, axis=0)
print("Feature means:", feature_means)

feature_stds = np.std(array, axis=0)
print("Feature standard deviations:", feature_stds)
````

##### <mark>Exercise 2 — Random Sampling from a Dataset</mark>
Create:

sample_ids = np.arange(1000)

Simulate selecting a mini-batch of 32 samples.

Requirements:

1. Select 32 random IDs.
2. Make sure sampling is without replacement.
3. Print the selected IDs.
4. Verify that exactly 32 IDs were selected.
5. Verify that there are no duplicate IDs.

````python
sample_ids = np.arange(1000)

rng = np.random.default_rng(42)

random_ids = rng.choice(
    sample_ids,
    size=32,
    replace=False
)

print("Selected IDs:")
print(random_ids)

print("Number of selected IDs:", random_ids.size)
print("Exactly 32:", random_ids.size == 32)

print("No duplicates:", np.unique(random_ids).size == random_ids.size)
````

##### <mark>Exercise 3 — Model Initialization & Distributions</mark>
```text
Create a weight matrix for a hypothetical neural network:

Input features: 128
Neurons: 64

Generate the weights using a normal distribution with:

mean = 0
standard deviation = 0.01

Then:

1. Print the shape of the weight matrix.
2. Calculate its mean.
3. Calculate its standard deviation.
4. Find the minimum and maximum values.
5. Create another weight matrix using a uniform distribution between -0.01 and 0.01.
6. Compare the statistics of the two matrices.

````python
# ----------------------------------------
# Normal distribution
# ----------------------------------------
weights1 = rng.normal(
    loc=0,
    scale=0.01,
    size=(64, 128)
)

print("Normal Distribution")
print("Shape:", weights1.shape)
print("Mean:", weights1.mean())
print("Std:", weights1.std())
print("Min:", weights1.min())
print("Max:", weights1.max())

# ----------------------------------------
# Uniform distribution
# ----------------------------------------
weights2 = rng.uniform(
    low=-0.01,
    high=0.01,
    size=(64, 128)
)

print("\nUniform Distribution")
print("Shape:", weights2.shape)
print("Mean:", weights2.mean())
print("Std:", weights2.std())
print("Min:", weights2.min())
print("Max:", weights2.max())
````
