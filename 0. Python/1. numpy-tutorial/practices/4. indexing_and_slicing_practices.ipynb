Create this dataset:

```text

##### <mark>Exercise 2 — Image Cropping</mark>

Create this grayscale image:

image = np.arange(64).reshape(8, 8)

Then:

1. Print the entire image.
2. Extract the top-left 4×4 region.
3. Extract the bottom-right 3×3 region.
4. Extract rows 2 to 5.
5. Extract columns 3 to 6.

````python
image = np.arange(64).reshape(8, 8)

print(image)
print("--------------------")
print(image[0:4,0:4])
print("--------------------")
print(image[5:8,5:8])
print("--------------------")
print(image[1:5])
print("--------------------")
print(image[:,1:5])
````

##### <mark>Exercise 3 — AI Batch Selection</mark>

Create a synthetic image batch:

images = np.zeros((16, 128, 128, 3), dtype=np.float32)

Then:

1. Select the first image.
2. Select images 0 through 3.
3. Select the first channel of all images.
4. Select the first 64×64 region from the first image.
5. Print the shape of every selected result.

````python
images = np.zeros((16, 128, 128, 3), dtype=np.float32)

first_image = images[0]
print(first_image.shape)

images_0_to_3 = images[0:4]
print(images_0_to_3.shape)

first_channel = images[:, :, :, 0]
print(first_channel.shape)

region = images[0, 0:64, 0:64, :]
print(region.shape)
````
