**Axis Practices**

````python
import numpy as np
````

- axis=0 -> aggregation on rows
- axis=1 -> aggregation on columns

---------

##### <mark>Exercise 1 — Dataset Feature Statistics</mark>

Create a NumPy array representing 100 samples and 4 features.

Calculate:

1. Mean of each feature using axis=0
2. Minimum value of each feature using axis=0
3. Maximum value of each feature using axis=0
4. Mean of each sample using axis=1
5. Print the shape of every result

````python
array = np.random.random((100, 4))

feature_mean = np.mean(array, axis=0)
feature_min = np.min(array, axis=0)
feature_max = np.max(array, axis=0)
sample_mean = np.mean(array, axis=1)

print("Feature Mean:")
print(feature_mean)
print("Shape:", feature_mean.shape)

print("\nFeature Minimum:")
print(feature_min)
print("Shape:", feature_min.shape)

print("\nFeature Maximum:")
print(feature_max)
print("Shape:", feature_max.shape)

print("\nSample Mean:")
print(sample_mean)
print("Shape:", sample_mean.shape)
````

##### <mark>Exercise 2 — Image Batch Analysis</mark>

Create a NumPy array representing:
```text
8 RGB images
32 × 32 pixels
3 channels

So the shape should be:

(8, 32, 32, 3)

Calculate:

1. The mean image across the batch using axis=0
2. The mean value of each RGB channel using axis=(0, 1, 2)
3. Print the shape of both results
4. Explain what each result represents

````python
images = np.random.random((8, 32, 32, 3))

mean_image = np.mean(images, axis=0)
channel_mean = np.mean(images, axis=(0, 1, 2))

print("Mean image:")
print(mean_image.shape)

print("Channel mean:")
print(channel_mean.shape)
````

##### <mark>Exercise 3 — Axis & keepdims</mark>

Create a feature matrix:
```text
5 samples
3 features

Calculate the feature mean in two ways:

x.mean(axis=0)

and:

x.mean(axis=0, keepdims=True)

Then:

1. Print both results
2. Print both shapes
3. Explain why the shapes are different
4. Also calculate the mean using axis=1, keepdims=True
5. Print its shape

````python
x = np.random.random((5, 3))

mean_axis0 = x.mean(axis=0)
mean_axis0_keepdims = x.mean(axis=0, keepdims=True)
mean_axis1_keepdims = x.mean(axis=1, keepdims=True)

print("x:")
print(x)

print("\nMean axis=0:")
print(mean_axis0)
print("Shape:", mean_axis0.shape)

print("\nMean axis=0, keepdims=True:")
print(mean_axis0_keepdims)
print("Shape:", mean_axis0_keepdims.shape)

print("\nMean axis=1, keepdims=True:")
print(mean_axis1_keepdims)
print("Shape:", mean_axis1_keepdims.shape)
````
