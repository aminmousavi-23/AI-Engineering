##### <mark>Exercise 1 — Dataset Features</mark>

Create a NumPy array representing the following dataset:

```text
Age    Height    Weight
25     180       75
30     175       80
22     168       65
35     182       90

Then:

1. Print the array.
2. Print the type of the array.
3. Determine the number of samples.
4. Determine the number of features.

````python
arr = np.array(
            [   [25 ,180 ,75],
                [30 ,175 ,80],
                [22 ,168 ,65],
                [35 ,182 ,90]
            ]
        )

print(f"Array : {arr}")
print(f"Array type : {arr.dtype!s}")
print("Samples:", arr.shape[0])
print("Features:", arr.shape[1])
````

##### <mark>Exercise 2 — Image Representation</mark>

Create a NumPy array representing this grayscale image:

```text
0    50   100
150  200  255
30   80   220

Then:

1. Print the array.
2. Print its data type.
3. Access the pixel at the second row and third column.
4. Print its value.

````python
array = np.array(
            [
                [0, 50, 100],
                [150, 200, 255],
                [30, 80, 220]
            ]
        )

print(array)
print(array.dtype)
print(array[1,2])
````

##### <mark>Exercise 3 — AI Batch</mark>

Create a NumPy array from the following batch:

```text
Sample 1 → [0.2, 0.5, 0.8]
Sample 2 → [0.1, 0.7, 0.4]
Sample 3 → [0.9, 0.3, 0.6]
Sample 4 → [0.4, 0.8, 0.2]

Then:

1. Print the array.
2. Determine the number of samples.
3. Determine the number of features.
4. Access the third feature of the second sample.

````python
array = np.array(
            [
                [0.2, 0.5, 0.8],
                [0.1, 0.7, 0.4],
                [0.9, 0.3, 0.6],
                [0.4, 0.8, 0.2],
            ]
        )

print(array)
print(array.shape[0])
print(array.shape[1])
print(array[1, 2])
````
