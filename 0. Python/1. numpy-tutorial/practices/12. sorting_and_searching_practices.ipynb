##### <mark>Exercise 1 — Model Score Ranking</mark>
Create an array containing 10 model confidence scores.

Then:

1. Sort the scores in ascending order.
2. Sort the scores in descending order.
3. Use np.argsort() to get the indices that produce the descending order.
4. Print the top 3 scores and their original indices.

````python
scores = np.array([
    0.72, 0.91, 0.45, 0.83, 0.67,
    0.98, 0.54, 0.76, 0.89, 0.61
])

# 1. Sort in ascending order
ascending_scores = np.sort(scores)

print("1. Ascending order:")
print(ascending_scores)


# 2. Sort in descending order
descending_scores = np.sort(scores)[::-1]

print("\n2. Descending order:")
print(descending_scores)


# 3. Indices that produce descending order
descending_indices = np.argsort(scores)[::-1]

print("\n3. Descending indices:")
print(descending_indices)


# 4. Top 3 scores and their original indices
top_3_indices = descending_indices[:3]
top_3_scores = scores[top_3_indices]

print("\n4. Top 3 scores:")
print(top_3_scores)

print("Original indices:")
print(top_3_indices)
````

##### <mark>Exercise 2 — Top-K Classification</mark>
Create:
```text
class_names = np.array([
    "cat",
    "dog",
    "car",
    "bird",
    "horse",
    "truck"
])

probabilities = np.array([
    0.12,
    0.81,
    0.05,
    0.63,
    0.27,
    0.44
])

Find the top 3 predictions.

Your output should contain:

1. Class name
2. Probability
3. Original class index

Use np.argsort() rather than manually sorting the values.

````python
class_names = np.array([
    "cat",
    "dog",
    "car",
    "bird",
    "horse",
    "truck"
])

probabilities = np.array([
    0.12,
    0.81,
    0.05,
    0.63,
    0.27,
    0.44
])

sorted_indices = np.argsort(probabilities)[::-1]

top_3_indices = sorted_indices[:3]

print("Top 3 Predictions:")

for index in top_3_indices:
    print(
        f"Class: {class_names[index]}, "
        f"Probability: {probabilities[index]}, "
        f"Original Index: {index}"
    )
````

##### <mark>Exercise 3 — Searching Model Errors</mark>

Create:
```text
errors = np.array([
    0.12, 0.85, 0.31, 0.04, 0.92,
    0.18, 0.67, 0.11, 0.73, 0.25
])

Perform the following:

1. Find the index of the largest error.
2. Find the index of the smallest error.
3. Find all indices where the error is greater than 0.7.
4. Sort the errors from largest to smallest.
5. Find the top 3 largest errors and their original indices.

````python
errors = np.array([
    0.12, 0.85, 0.31, 0.04, 0.92,
    0.18, 0.67, 0.11, 0.73, 0.25
])

# 1. Index of the largest error
largest_error_index = np.argmax(errors)

print("1. Index of largest error:")
print(largest_error_index)


# 2. Index of the smallest error
smallest_error_index = np.argmin(errors)

print("\n2. Index of smallest error:")
print(smallest_error_index)


# 3. All indices where error > 0.7
high_error_indices = np.where(errors > 0.7)[0]

print("\n3. Indices where error > 0.7:")
print(high_error_indices)


# 4. Sort errors from largest to smallest
descending_indices = np.argsort(errors)[::-1]
descending_errors = errors[descending_indices]

print("\n4. Errors from largest to smallest:")
print(descending_errors)


# 5. Top 3 largest errors and their original indices
top_3_indices = descending_indices[:3]
top_3_errors = errors[top_3_indices]

print("\n5. Top 3 largest errors:")
for index, error in zip(top_3_indices, top_3_errors):
    print(f"Index: {index}, Error: {error}")
````
