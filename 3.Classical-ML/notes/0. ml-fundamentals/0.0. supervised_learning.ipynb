# 01. Supervised Learning — یادگیری نظارت‌شده

## ۱. تعریف

**Supervised Learning** یکی از روش‌های اصلی یادگیری ماشین است که در آن مدل با استفاده از داده‌های آموزشی دارای پاسخ صحیح (Labeled Data) آموزش می‌بیند تا رابطه میان ورودی‌ها و خروجی مورد انتظار را یاد بگیرد و برای داده‌های جدید پیش‌بینی انجام دهد.

هدف اصلی، یادگیری رابطه‌ای است که بتوان از آن برای پیش‌بینی نمونه‌هایی استفاده کرد که مدل قبلاً ندیده است.

## ۲. مفاهیم اصلی

### Features (X)

ویژگی‌ها یا متغیرهای ورودی هستند که اطلاعات لازم برای پیش‌بینی را در اختیار مدل قرار می‌دهند.

مثال در پیش‌بینی قیمت خانه:

- `Area`: متراژ خانه
- `Rooms`: تعداد اتاق‌ها
- `Age`: سن ساختمان

### Target (y)

متغیری است که مدل باید آن را پیش‌بینی کند.

مثال:

- `Price`: قیمت خانه

### Label

برچسب یا پاسخ صحیح هر نمونه است. این اصطلاح به‌ویژه در مسائل Classification برای نمایش کلاس واقعی نمونه استفاده می‌شود.

مثال:

- `Spam`
- `Not Spam`

### Training Data

مجموعه داده‌هایی است که برای آموزش مدل استفاده می‌شود. در Supervised Learning، این داده‌ها شامل Features و پاسخ‌های واقعی هستند.

### Model

مدل، رابطه‌ای است که از داده‌های آموزشی یاد گرفته می‌شود و برای تولید پیش‌بینی روی ورودی‌های جدید استفاده می‌شود.

### Prediction

خروجی‌ای است که مدل برای یک ورودی جدید تولید می‌کند.

### Error

اختلاف میان خروجی پیش‌بینی‌شده و مقدار واقعی است. روش محاسبه خطا به نوع مسئله و معیار ارزیابی بستگی دارد.

## ۳. چرخه یادگیری نظارت‌شده

۱. جمع‌آوری داده‌های دارای پاسخ واقعی  
۲. تعیین Features و Target  
۳. انتخاب الگوریتم مناسب  
۴. آموزش مدل با استفاده از `fit()`  
۵. تولید پیش‌بینی با استفاده از `predict()`  
۶. ارزیابی عملکرد مدل روی داده‌های مناسب  
۷. بهبود مدل در صورت نیاز

## ۴. انواع اصلی Supervised Learning

### Regression

زمانی استفاده می‌شود که خروجی مورد انتظار یک مقدار عددی باشد.

مثال‌ها:

- پیش‌بینی قیمت خانه
- پیش‌بینی درآمد
- تخمین زمان تحویل سفارش

### Classification

زمانی استفاده می‌شود که خروجی مورد انتظار یک کلاس یا دسته باشد.

مثال‌ها:

- تشخیص ایمیل اسپم
- تشخیص تراکنش مشکوک
- پیش‌بینی تأیید یا رد درخواست وام

## ۵. مثال عملی

فرض کن دیتاست زیر را داریم:

| Area | Rooms | Price |
|---:|---:|---:|
| 60 | 1 | 3 |
| 80 | 2 | 4 |
| 120 | 3 | 6 |

در این مثال:

- `Area` و `Rooms` ویژگی‌های ورودی هستند.
- `Price` متغیر هدف است.
- ردیف‌های دیتاست نمونه‌های آموزشی هستند.
- مدل تلاش می‌کند رابطه بین ویژگی‌ها و قیمت را یاد بگیرد.

پس از آموزش، می‌توان ویژگی‌های یک خانه جدید را به مدل داد تا قیمت آن را پیش‌بینی کند.

## ۶. آموزش و پیش‌بینی در Scikit-learn

```python
from sklearn.linear_model import LinearRegression

X = [[50], [70], [90], [110], [130]]
y = [2.5, 3.5, 4.5, 5.5, 6.5]

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[100]])

print(prediction[0])
```

- `X`: ویژگی‌های ورودی
- `y`: مقادیر واقعی هدف
- `fit()`: آموزش مدل
- `predict()`: پیش‌بینی برای ورودی جدید

## ۷. نکات مهم

- در Supervised Learning، پاسخ واقعی نمونه‌های آموزشی مشخص است.
- مدل باید از داده‌ها الگو یاد بگیرد، نه اینکه صرفاً آن‌ها را حفظ کند.
- کیفیت داده‌های آموزشی بر عملکرد مدل تأثیر می‌گذارد.
- عملکرد خوب روی داده‌های آموزشی لزوماً به معنای عملکرد خوب روی داده‌های جدید نیست.
- برای ارزیابی قابل‌اعتماد، معمولاً بخشی از داده‌ها برای ارزیابی مدل کنار گذاشته می‌شود.
- انتخاب Features مناسب و جلوگیری از Data Leakage برای ساخت مدل قابل‌اعتماد اهمیت دارد.
