# 02. Regression vs Classification

## ۱. تفاوت اصلی

در مسائل Supervised Learning، نوع مسئله را بر اساس متغیر هدف (Target) تعیین می‌کنیم.

- **Regression:** پیش‌بینی یک مقدار عددی
- **Classification:** پیش‌بینی یک کلاس یا دسته

مثال:

- پیش‌بینی قیمت خانه: Regression
- پیش‌بینی قبولی یا مردودی: Classification
- تشخیص تراکنش عادی یا مشکوک: Classification

## ۲. Regression

### تعریف

Regression روشی در یادگیری نظارت‌شده است که برای پیش‌بینی مقادیر عددی استفاده می‌شود.

### مثال‌های کاربردی

- پیش‌بینی قیمت خانه
- پیش‌بینی درآمد ماه آینده
- تخمین زمان تحویل سفارش
- پیش‌بینی میزان مصرف انرژی
- پیش‌بینی تعداد سفارش‌ها در یک بازه زمانی

### الگوریتم‌های نمونه

- Linear Regression
- Polynomial Regression
- Ridge Regression
- Lasso Regression

### معیارهای ارزیابی نمونه

- MAE
- MSE
- RMSE
- R²

این معیارها در مباحث بعدی به‌صورت دقیق بررسی می‌شوند.

## ۳. Classification

### تعریف

Classification روشی در یادگیری نظارت‌شده است که برای پیش‌بینی کلاس یا دسته یک نمونه استفاده می‌شود.

### مثال‌های کاربردی

- تشخیص ایمیل اسپم
- تشخیص تراکنش مشکوک
- پیش‌بینی تأیید یا رد درخواست وام
- تشخیص نوع بیماری از داده‌های پزشکی
- طبقه‌بندی تصاویر

### انواع Classification

**Binary Classification:** انتخاب بین دو کلاس؛ مانند Spam و Not Spam.

**Multiclass Classification:** انتخاب یک کلاس از میان چند کلاس؛ مانند تشخیص Cat، Dog یا Bird.

**Multilabel Classification:** اختصاص هم‌زمان چند برچسب به یک نمونه؛ مانند شناسایی Dog و Park در یک تصویر.

### الگوریتم‌های نمونه

- Logistic Regression
- K-Nearest Neighbors (KNN)
- Decision Trees
- Random Forest
- Support Vector Machine (SVM)
- Naive Bayes

### معیارهای ارزیابی نمونه

- Accuracy
- Precision
- Recall
- F1-Score
- ROC-AUC

## ۴. Probability و Predicted Class

برخی مدل‌های Classification می‌توانند احتمال تخمینی تعلق نمونه به کلاس‌ها را تولید کنند.

برای مثال:

- احتمال تخمینی کلاس عادی: 0.20
- احتمال تخمینی کلاس مشکوک: 0.80

با استفاده از یک قانون تصمیم‌گیری یا Classification Threshold می‌توان احتمال را به یک برچسب تبدیل کرد.

احتمال تخمینی مدل لزوماً احتمال واقعی نیست؛ میزان قابل‌اعتماد بودن آن به مدل و کالیبراسیون احتمال‌ها بستگی دارد.

## ۵. مقایسه Regression و Classification

| ویژگی | Regression | Classification |
|---|---|---|
| هدف | مقدار عددی | کلاس |
| مثال | قیمت خانه | اسپم یا غیر اسپم |
| خروجی نمونه | 5.2 | Spam |
| الگوریتم نمونه | Linear Regression | Logistic Regression |
| معیار نمونه | MAE، RMSE | Precision، Recall، F1 |

نکته: خروجی عددی لزوماً به معنای Regression نیست. اگر اعداد صرفاً کد کلاس‌ها باشند، مسئله همچنان می‌تواند Classification باشد.

## ۶. مثال عملی با Scikit-learn

### Regression

```python
from sklearn.linear_model import LinearRegression

X = [[50], [70], [90], [110], [130]]
y = [2.5, 3.5, 4.5, 5.5, 6.5]

model = LinearRegression()
model.fit(X, y)

print(model.predict([[100]]))
```

### Classification

```python
from sklearn.tree import DecisionTreeClassifier

X = [[1], [2], [3], [4], [5], [6]]
y = ["Fail", "Fail", "Fail", "Pass", "Pass", "Pass"]

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

print(model.predict([[5]]))
```

در هر دو مثال:

- `fit()` برای آموزش مدل استفاده می‌شود.
- `predict()` برای تولید پیش‌بینی به کار می‌رود.

تفاوت در نوع Target و خروجی مورد انتظار است.

## ۷. روش تشخیص نوع مسئله

۱. ابتدا Target را مشخص کن.
۲. بررسی کن که هدف، پیش‌بینی مقدار عددی است یا دسته‌بندی.
۳. اگر هدف پیش‌بینی یک مقدار عددی باشد، معمولاً Regression است.
۴. اگر هدف تعیین کلاس یا برچسب باشد، Classification است.
۵. الگوریتم و معیار ارزیابی را متناسب با مسئله انتخاب کن.
