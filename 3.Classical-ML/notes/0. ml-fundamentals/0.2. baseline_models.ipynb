# 03. Baseline Models — مدل‌های پایه

## ۱. تعریف

**Baseline Model** یک مدل یا روش ساده برای تولید پیش‌بینی است که به‌عنوان نقطه مرجع (Reference Point) برای ارزیابی مدل‌های پیچیده‌تر استفاده می‌شود.

هدف این است که بررسی کنیم آیا مدل اصلی واقعاً عملکرد بهتری از یک روش ساده دارد یا خیر.

Baseline لزوماً یک الگوریتم پیچیده Machine Learning نیست؛ ممکن است فقط از یک قانون ساده استفاده کند.

## ۲. چرا Baseline اهمیت دارد؟

- ایجاد یک نقطه شروع برای مقایسه مدل‌ها
- جلوگیری از پیچیده‌کردن غیرضروری سیستم
- بررسی ارزش واقعی استفاده از الگوریتم‌های پیچیده‌تر
- کمک به شناسایی عملکرد ضعیف مدل
- ایجاد معیاری برای ارزیابی پیشرفت مدل در طول توسعه

## ۳. Classification Baseline

در Classification، یکی از روش‌های ساده این است که همیشه کلاس پرتکرار داده‌های آموزشی را پیش‌بینی کنیم.

مثال:

فرض کن ۱۰۰۰ تراکنش داریم:

- ۹۵۰ تراکنش عادی
- ۵۰ تراکنش متقلبانه

اگر Baseline همیشه کلاس عادی را پیش‌بینی کند، Accuracy آن روی همین مجموعه برابر با ۹۵٪ خواهد بود.

این نتیجه نشان می‌دهد که Accuracy بالا لزوماً به معنای عملکرد مناسب در شناسایی کلاس اقلیت نیست.

### پیاده‌سازی با Scikit-learn

```python
from sklearn.dummy import DummyClassifier

X = [[1], [2], [3], [4], [5], [6]]
y = ["No", "No", "No", "No", "Yes", "No"]

baseline = DummyClassifier(
    strategy="most_frequent"
)

baseline.fit(X, y)

print(baseline.predict([[7], [8]]))
```

استراتژی `most_frequent` همیشه پرتکرارترین کلاس مشاهده‌شده در Targetهای آموزشی را پیش‌بینی می‌کند.

## ۴. Regression Baseline

در Regression، می‌توان برای تمام نمونه‌ها مقدار ثابتی مانند میانگین یا میانه Targetهای آموزشی را پیش‌بینی کرد.

مثال:

قیمت خانه‌های آموزشی:

`[2, 3, 4, 5, 6]`

میانگین برابر با ۴ است.

در نتیجه Baseline با استراتژی `mean` برای هر ورودی، مقدار ۴ را پیش‌بینی می‌کند.

### پیاده‌سازی با Scikit-learn

```python
from sklearn.dummy import DummyRegressor

X = [[50], [70], [90], [110], [130]]
y = [2, 3, 4, 5, 6]

baseline = DummyRegressor(
    strategy="mean"
)

baseline.fit(X, y)

print(baseline.predict([[60], [100], [150]]))
```

خروجی:

`[4. 4. 4.]`

## ۵. استراتژی‌های مهم Baseline

### DummyClassifier

- `most_frequent`: پیش‌بینی پرتکرارترین کلاس
- `prior`: پیش‌بینی پرتکرارترین کلاس و ارائه توزیع فراوانی کلاس‌ها به‌عنوان احتمال
- `uniform`: انتخاب تصادفی کلاس‌ها با احتمال برابر
- `stratified`: تولید پیش‌بینی تصادفی متناسب با فراوانی کلاس‌های آموزشی
- `constant`: پیش‌بینی یک کلاس ثابت تعیین‌شده

### DummyRegressor

- `mean`: پیش‌بینی میانگین Targetهای آموزشی
- `median`: پیش‌بینی میانه Targetهای آموزشی
- `quantile`: پیش‌بینی یک صدک تعیین‌شده
- `constant`: پیش‌بینی یک مقدار ثابت تعیین‌شده

## ۶. مقایسه Baseline و مدل اصلی

برای مقایسه معتبر:

۱. داده‌های آموزش و ارزیابی را به‌درستی جدا کن.

۲. Baseline و مدل اصلی را با داده‌های آموزشی یکسان آموزش بده.

۳. هر دو مدل را روی مجموعه ارزیابی یکسان بررسی کن.

۴. معیارهای مناسب با هدف مسئله را انتخاب کن.

۵. مشخص کن آیا مدل اصلی بهبود معناداری در عملکرد ایجاد کرده است یا خیر.

## ۷. نکات مهم

- Baseline یک نقطه مرجع است، نه لزوماً یک مدل مناسب برای استفاده در محیط عملیاتی.
- عملکرد Baseline به نوع مسئله، توزیع داده‌ها و معیار ارزیابی وابسته است.
- Accuracy بالا در داده‌های نامتوازن ممکن است گمراه‌کننده باشد.
- Baseline و مدل اصلی باید روی داده‌های ارزیابی یکسان مقایسه شوند.
- اطلاعات مجموعه ارزیابی نباید برای آموزش مدل استفاده شوند.
- اگر مدل اصلی بهتر از Baseline نیست، باید علت را بررسی کرد؛ پیچیدگی بیشتر به‌تنهایی مزیت محسوب نمی‌شود.
