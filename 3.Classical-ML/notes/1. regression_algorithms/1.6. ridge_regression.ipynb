# Ridge Regression — خلاصه

**Ridge Regression** نوعی Linear Regression است که از L2 Regularization برای کنترل اندازه‌ی ضرایب مدل استفاده می‌کند.

## فرمول

\[
Loss = MSE + \alpha\sum_{j=1}^{n}w_j^2
\]

## اصطلاحات کلیدی

- **L2 Regularization:** جریمه بر اساس مجموع مربع ضرایب.
- **Alpha:** شدت جریمه.
- **Coefficient:** ضریب هر Feature.
- **Multicollinearity:** همبستگی خطی بالا بین ویژگی‌ها.
- **StandardScaler:** ابزار هم‌مقیاس‌سازی ویژگی‌ها.
- **Pipeline:** ترکیب مراحل پیش‌پردازش و آموزش مدل.

## نکات مهم

۱. Ridge از L2 Regularization استفاده می‌کند.

۲. ضرایب را به سمت صفر کوچک می‌کند، اما معمولاً دقیقاً صفر نمی‌کند.

۳. افزایش Alpha جریمه را قوی‌تر می‌کند.

۴. Alpha بیش از حد بزرگ می‌تواند باعث Underfitting شود.

۵. StandardScaler معمولاً قبل از Ridge استفاده می‌شود.

۶. Ridge برای مقابله با مشکلاتی مانند Multicollinearity نیز مفید است.

۷. برای انتخاب Alpha مناسب می‌توان از Cross-Validation استفاده کرد.

# تمرین — Ridge Regression

**A. تفاوت اصلی Linear Regression و Ridge Regression چیست؟**

در Linear Regression، مدل تلاش می‌کند مجموع خطاهای پیش‌بینی را کمینه کند.

اما Ridge Regression علاوه بر خطای پیش‌بینی، یک جریمه برای بزرگ‌بودن ضرایب مدل در نظر می‌گیرد.

بنابراین Ridge از L2 Regularization استفاده می‌کند تا ضرایب را کوچک‌تر نگه دارد و احتمال Overfitting را کاهش دهد.

**B. چرا Ridge از L2 Regularization استفاده می‌کند؟**

زیرا L2 Regularization با جریمه‌کردن مربع ضرایب، آن‌ها را کوچک‌تر می‌کند.

این کار به کنترل پیچیدگی مدل و افزایش پایداری آن کمک می‌کند.

**C. اگر مقدار alpha از 0.1 به 100 افزایش پیدا کند، چه اتفاقی می‌افتد؟**

با افزایش alpha، قدرت Regularization بیشتر می‌شود.

در نتیجه ضرایب مدل بیشتر به سمت صفر حرکت می‌کنند.

اگر alpha بیش از حد بزرگ باشد، ممکن است مدل دچار Underfitting شود.

**D. چرا Ridge معمولاً ضرایب را دقیقاً صفر نمی‌کند؟**

زیرا Ridge از L2 Regularization استفاده می‌کند.

L2 معمولاً ضرایب را به صفر نزدیک می‌کند، اما برخلاف L1 Regularization در Lasso، آن‌ها را دقیقاً صفر نمی‌کند.

**E. چرا StandardScaler قبل از Ridge مفید است؟**

زیرا Featureها ممکن است مقیاس‌های متفاوتی داشته باشند.

StandardScaler باعث می‌شود Featureها در مقیاس مشابهی قرار بگیرند و Regularization تأثیر متعادل‌تری روی ضرایب داشته باشد.

**F. Ridge چگونه با Featureهای بسیار هم‌بسته برخورد می‌کند؟**

وقتی Featureها هم‌بستگی بالایی دارند، ضرایب مدل Linear Regression ممکن است ناپایدار شوند.

Ridge با محدودکردن اندازه ضرایب، معمولاً پایداری مدل را افزایش می‌دهد و اثر Multicollinearity را کاهش می‌دهد.


## پیاده‌سازی با Python

**نوع تمرین: کدنویسی Python الزامی است. از scikit-learn استفاده کن.**

داده‌های زیر را در نظر بگیر:

```python
X = [[50, 1], [70, 2], [90, 3], [110, 3], [130, 4]]
y = [2.0, 2.8, 3.8, 4.5, 5.5]
```

وظایف:

۱. با استفاده از `Ridge` یک مدل با `alpha=0.1` بساز.

۲. یک مدل دیگر با `alpha=100` بساز.

۳. هر دو مدل را با داده‌های بالا آموزش بده.

۴. مقدار `coef_` هر دو مدل را چاپ و مقایسه کن.

۵. با استفاده از هر دو مدل، مقدار Target را برای ورودی زیر پیش‌بینی کن:

```python
[[100, 3]]
```

۶. در کامنت توضیح بده چرا ضرایب دو مدل با یکدیگر متفاوت هستند.

۷. توضیح بده آیا صرفاً کوچک‌تر بودن ضرایب به معنی بهتر بودن مدل است؟

**نکته:** در این تمرین برای مقایسه‌ی ضرایب، بهتر است هر دو مدل را با `StandardScaler` و `Pipeline` بسازی تا مقیاس ویژگی‌ها یکسان باشد.

</div>

````python
from sklearn.linear_model import Ridge
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Define the dataset
X = [
    [50, 1],
    [70, 2],
    [90, 3],
    [110, 3],
    [130, 4]
]

y = [2.0, 2.8, 3.8, 4.5, 5.5]

# Create Ridge models with different alpha values
ridge_01 = Pipeline([
    ("scaler", StandardScaler()),
    ("model", Ridge(alpha=0.1))
])

ridge_100 = Pipeline([
    ("scaler", StandardScaler()),
    ("model", Ridge(alpha=100))
])

# Train both models
ridge_01.fit(X, y)
ridge_100.fit(X, y)

# Display coefficients
print("Ridge alpha=0.1")
print(ridge_01.named_steps["model"].coef_)

print("\nRidge alpha=100")
print(ridge_100.named_steps["model"].coef_)

# Predict a new sample
sample = [[100, 3]]

prediction_01 = ridge_01.predict(sample)
prediction_100 = ridge_100.predict(sample)

print("\nPrediction with alpha=0.1:", prediction_01)
print("Prediction with alpha=100:", prediction_100)

# Analysis:
# A higher alpha applies stronger regularization.
# Therefore, coefficients are pushed closer to zero.
#
# Smaller coefficients do not necessarily mean a better model.
# If alpha is too large, the model may underfit the training data.
#
# We should evaluate models using validation data or cross-validation.
````

````python

# Analysis:

# 1. The model with alpha=0.1 applies weak regularization.
#    Therefore, its coefficients are less restricted.

# 2. The model with alpha=100 applies much stronger regularization.
#    Therefore, its coefficients are pushed closer to zero.

# 3. Increasing alpha reduces the magnitude of the coefficients.
#    However, Ridge usually does not make coefficients exactly zero.

# 4. The predictions of the two models may be different
#    because they use different regularization strengths.

# 5. Smaller coefficients do not necessarily mean a better model.
#    A very high alpha may cause underfitting.

# 6. We should compare models using validation data
#    or cross-validation to select the best alpha value.

# 7. StandardScaler is used because the features have different scales.
#    It helps Ridge apply regularization more consistently.

# 8. The printed coefficients belong to the scaled features,
#    not the original features.
````
