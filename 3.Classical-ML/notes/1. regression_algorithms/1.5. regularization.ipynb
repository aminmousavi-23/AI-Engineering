# Regularization — خلاصه

**Regularization** تکنیکی برای کنترل پیچیدگی مدل و کاهش خطر Overfitting است.

در این روش، یک جریمه به تابع هزینه اضافه می‌شود تا مدل به سمت ضرایب کوچک‌تر هدایت شود.

## فرمول کلی

\[
Total\ Loss = Original\ Loss + Penalty
\]

## اصطلاحات کلیدی

- **Regularization:** کنترل پیچیدگی مدل با اعمال جریمه.
- **Penalty:** جریمه‌ای که به تابع هزینه اضافه می‌شود.
- **Alpha (α):** شدت جریمه در scikit-learn.
- **L1 Regularization:** جریمه بر اساس قدرمطلق ضرایب؛ مبنای Lasso.
- **L2 Regularization:** جریمه بر اساس مربع ضرایب؛ مبنای Ridge.
- **Feature Scaling:** هم‌مقیاس‌سازی ویژگی‌ها.
- **Overfitting:** یادگیری بیش از حد الگوهای داده‌های آموزشی.

## نکات مهم

۱. Regularization به کنترل پیچیدگی مدل کمک می‌کند.

۲. افزایش Alpha جریمه را قوی‌تر می‌کند.

۳. Alpha بیش از حد بزرگ ممکن است باعث Underfitting شود.

۴. L1 می‌تواند بعضی ضرایب را صفر کند.

۵. L2 ضرایب را کوچک می‌کند، اما معمولاً دقیقاً صفر نمی‌کند.

۶. Feature Scaling برای بسیاری از مدل‌های Regularized اهمیت دارد.

۷. مقدار مناسب Alpha را می‌توان با Cross-Validation انتخاب کرد.
