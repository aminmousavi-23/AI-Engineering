# Regression Loss Functions

## تعریف

`Loss Function` میزان خطای پیش‌بینی مدل را اندازه‌گیری می‌کند و در فرایند آموزش می‌تواند به مدل کمک کند پارامترهایش را بهبود دهد.

خطای هر پیش‌بینی:

\[
Error=y-\hat{y}
\]

## ۱. MAE — Mean Absolute Error

میانگین قدرمطلق خطاها:

\[
MAE=\frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
\]

- تفسیر ساده دارد.
- واحد آن با واحد Target یکسان است.
- نسبت به MSE تأکید کمتری بر خطاهای بزرگ دارد.

## ۲. MSE — Mean Squared Error

میانگین مربع خطاها:

\[
MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
\]

- خطاهای بزرگ را بیشتر جریمه می‌کند.
- واحد آن مربع واحد Target است.

## ۳. RMSE — Root Mean Squared Error

ریشه دوم MSE:

\[
RMSE=\sqrt{MSE}
\]

- خطاهای بزرگ تأثیر بیشتری دارند.
- واحد آن با واحد Target یکسان است.

## پیاده‌سازی در Python

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    root_mean_squared_error
)

mae = mean_absolute_error(y_true, y_pred)
mse = mean_squared_error(y_true, y_pred)
rmse = root_mean_squared_error(y_true, y_pred)
```

## نکته مهم

در هر سه معیار، مقدار کمتر معمولاً بهتر است. انتخاب معیار مناسب به مسئله و اهمیت خطاهای بزرگ بستگی دارد.

# تمرین عملی موضوع ۱۰ — Regression Loss Functions

## بخش اول — پیاده‌سازی با Python

با استفاده از داده‌های زیر، معیارهای خطا را محاسبه کن.

```python
y_true = [10, 20, 30, 40, 50]
y_pred = [12, 18, 33, 36, 55]
```

### وظایف

۱. با استفاده از `mean_absolute_error` مقدار **MAE** را محاسبه کن.

۲. با استفاده از `mean_squared_error` مقدار **MSE** را محاسبه کن.

۳. مقدار **RMSE** را محاسبه کن. اگر نسخه‌ی `scikit-learn` تو تابع `root_mean_squared_error` را پشتیبانی نمی‌کند، از ریشه‌ی دوم MSE استفاده کن.

۴. هر سه مقدار را چاپ کن.

## بخش دوم — تحلیل نتایج

پس از اجرای کد، به این سؤال‌ها پاسخ بده:

**A.** کدام معیار مستقیماً میانگین قدرمطلق خطاها را نشان می‌دهد؟

**B.** چرا مقدار MSE با واحد اصلی Target یکسان نیست؟

**C.** اگر یک پیش‌بینی خطای بسیار بزرگی داشته باشد، کدام معیار نسبت به آن حساس‌تر است: MAE یا MSE؟ چرا؟

**D.** آیا کمتر بودن یک معیار خطا همیشه به این معنی است که مدل برای مسئله‌ی واقعی بهتر است؟ توضیح بده.

</div>

````python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    root_mean_squared_error
)


y_true = [10, 20, 30, 40, 50]
y_pred = [12, 18, 33, 36, 55]

print(f"MAE : {mean_absolute_error(y_true, y_pred)}")
print(f"MSE : {mean_squared_error(y_true, y_pred)}")
print(f"RMSE : {root_mean_squared_error(y_true, y_pred)}")

# Answer A:
# MAE directly represents the mean of the absolute errors.
# In this example, the MAE is 3.8.

# Answer B:
# In MSE, the errors are squared, so the unit is also squared.
# For example, if the target is measured in million tomans,
# the unit of MSE will be (million tomans)^2.
# In contrast, RMSE returns to the original target unit
# because it takes the square root of MSE.

# Answer C:
# MSE is more sensitive to large errors than MAE.
# This is because MSE squares the errors, giving larger errors
# a greater impact on the final result.
# For example, an error of 10 contributes 100 to the squared error,
# whereas its absolute error is only 10.

# Answer D:
# No, a lower error metric does not always mean that a model
# is better for a real-world problem.
# The choice of metric depends on the problem's objectives.
# If large errors are very costly, MSE or RMSE may be more suitable.
# If we want to measure the average error magnitude in a simple,
# interpretable way, MAE may be a good choice.
# We should also consider the data distribution, outliers,
# and the real-world cost of prediction errors.
````
