# Lasso Regression

**Lasso Regression** نوعی Linear Regression است که از L1 Regularization استفاده می‌کند.

فرمول:

\[
Loss = MSE + \alpha \sum |w_j|
\]

### مفاهیم کلیدی

- **L1 Regularization:** جریمه مجموع قدر مطلق ضرایب.
- **Alpha:** کنترل قدرت Regularization.
- **Coefficient:** ضریب هر Feature در مدل.
- **Feature Selection:** انتخاب Featureهای مؤثرتر.
- **Zero Coefficient:** ضریبی که دقیقاً صفر شده است.

### رفتار Lasso

- Alpha کوچک: Regularization ضعیف.
- Alpha بزرگ: Regularization قوی.
- Alpha بسیار بزرگ: احتمال Underfitting.
- بعضی ضرایب می‌توانند دقیقاً صفر شوند.

### تفاوت Ridge و Lasso

- Ridge از L2 استفاده می‌کند و ضرایب را کوچک می‌کند.
- Lasso از L1 استفاده می‌کند و می‌تواند بعضی ضرایب را صفر کند.

### کتابخانه

`sklearn.linear_model.Lasso`

### تمرین موضوع ۱۵ — Lasso Regression

**A. تفاوت اصلی Lasso و Ridge چیست؟**

Ridge از L2 Regularization استفاده می‌کند و ضرایب را کوچک می‌کند، اما معمولاً آن‌ها را دقیقاً صفر نمی‌کند.

Lasso از L1 Regularization استفاده می‌کند و می‌تواند بعضی ضرایب را دقیقاً صفر کند.

**B. چرا Lasso می‌تواند بعضی از Coefficientها را دقیقاً صفر کند؟**

به دلیل ماهیت L1 Regularization که مجموع قدر مطلق ضرایب را جریمه می‌کند.

این ویژگی باعث می‌شود در فرایند بهینه‌سازی، بعضی ضرایب دقیقاً صفر شوند.

**C. اگر مقدار alpha را افزایش دهیم، چه اتفاقی برای Coefficientها می‌افتد؟**

با افزایش alpha، قدرت Regularization بیشتر می‌شود.

در نتیجه ضرایب بیشتر به سمت صفر حرکت می‌کنند و احتمال صفرشدن بعضی از آن‌ها افزایش می‌یابد.

اگر alpha بیش از حد بزرگ باشد، مدل ممکن است Underfitting شود.

**D. منظور از Feature Selection در Lasso چیست؟**

Feature Selection یعنی انتخاب Featureهای مؤثر و حذف Featureهای کم‌اثر از مدل.

Lasso با صفرکردن بعضی Coefficientها باعث می‌شود Featureهای مربوط به آن ضرایب در پیش‌بینی مدل نقشی نداشته باشند.

**E. آیا صفرشدن ضریب یک Feature به معنی بی‌اهمیت‌بودن قطعی آن در دنیای واقعی است؟ چرا؟**

خیر.

زیرا صفرشدن ضریب فقط نشان می‌دهد مدل با توجه به داده‌های آموزشی و مقدار alpha انتخاب‌شده، نیازی به آن Feature برای پیش‌بینی نداشته است.

ممکن است آن Feature در شرایط دیگر یا همراه با داده‌های متفاوت مفید باشد.

**F. اگر تمام Featureها برای پیش‌بینی مفید باشند، چرا ممکن است Ridge انتخاب مناسب‌تری نسبت به Lasso باشد؟**

زیرا Ridge تمام Featureها را در مدل حفظ می‌کند و فقط اندازه Coefficientها را کاهش می‌دهد.

اما Lasso ممکن است بعضی Coefficientها را صفر کند و Featureهای مربوطه را از مدل کنار بگذارد.

بنابراین وقتی انتظار داریم بیشتر Featureها اطلاعات مفیدی داشته باشند، Ridge می‌تواند انتخاب مناسب‌تری باشد.

### تمرین عملی — مقایسه Lasso با Alphaهای مختلف

این تمرین به کدنویسی Python نیاز دارد.

کتابخانه موردنیاز: `scikit-learn`

داده:

```python
X = [
    [50, 1, 10],
    [70, 2, 20],
    [90, 3, 30],
    [110, 3, 40],
    [130, 4, 50],
    [150, 5, 60],
    [170, 5, 70],
    [190, 6, 80]
]

y = [2.0, 2.8, 3.8, 4.5, 5.5, 6.2, 7.1, 7.8]
```

**وظایف:**

1. با استفاده از `Pipeline` و `StandardScaler` یک مدل Lasso با `alpha=0.01` بساز.
2. یک مدل دیگر با `alpha=1.0` بساز.
3. هر دو مدل را آموزش بده.
4. ضرایب هر دو مدل را با `coef_` چاپ کن.
5. بررسی کن آیا در یکی از مدل‌ها ضریبی دقیقاً صفر شده است یا خیر.
6. توضیح بده چرا ضرایب دو مدل متفاوت هستند.
7. آیا مدل با alpha بزرگ‌تر الزاماً عملکرد بهتری دارد؟

</div>


````python
from sklearn.linear_model import Lasso
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Define the dataset
X = [
    [50, 1, 10],
    [70, 2, 20],
    [90, 3, 30],
    [110, 3, 40],
    [130, 4, 50],
    [150, 5, 60],
    [170, 5, 70],
    [190, 6, 80]
]

y = [2.0, 2.8, 3.8, 4.5, 5.5, 6.2, 7.1, 7.8]

# Create the first Lasso model with alpha=0.01
lasso_001 = Pipeline([
    ("scaler", StandardScaler()),
    ("model", Lasso(alpha=0.01))
])

# Create the second Lasso model with alpha=1.0
lasso_1 = Pipeline([
    ("scaler", StandardScaler()),
    ("model", Lasso(alpha=1.0))
])

# Train both models
lasso_001.fit(X, y)
lasso_1.fit(X, y)

# Get coefficients from both models
coef_001 = lasso_001.named_steps["model"].coef_
coef_1 = lasso_1.named_steps["model"].coef_

# Print coefficients
print("Coefficients with alpha=0.01:")
print(coef_001)

print("\nCoefficients with alpha=1.0:")
print(coef_1)

# Check whether any coefficient is exactly zero
print("\nZero coefficients with alpha=0.01:")
print(coef_001 == 0)

print("\nZero coefficients with alpha=1.0:")
print(coef_1 == 0)

# Analysis:

# 1. The model with alpha=0.01 applies weak regularization.
#    Therefore, its coefficients are less restricted.

# 2. The model with alpha=1.0 applies stronger regularization.
#    Therefore, its coefficients are pushed closer to zero.

# 3. Lasso can make some coefficients exactly zero.
#    This behavior is called Feature Selection.

# 4. A coefficient equal to zero means that the corresponding
#    feature does not contribute to the model's predictions.

# 5. A higher alpha does not necessarily produce a better model.
#    Excessive regularization may cause underfitting.

# 6. We should evaluate both models using validation data
#    or cross-validation to select the appropriate alpha.
````
