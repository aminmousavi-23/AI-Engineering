# Gradient Descent — خلاصه

**Gradient Descent** یک الگوریتم بهینه‌سازی است که پارامترهای مدل را به‌صورت مرحله‌ای تغییر می‌دهد تا مقدار `Loss Function` کاهش پیدا کند.

## اصطلاحات کلیدی

- **Parameter:** مقداری که مدل یاد می‌گیرد؛ مانند `weight` و `bias`.
- **Loss Function:** تابعی برای اندازه‌گیری خطای مدل.
- **Gradient:** جهت و میزان تغییر Loss نسبت به یک پارامتر را نشان می‌دهد.
- **Learning Rate:** اندازه‌ی تغییر پارامترها در هر مرحله.
- **Optimization:** فرایند پیدا کردن پارامترهایی که تابع هدف را بهتر می‌کنند.
- **Convergence:** نزدیک شدن فرایند به یک وضعیت پایدار یا نقطه‌ی مناسب.

## فرمول به‌روزرسانی

\[
\theta_{\text{new}}=\theta_{\text{old}}-\eta \frac{\partial L}{\partial \theta}
\]

در این فرمول، \(\eta\) همان Learning Rate و مشتق Loss نسبت به پارامتر، گرادیان است.

## نکات مهم

۱. نرخ یادگیری خیلی کوچک می‌تواند آموزش را کند کند.

۲. نرخ یادگیری خیلی بزرگ ممکن است باعث نوسان یا دور شدن از کمینه شود.

۳. Gradient Descent لزوماً تضمین نمی‌کند به کمینه‌ی سراسری برسد؛ نتیجه به تابع Loss، شکل مسئله و تنظیمات الگوریتم بستگی دارد.

۴. در آموزش مدل، پیش‌بینی، محاسبه‌ی Loss، محاسبه‌ی گرادیان و به‌روزرسانی پارامترها تکرار می‌شوند.

۵. در شبکه‌های عصبی، Backpropagation گرادیان‌ها را محاسبه می‌کند و Optimizer از آن‌ها برای به‌روزرسانی پارامترها استفاده می‌کند.

**D.** تفاوت Gradient Descent با Backpropagation چیست؟

**G.** اگر گرادیان صفر باشد، آیا وزن در این مرحله تغییر می‌کند؟ توضیح بده.

```python
# ==========================================
# Part 1 — Conceptual Questions
# ==========================================

# Answer A:
# Gradient Descent is an optimization algorithm used to minimize
# a loss function by iteratively updating the model's parameters.


# Answer B:
# Gradient indicates the slope of the loss function with respect
# to a parameter, showing how the loss changes as the parameter changes.
# Learning Rate determines the step size used to update the parameter.


# Answer C:
# A very large Learning Rate can cause the algorithm to overshoot
# the minimum of the loss function.
# This may cause oscillations, increase the loss, or prevent convergence.
# However, a large Learning Rate does not necessarily increase the loss
# at every single step.


# Answer D:
# Gradient Descent is an optimization algorithm that uses gradients
# to update model parameters.
# Backpropagation calculates gradients in neural networks.
# In other words, Backpropagation calculates the gradients,
# while Gradient Descent uses them to update the parameters.


# ==========================================
# Part 2 — Calculation Questions
# ==========================================

# Given:
# Current weight = 5
# Gradient = 4
# Learning Rate = 0.2

# Answer E:
# Formula:
# new_weight = old_weight - learning_rate * gradient
#
# new_weight = 5 - (0.2 * 4)
# new_weight = 5 - 0.8
# new_weight = 4.2

# The new weight is 4.2.


# Answer F:
# If the gradient is positive, the parameter decreases
# when using the standard Gradient Descent update rule
# with a positive Learning Rate.
# This update moves the parameter in the negative gradient direction.


# Answer G:
# If the gradient is zero, the parameter does not change
# during this standard Gradient Descent update.
# This is because:
# new_weight = old_weight - learning_rate * 0
# new_weight = old_weight
#
# A zero gradient means the loss function has zero slope
# with respect to this parameter at the current point.
# However, this point is not necessarily a global minimum.
```
