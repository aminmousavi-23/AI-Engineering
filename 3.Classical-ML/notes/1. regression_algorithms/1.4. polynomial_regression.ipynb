# Polynomial Regression — خلاصه

**Polynomial Regression** روشی برای مدل‌کردن رابطه‌های منحنی بین ویژگی‌ها و Target است. در این روش، توان‌های ویژگی‌ها مانند \(x^2\) و \(x^3\) به داده‌ها اضافه می‌شوند.

## فرمول درجه‌ی دوم

\[
\hat y=b+w_1x+w_2x^2
\]

## اصطلاحات کلیدی

- **Polynomial Degree:** بالاترین درجه‌ی چندجمله‌ای که در مدل استفاده می‌شود.
- **PolynomialFeatures:** ابزار `scikit-learn` برای تولید ویژگی‌های چندجمله‌ای.
- **LinearRegression:** الگوریتمی که ضرایب مدل را یاد می‌گیرد.
- **Underfitting:** ساده بودن بیش از حد مدل برای یادگیری رابطه‌ی داده‌ها.
- **Overfitting:** یادگیری بیش از حد جزئیات و نویز داده‌های آموزشی.

## نکات مهم

۱. Polynomial Regression می‌تواند رابطه‌ی غیرخطی بین ورودی و خروجی را مدل کند.

۲. مدل همچنان نسبت به ضرایب خود خطی است.

۳. افزایش Degree پیچیدگی مدل را بیشتر می‌کند، اما تضمین نمی‌کند عملکرد آن بهتر شود.

۴. در `scikit-learn` می‌توان `PolynomialFeatures` و `LinearRegression` را با `Pipeline` ترکیب کرد.

۵. Degree مناسب را با داده‌های اعتبارسنجی یا Cross-Validation انتخاب می‌کنیم.

# تمرین — Polynomial Regression

## بخش اول: درک مفهومی

**نوع تمرین: مفهومی؛ نیازی به کدنویسی ندارد.**

**A.** تفاوت اصلی Linear Regression و Polynomial Regression چیست؟

**B.** اگر `degree=3` باشد، با یک ویژگی ورودی \(x\)، چه ویژگی‌های چندجمله‌ای ساخته می‌شوند؟

**C.** آیا افزایش Degree همیشه باعث بهتر شدن عملکرد مدل روی داده‌های جدید می‌شود؟ چرا؟

**D.** چرا Polynomial Regression با وجود مدل‌کردن منحنی، همچنان نسبت به ضرایب خود خطی است؟

## بخش دوم: پیاده‌سازی

**نوع تمرین: کدنویسی Python الزامی است. از `scikit-learn` و ابزارهای `PolynomialFeatures` و `LinearRegression` استفاده کن.**

داده‌های زیر را در نظر بگیر:

```python
X = [[1], [2], [3], [4], [5]]
y = [3, 6, 11, 18, 27]
```

وظایف:

۱. یک مدل Polynomial Regression با `degree=2` بساز.

۲. مدل را با داده‌های بالا آموزش بده.

۳. قیمت یا مقدار خروجی را برای ورودی‌های `6` و `7` پیش‌بینی کن.

۴. مقدار `degree` را به `1` تغییر بده و پیش‌بینی‌ها را با مدل درجه‌ی دوم مقایسه کن.

۵. در یک کامنت توضیح بده چرا مدل درجه‌ی دوم برای این داده‌ها مناسب‌تر است.

</div>

````python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5]]
y = [3, 6, 11, 18, 27]


# ==========================================
# Part 1 — Conceptual Questions
# ==========================================

# Answer A:
# Linear Regression models the relationship between input and output
# using a straight line.
# Polynomial Regression adds polynomial features such as x^2 and x^3
# to model curved relationships.


# Answer B:
# If degree=3 and there is one input feature x,
# PolynomialFeatures generates:
# x, x^2, and x^3.
# These features allow the model to learn more complex relationships.


# Answer C:
# No, increasing the degree does not always improve performance
# on unseen data.
# A high degree can cause overfitting by learning noise in the
# training data instead of the underlying relationship.
# We should evaluate the model using validation data or cross-validation.


# Answer D:
# Polynomial Regression can model a nonlinear relationship between
# the input x and the target y.
# However, it is still linear in its coefficients.
# For example:
# y = b + w1*x + w2*x^2
# The coefficients b, w1, and w2 are combined linearly.


# ==========================================
# Part 2 — Polynomial Regression, degree=2
# ==========================================

model_degree_2 = make_pipeline(
    PolynomialFeatures(degree=2, include_bias=False),
    LinearRegression()
)

model_degree_2.fit(X, y)

predictions_degree_2 = model_degree_2.predict([[6], [7]])

print("Degree 2 predictions:", predictions_degree_2)
# Expected output: [48. 63.]


# ==========================================
# Part 3 — Linear Regression, degree=1
# ==========================================

model_degree_1 = make_pipeline(
    PolynomialFeatures(degree=1, include_bias=False),
    LinearRegression()
)

model_degree_1.fit(X, y)

predictions_degree_1 = model_degree_1.predict([[6], [7]])

print("Degree 1 predictions:", predictions_degree_1)
# Expected output: approximately [32.6, 38.8]


# ==========================================
# Answer E — Comparing the Models
# ==========================================

# The degree=2 model is more suitable for this dataset because
# the target values follow the relationship:
# y = x^2 + 2*x
#
# For example:
# When x=6, y=6^2 + 2*6 = 48.
# When x=7, y=7^2 + 2*7 = 63.
#
# The degree=1 model can only fit a straight line, so it cannot
# represent this quadratic relationship exactly.
#
# Important:
# A higher-degree model is not always better.
# We should choose the degree based on the underlying relationship
# and performance on validation data or cross-validation.
````
