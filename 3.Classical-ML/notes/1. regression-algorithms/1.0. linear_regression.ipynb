# Linear Regression

## تعریف

**Linear Regression** یک الگوریتم `Supervised Learning` از نوع `Regression` است که برای پیش‌بینی مقادیر عددی با یادگیری یک رابطه خطی به کار می‌رود.

مثال‌ها:
- پیش‌بینی قیمت خانه
- تخمین فروش ماه آینده
- پیش‌بینی میزان مصرف انرژی

## معادله مدل

در حالت یک ویژگی:

\[
\hat{y}=wx+b
\]

- \(x\): ویژگی ورودی (`Feature`)
- \(\hat{y}\): مقدار پیش‌بینی‌شده (`Prediction`)
- \(w\): ضریب ویژگی (`Weight` یا `Coefficient`)
- \(b\): مقدار ثابت (`Bias` یا `Intercept`)

در حالت چند ویژگی:

\[
\hat{y}=w_1x_1+\cdots+w_nx_n+b
\]

## نحوه یادگیری

مدل در مرحله `Training`، ضرایب خود را از روی داده‌های آموزشی یاد می‌گیرد تا خطای پیش‌بینی کاهش پیدا کند.

## پیاده‌سازی در Python

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)

predictions = model.predict(X_test)
```

## اصطلاحات مهم

- `Training`: فرایند یادگیری مدل از داده‌ها
- `Coefficient`: ضریب مربوط به هر ویژگی
- `Intercept`: مقدار ثابت معادله
- `Prediction Error`: اختلاف مقدار واقعی و پیش‌بینی‌شده
- `Generalization`: توانایی مدل در پیش‌بینی داده‌های دیده‌نشده

**نکته:** Linear Regression برای پیش‌بینی مقادیر عددی استفاده می‌شود، نه مستقیماً برای پیش‌بینی کلاس‌هایی مانند «قبول» و «رد».

# پاسخ تمرین ۰۸ — Linear Regression

### A. قیمت پیش‌بینی‌شده برای خانه ۱۰۰ متری چقدر است؟

معادله مدل:

\[
\hat{y}=0.05x+1
\]

با جای‌گذاری \(x=100\):

\[
\hat{y}=0.05(100)+1=6
\]

**پاسخ:** مقدار پیش‌بینی‌شده برابر با ۶ است.

### B. مقادیر Weight و Bias چیست؟

با مقایسه معادله با فرم استاندارد:

\[
\hat{y}=wx+b
\]

نتیجه می‌گیریم:

- `Weight` یا `Coefficient`: برابر با \(0.05\)
- `Bias` یا `Intercept`: برابر با \(1\)

ضریب \(0.05\) نشان می‌دهد که با افزایش یک واحد در \(x\)، مقدار پیش‌بینی‌شده ۰٫۰۵ واحد افزایش می‌یابد.

### C. چرا Linear Regression برای پیش‌بینی قیمت مناسب است؟

زیرا قیمت یک مقدار عددی است و می‌خواهیم مقدار آن را پیش‌بینی کنیم.

- `Regression`: پیش‌بینی مقادیر عددی مانند قیمت یا فروش
- `Classification`: پیش‌بینی کلاس‌هایی مانند قبول/رد یا تقلبی/واقعی

### D. آیا عملکرد خوب روی داده‌های آموزشی، دقت روی داده‌های جدید را تضمین می‌کند؟

خیر. عملکرد خوب روی داده‌های آموزشی به‌تنهایی کافی نیست. ممکن است مدل الگوهای خاص داده‌های آموزشی را یاد گرفته باشد و روی داده‌های جدید عملکرد ضعیف‌تری داشته باشد.

برای بررسی توانایی `Generalization` باید عملکرد مدل را روی داده‌های دیده‌نشده ارزیابی کنیم؛ برای مثال با استفاده از `Test Set` و در مرحله انتخاب مدل با `Cross-Validation`.

# تمرین ۰۸ — بخش ۲: کدنویسی

**نوع تمرین:** نیازمند کدنویسی Python

**کتابخانه:** `scikit-learn`

## صورت مسئله

فرض کن اطلاعات زیر مربوط به متراژ و قیمت چند خانه است. قیمت‌ها بر حسب میلیارد تومان هستند.

| متراژ (متر مربع) | قیمت (میلیارد تومان) |
|---:|---:|
| ۵۰ | ۲٫۰ |
| ۷۰ | ۲٫۸ |
| ۹۰ | ۳٫۶ |
| ۱۱۰ | ۴٫۴ |
| ۱۳۰ | ۵٫۲ |

با استفاده از `LinearRegression` مدلی بساز که قیمت خانه را پیش‌بینی کند.

## وظایف

۱. داده‌های ورودی را در `X` و قیمت‌ها را در `y` قرار بده.

۲. یک مدل `LinearRegression` ایجاد کن.

۳. مدل را با `fit()` آموزش بده.

۴. مقادیر `Coefficient` و `Intercept` را چاپ کن.

۵. قیمت یک خانه با متراژ ۸۰ متر را با `predict()` پیش‌بینی کن.

۶. قیمت یک خانه با متراژ ۱۰۰ متر را هم پیش‌بینی کن.

## سؤال‌های تحلیلی

**A.** مقدار `Coefficient` چه چیزی را درباره رابطه متراژ و قیمت نشان می‌دهد؟

**B.** چرا ورودی `predict()` را به شکل `[[80]]` می‌نویسیم، نه `[80]`؟

**C.** آیا این مدل تضمین می‌کند قیمت واقعی تمام خانه‌ها را درست پیش‌بینی کند؟ چرا؟

</div>

````python

from sklearn.linear_model import LinearRegression

X = [[50], [70], [90], [110], [130]]
y = [2.0, 2.8, 3.6, 4.4, 5.2]

model = LinearRegression()
model.fit(X, y)

predictions = model.predict([[80], [100]])

print("Predicted prices:", predictions)
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

# -----------------------------------------
# Analytical Answers
# -----------------------------------------

# A. What does Coefficient = 0.04 mean?
# Answer:
# For every 1 square meter increase in area,
# the predicted price increases by 0.04 billion toman.
# This means approximately 40 million toman per square meter.
# This interpretation applies to the learned linear relationship.

# B. Why do we use [[80]] instead of [80]?
# Answer:
# scikit-learn expects X to be a 2D structure:
# (number_of_samples, number_of_features).
#
# [80] represents a 1D list.
# [[80]] represents one sample with one feature.
#
# For two houses, we use [[80], [100]]:
# 2 samples, 1 feature per sample.

# C. Does this model guarantee accurate predictions for all houses?
# Answer:
# No. The model learns a relationship from the training data.
# Real house prices may also depend on location, age, condition,
# number of rooms, and other factors.
#
# We should evaluate the model on unseen data.
# Good predictions on this small, perfectly linear dataset
# do not guarantee accurate predictions in the real world.
````
