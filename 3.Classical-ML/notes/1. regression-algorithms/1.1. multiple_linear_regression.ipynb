# Multiple Linear Regression

## تعریف

**Multiple Linear Regression** الگوریتمی برای پیش‌بینی یک متغیر عددی (`Target`) با استفاده از دو یا چند ویژگی (`Features`) است.

## معادله

\[
\hat{y}=w_1x_1+w_2x_2+\cdots+w_nx_n+b
\]

- \(x_i\): ویژگی ورودی
- \(w_i\): ضریب ویژگی (`Coefficient`)
- \(b\): مقدار ثابت (`Intercept`)
- \(\hat{y}\): مقدار پیش‌بینی‌شده

## تفسیر ضرایب

ضریب هر ویژگی نشان می‌دهد با افزایش یک واحد در آن ویژگی، مقدار پیش‌بینی‌شده چقدر تغییر می‌کند، درحالی‌که سایر ویژگی‌ها ثابت نگه داشته می‌شوند.

## پیاده‌سازی در Python

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)

predictions = model.predict(X_test)

print(model.coef_)
print(model.intercept_)
```

در `X`:
- هر سطر یک نمونه (`Sample`) است.
- هر ستون یک ویژگی (`Feature`) است.

**نکته:** افزایش تعداد ویژگی‌ها لزوماً به بهبود `Generalization` منجر نمی‌شود.

# تمرین ۰۹ — بخش ۲

**نوع تمرین:** نیازمند کدنویسی Python

**کتابخانه:** `scikit-learn`

## صورت مسئله

می‌خواهیم با استفاده از سه ویژگی، قیمت خانه‌ها را پیش‌بینی کنیم:

- `Area`: متراژ خانه
- `Rooms`: تعداد اتاق‌ها
- `Age`: سن ساختمان

داده‌های آموزشی:

| Area | Rooms | Age | Price |
|---:|---:|---:|---:|
| 50 | 1 | 10 | 2.0 |
| 70 | 2 | 5 | 2.8 |
| 90 | 3 | 3 | 3.8 |
| 110 | 3 | 2 | 4.5 |
| 130 | 4 | 1 | 5.5 |

قیمت‌ها بر حسب میلیارد تومان هستند.

## وظایف

۱. ویژگی‌ها را در `X` و قیمت‌ها را در `y` قرار بده.

۲. یک مدل `LinearRegression` بساز و با `fit()` آموزش بده.

۳. ضرایب مدل (`model.coef_`) و مقدار ثابت (`model.intercept_`) را چاپ کن.

۴. قیمت خانه‌ای با مشخصات زیر را پیش‌بینی کن:

- متراژ: ۸۰ متر
- تعداد اتاق‌ها: ۲
- سن ساختمان: ۵ سال

۵. قیمت خانه دیگری را پیش‌بینی کن:

- متراژ: ۱۰۰ متر
- تعداد اتاق‌ها: ۳
- سن ساختمان: ۲ سال

## سؤال‌های تحلیلی

**A.** چرا `model.coef_` به‌جای یک عدد، سه ضریب برمی‌گرداند؟

**B.** چرا ورودی خانه اول باید به شکل `[[80, 2, 5]]` باشد؟

**C.** اگر ترتیب ویژگی‌ها را هنگام پیش‌بینی تغییر بدهیم، مثلاً `[2, 80, 5]`، چه مشکلی ممکن است ایجاد شود؟

**D.** آیا ضرایب مثبت و منفی مدل به‌تنهایی ثابت می‌کنند که ویژگی‌ها علت افزایش یا کاهش قیمت هستند؟ توضیح بده.

</div>

````python

from sklearn.linear_model import LinearRegression

X = [
    [50, 1, 10],
    [70, 2, 5],
    [90, 3, 3],
    [110, 3, 2],
    [130, 4, 1]
]
y = [2.0, 2.8, 3.8, 4.5, 5.5]

model = LinearRegression()
model.fit(X, y)

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)

house_1 = [[80, 2, 5]]
prediction_1 = model.predict(house_1)

print("House 1 predicted price:", prediction_1[0])

house_2 = [[100, 3, 2]]
prediction_2 = model.predict(house_2)

print("House 2 predicted price:", prediction_2[0])


# -----------------------------------------
# Analytical Answers
# -----------------------------------------

# A. Why does model.coef_ return three coefficients?
# Answer:
# Because the model has three input features:
# Area, Rooms, and Age.
# Each feature has its own learned coefficient.

# B. Why do we use [[80, 2, 5]]?
# Answer:
# scikit-learn expects a 2D input:
# (number_of_samples, number_of_features).
# Here, we have one house and three features.

# C. What happens if we change the feature order?
# Answer:
# The model associates each coefficient with the feature
# in the position used during training.
# Changing the order can produce an incorrect prediction
# because the model interprets each value as a different feature.
# Keep the same feature order during training and prediction.

# D. Do positive and negative coefficients prove causation?
# Answer:
# No. Coefficients describe the relationships learned by the model,
# assuming the other features remain constant.
# They do not, by themselves, prove that a feature causes
# the price to increase or decrease.
````
