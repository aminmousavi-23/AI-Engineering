##### <mark>Exercise 1 — Mathematical Transformations</mark>

Create:

x = np.array([-4, -1, 0, 1, 4, 9], dtype=np.float32)

Apply:

1. Absolute value
2. Square
3. Square root of the absolute values
4. Round the square-root result to 2 decimal places

Print every result.

````python
x = np.array([-4, -1, 0, 1, 4, 9], dtype=np.float32)

absolute_value = np.abs(x)
square_value = np.square(x)
sqrt_value_from_abs = np.sqrt(absolute_value)
round_value_from_square_root = np.round(sqrt_value_from_abs, 2)

print(absolute_value)
print(square_value)
print(sqrt_value_from_abs)
print(round_value_from_square_root)
````

##### <mark>Exercise 2 — Prediction Error</mark>

Given:
```text
predictions = np.array([0.8, 0.6, 0.9, 0.4])
targets = np.array([1.0, 0.5, 1.0, 0.2])

Use NumPy mathematical operations to calculate:

1. Error
2. Absolute error
3. Squared error
4. Square root of the squared error

Then verify that:

sqrt(squared_error) == absolute_error

````python
predictions = np.array([0.8, 0.6, 0.9, 0.4])
targets = np.array([1.0, 0.5, 1.0, 0.2])

error = predictions - targets
absolute_error = np.abs(error)
squared_error = np.square(error)
sqrt_of_squared_error = np.sqrt(squared_error)

print(error)
print(absolute_error)
print(squared_error)
print(sqrt_of_squared_error)
print(np.sqrt(squared_error) == absolute_error)
````

##### <mark>Exercise 3 — AI Score Transformation</mark>

Create:

scores = np.array([0.5, 1.0, 1.5, 2.0])

Calculate:

1. np.exp(scores)
2. np.log(scores)
3. np.square(scores)
4. np.sqrt(scores)

Print each result and briefly explain what mathematical transformation was applied.

````python
scores = np.array([0.5, 1.0, 1.5, 2.0])

exp_scores = np.exp(scores) #e²
log_scores = np.log(scores) #ln(2)
square_scores = np.square(scores) #2² = 4
sqrt_scores = np.sqrt(scores) #√2

print(exp_scores) 
print(log_scores)
print(square_scores)
print(sqrt_scores)
````
