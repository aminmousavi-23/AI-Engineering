##### <mark>Exercise 1 — Dataset Reshaping</mark>

Create:

x = np.arange(60)

Reshape it into:

- 10 samples
- 6 features

Then:

1. Print the original shape.
2. Print the new shape.
3. Print the reshaped array.
4. Verify that the number of elements did not change.

```python
x = np.arange(60)
print(x.shape)
print(x.size)

x = x.reshape(10, 6)
print(x.shape)
print(x.size)
```

##### <mark>Exercise 2 — Flatten Image Batch</mark>

Create a grayscale image batch:

images = np.zeros((32, 28, 28), dtype=np.float32)

Reshape it into:

(samples, features)

Requirements:

1. Keep the number of samples unchanged.
2. Use -1 for the feature dimension.
3. Print the final shape.
4. Explain why the final shape is (32, 784).

```python
images = np.zeros((32, 28, 28), dtype=np.float32)

images = images.reshape(32, -1)

print(images.shape)
# the shape is (32, 784) because 28 * 28 is 784
```

##### <mark>Exercise 3 — RGB Image Preprocessing</mark>

Create:

images = np.zeros((16, 64, 64, 3), dtype=np.float32)

Reshape it into a 2D ML feature matrix.

Requirements:

1. Keep 16 samples.
2. Use -1 for the feature dimension.
3. Print the original shape.
4. Print the new shape.
5. Calculate how many features each image contains.
6. Verify that size is unchanged.

```python
images = np.zeros((16, 64, 64, 3), dtype=np.float32)
print(images.shape)
print(images.size)

images = images.reshape(16, -1)
print(images.shape)
print(images.shape[1])
print(images.size)
```
