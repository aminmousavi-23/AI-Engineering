##### <mark>Exercise 1 — Sensor Data Cleaning</mark>
Create an array of sensor temperature readings that includes some faulty values (e.g., -999 representing sensor errors):
```text
temps = np.array([22.5, -999, 25.1, 23.8, -999, 21.9, 50.2, 24.0])

Filter out only the valid readings (between 0 and 45 degrees) using Boolean Indexing,
 and compute the mean of the actual valid temperatures.

````python
temps = np.array([22.5, -999, 25.1, 23.8, -999, 21.9, 50.2, 24.0])

valid_temp_range = (45 >= temps) & (temps >= 0)

valid_temps = temps[valid_temp_range]

print(valid_temps)
print(np.mean(valid_temps))
````

##### <mark>Exercise 2 — Filtering Dataset Samples by Multiple Conditions</mark>
```text
Suppose you have an array of student scores (0 to 100). 
Extract the students whose scores fall between 60 and 80 
    (i.e., not failing, not excellent — the "needs attention" group),
and print how many students are in that group.

````python
scores = np.random.randint(0, 101, size=8)

need_attention_condition = (scores >= 60) & (scores <= 80)

need_attention_scores = scores[need_attention_condition]

print(scores)
print(need_attention_scores)

print("Number of students:", need_attention_scores.size)
````

##### <mark>Exercise 3 — Simulating Model Confidence Filtering</mark>
```text
Create an array of 20 random numbers between 0 and 1 (representing a classification model's predicted probabilities):

preds = np.random.rand(20)

Keep only the predictions where confidence > 0.7 or confidence < 0.3 (i.e., the model is "confident" about them), 
and discard the uncertain range (0.3 to 0.7). 
Report the count and percentage of "confident" samples.

````python
preds = np.random.rand(20)

confidence_condition = (preds > 0.7) | (preds < 0.3)

confidence_preds = preds[confidence_condition]

confident_count = confidence_preds.size
confident_percentage = confident_count / preds.size * 100

print("Confident samples:", confident_count)
print("Confident percentage:", confident_percentage)
````
