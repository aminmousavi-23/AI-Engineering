<mark>Exercise 1 — Dataset Inspection</mark>

Create a NumPy array representing:

- 5000 samples
- 30 features
- float32 data type

Print:

1. shape
2. ndim
3. size
4. dtype
5. itemsize
6. nbytes

Then explain what each value means.

```python
array = np.zeros((5000, 30), dtype=np.float32)

print(array.shape) # element in each dimention
print(array.ndim) # number of dimentions
print(array.size) # all element's count
print(array.dtype) # array element's type
print(array.itemsize) # memeory size which each element take in bytes
print(array.nbytes) # memeory size which all elements take in bytes
```
