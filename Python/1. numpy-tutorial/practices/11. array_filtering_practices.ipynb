##### <mark>Exercise 1 — Model Confidence Filtering</mark>

Create an array of 20 random confidence scores between 0 and 1.

Filter and print:

1. Scores greater than or equal to 0.8
2. Scores lower than 0.3
3. Scores between 0.4 and 0.7
4. The number of elements in each filtered result

````python
scores = np.random.random(20)

greater_equal_08 = scores[scores >= 0.8]
lower_03 = scores[scores < 0.3]
between_04_07 = scores[(scores >= 0.4) & (scores <= 0.7)]

print(f"All Scores: {scores}")

print(f"Scores >= 0.8: {greater_equal_08}")
print(f"Count: {greater_equal_08.size}")

print(f"Scores < 0.3: {lower_03}")
print(f"Count: {lower_03.size}")

print(f"Scores between 0.4 and 0.7: {between_04_07}")
print(f"Count: {between_04_07.size}")
````

##### <mark>Exercise 2 — Dataset Sample Filtering</mark>
```text
Create this dataset:

x = np.array([
    [25, 180, 75],
    [31, 175, 82],
    [22, 165, 60],
    [40, 190, 95],
    [28, 172, 68],
    [35, 185, 88]
])

Columns represent:

Age, Height, Weight

Filter and print:

1. Samples with weight greater than 80
2. Samples with height greater than 175
3. Samples with weight between 65 and 90
4. Samples where height is greater than 175 and weight is greater than 80

````python
import numpy as np

x = np.array([
    [25, 180, 75],
    [31, 175, 82],
    [22, 165, 60],
    [40, 190, 95],
    [28, 172, 68],
    [35, 185, 88]
])

print("1. Samples with weight greater than 80:")
print(x[x[:, 2] > 80])

print("\n2. Samples with height greater than 175:")
print(x[x[:, 1] > 175])

print("\n3. Samples with weight between 65 and 90:")
print(x[(x[:, 2] >= 65) & (x[:, 2] <= 90)])

print("\n4. Samples with height greater than 175 and weight greater than 80:")
print(x[(x[:, 1] > 175) & (x[:, 2] > 80)])
````

##### <mark>Exercise 3 — Classification Result Filtering</mark>
```text
Create:

predictions = np.array([
    0, 1, 2, 1, 0, 2, 2, 1, 0, 2
])

confidence = np.array([
    0.95, 0.62, 0.88, 0.41, 0.91,
    0.77, 0.93, 0.55, 0.84, 0.68
])

Find:

1. All predictions with confidence >= 0.8
2. The corresponding prediction classes
3. The indices of these high-confidence predictions using np.where
4. All predictions belonging to class 2
5. All class 2 predictions with confidence >= 0.8

````python
predictions = np.array([
    0, 1, 2, 1, 0, 2, 2, 1, 0, 2
])

confidence = np.array([
    0.95, 0.62, 0.88, 0.41, 0.91,
    0.77, 0.93, 0.55, 0.84, 0.68
])

# 1. All predictions with confidence >= 0.8
high_confidence = confidence >= 0.8

print("1. High-confidence scores:")
print(confidence[high_confidence])


# 2. Corresponding prediction classes
print("\n2. Corresponding prediction classes:")
print(predictions[high_confidence])


# 3. Indices of high-confidence predictions
high_confidence_indices = np.where(high_confidence)[0]

print("\n3. Indices of high-confidence predictions:")
print(high_confidence_indices)


# 4. All predictions belonging to class 2
class_2 = predictions == 2

print("\n4. All class 2 predictions:")
print(predictions[class_2])


# 5. Class 2 predictions with confidence >= 0.8
class_2_high_confidence = (predictions == 2) & (confidence >= 0.8)

print("\n5. Class 2 predictions with confidence >= 0.8:")
print(predictions[class_2_high_confidence])
````
