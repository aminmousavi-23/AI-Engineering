##### <mark>Exercise 1 — Feature Normalization</mark>
```text
You have a dataset of 5 samples with 3 features each:

data = np.array([[10, 200, 3000],
                  [12, 220, 2800],
                  [ 9, 195, 3100],
                  [15, 240, 2950],
                  [11, 210, 3050]])

Compute the mean and standard deviation of each feature (column-wise, so axis=0), 
then use broadcasting to normalize the entire dataset (subtract the mean and divide by the std) in a single expression — no loops.

````python
data = np.array([[10, 200, 3000],
                  [12, 220, 2800],
                  [ 9, 195, 3100],
                  [15, 240, 2950],
                  [11, 210, 3050]])

mean = np.mean(data, axis=0)
std = np.std(data, axis=0)

print(f"Mean: {mean}")
print(f"Standard Deviation: {std}")
print(f"Normalized: {(data - mean) / std}")
````

Simulate the output of a linear layer before adding bias, for a batch of 4 samples with 3 output neurons:

```text
layer_output = np.array([[1.2, 0.5, -0.3],

bias = np.array([0.1, -0.05, 0.2])              # shape (3,)

##### <mark>Exercise 3 — Distance from Multiple Centroids (K-Means style)</mark> ❌Need practice later
You have 4 data points (2D) and 2 cluster centroids:
```text
points = np.array([[1, 2], [3, 4], [5, 1], [0, 0]])       # shape (4, 2)
centroids = np.array([[2, 2], [4, 4]])                      # shape (2, 2)

Using broadcasting (hint: you'll need to reshape one array to add a new axis, e.g., points[:, np.newaxis, :]), 
compute the Euclidean distance from every point to every centroid, 
resulting in a distance matrix of shape (4, 2). 
Then determine which centroid each point is closest to.

````python
points = np.array([[1, 2], [3, 4], [5, 1], [0, 0]])    # shape (4, 2)
centroids = np.array([[2, 2], [4, 4]])                 # shape (2, 2)

# (4, 1, 2) - (2, 2) -> broadcasts to (4, 2, 2)
diff = points[:, np.newaxis, :] - centroids

# Euclidean distance matrix, shape (4, 2)
distances = np.sqrt((diff ** 2).sum(axis=2))

# Index of the closest centroid for each point
closest = np.argmin(distances, axis=1)

print(distances.round(3))
print(distances.shape)
print(closest)
````
