**NumPy Performance Practices**

```python
import numpy as np
```

| تکنیک            | کار                                | مثال                                          |
| ---------------- | ---------------------------------- | --------------------------------------------- |
| Vectorization    | حذف loopهای غیرضروری Python        | `x * 2`                                       |
| Aggregation      | اجرای سریع عملیات روی کل Array     | `x.mean()`                                    |
| Broadcasting     | عملیات روی ابعاد سازگار            | `X - mean`                                    |
| مناسب بودن dtype | کاهش Memory                        | `float32` به‌جای `float64` در صورت مناسب بودن |
| View             | جلوگیری از Copy غیرضروری           | `x[:1000]`                                    |
| Copy             | ایجاد داده مستقل                   | `x[:1000].copy()`                             |
| In-place         | کاهش allocation                    | `x *= 2`                                      |
| `np.empty()`     | جلوگیری از initialization غیرضروری | `np.empty(shape)`                             |
| کاهش Dataset     | پردازش فقط داده موردنیاز           | `x[:1000]`                                    |
| Benchmark        | اندازه‌گیری واقعی performance      | `time.perf_counter()`                         |
| `timeit`         | benchmark دقیق‌تر برای عملیات کوچک | `timeit.timeit(...)`                          |
