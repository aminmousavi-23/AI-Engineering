**Array Creation Practices**

- np.array() -> Convert existing data → ndarray

- np.zeros() -> All zeros

- np.ones() -> All ones

- np.full() -> All same value

- np.empty() -> Uninitialized array

- np.arange() -> Sequence based on step

- np.linspace() -> Fixed number of evenly spaced values

````python
import numpy as np
````
