Exercise 1 — Dataset Features

Create a NumPy array representing the following dataset:

Age    Height    Weight

25     180       75

30     175       80

22     168       65

35     182       90

Then:

1. Print the array.
2. Print the type of the array.
3. Determine the number of samples.
4. Determine the number of features.

```python
arr = np.array([[25 ,180 ,75],
                    [30 ,175 ,80],
                    [22 ,168 ,65],
                    [35 ,182 ,90]]
                )

print(f"Array : {arr}")
print(f"Array type : {str(arr.dtype)}")
print("Samples:", arr.shape[0])
print("Features:", arr.shape[1])
```
