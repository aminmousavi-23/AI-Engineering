**Aggregation Practices**

````python
import numpy as np
````

| کد NumPy       | نام فارسی                   | کاربرد                                                     |
| -------------- | --------------------------- | ---------------------------------------------------------- |
| `np.sum(x)`    | **مجموع**                   | جمع تمام عناصر آرایه                                       |
| `np.mean(x)`   | **میانگین**                 | محاسبه میانگین عناصر                                       |
| `np.min(x)`    | **کمینه / حداقل**           | پیدا کردن کوچک‌ترین مقدار                                  |
| `np.max(x)`    | **بیشینه / حداکثر**         | پیدا کردن بزرگ‌ترین مقدار                                  |
| `np.std(x)`    | **انحراف معیار**            | اندازه‌گیری میزان پراکندگی داده‌ها نسبت به میانگین         |
| `np.var(x)`    | **واریانس**                 | اندازه‌گیری پراکندگی داده‌ها بر اساس مربع فاصله از میانگین |
| `np.median(x)` | **میانه**                   | مقدار وسط داده‌ها پس از مرتب‌سازی                          |
| `np.argmin(x)` | **اندیس کمینه**             | پیدا کردن **اندیس** کوچک‌ترین مقدار                        |
| `np.argmax(x)` | **اندیس بیشینه**            | پیدا کردن **اندیس** بزرگ‌ترین مقدار                        |
| `np.any(x)`    | **آیا حداقل یکی درست است؟** | اگر حداقل یک عنصر `True` یا غیرصفر باشد → `True`           |
| `np.all(x)`    | **آیا همه درست هستند؟**     | اگر تمام عناصر `True` یا غیرصفر باشند → `True`             |

##### <mark>Exercise 1 — Model Performance Summary</mark>

A classification model produced the following confidence scores:
```text
confidence = np.array([
    0.91, 0.72, 0.88, 0.45,
    0.97, 0.63, 0.81, 0.55
])

Calculate and print:

1. Total confidence
2. Mean confidence
3. Minimum confidence
4. Maximum confidence
5. Median confidence
6. Standard deviation
7. Index of the lowest confidence
8. Index of the highest confidence

````python
confidence = np.array([
    0.91, 0.72, 0.88, 0.45,
    0.97, 0.63, 0.81, 0.55
])

print(np.sum(confidence))
print(np.mean(confidence))
print(np.min(confidence))
print(np.max(confidence))
print(np.max(confidence))
print(np.median(confidence))
print(np.std(confidence))
print(np.argmin(confidence))
print(np.argmax(confidence))
````

##### <mark>Exercise 2 — Dataset Quality Check</mark>

Create:
```text
feature = np.array([
    10, 12, 11, 13, 10,
    12, 11, 500
], dtype=np.float32)

This represents one numerical feature from a dataset.

Calculate:

1. Mean
2. Median
3. Standard deviation
4. Minimum
5. Maximum

Then determine whether the mean and median tell a similar story.

Goal: Understand how an extreme value can affect statistical aggregation.

````python
feature = np.array([
    10, 12, 11, 13, 10,
    12, 11, 500
], dtype=np.float32)

print(np.mean(feature))
print(np.median(feature))
print(np.std(feature))
print(np.min(feature))
print(np.max(feature))
````

##### <mark>Exercise 3 — Classification Output Validation</mark>

A model outputs probabilities for 5 classes:
```text
probabilities = np.array([
    0.05,
    0.15,
    0.60,
    0.10,
    0.10
])

Assume:

0 → Cat
1 → Dog
2 → Bird
3 → Car
4 → Horse

Write code to:

1. Find the predicted class using argmax.
2. Find the highest probability using max.
3. Calculate the sum of all probabilities.
4. Verify that the probabilities sum to approximately 1.0.
5. Use np.any() to check whether any class has probability greater than 0.5.
6. Use np.all() to check whether every class has probability greater than 0.01.

````python
probabilities = np.array([
    0.05,
    0.15,
    0.60,
    0.10,
    0.10
])

print(np.argmax(probabilities))
print(np.max(probabilities))
print(np.sum(probabilities))
print(np.sum(probabilities) == 1)
print(np.any(probabilities > 0.5))
print(np.all(probabilities > 0.01))
````
