##### <mark>Exercise 1 — Merging Data Batches</mark>
```text
You receive two separate batches of feature data collected at different times:

batch1 = np.array([[1, 2, 3], [4, 5, 6]])
batch2 = np.array([[7, 8, 9], [10, 11, 12]])

Combine them into a single dataset with all rows stacked together (resulting shape should be (4, 3)). 
Then combine them side-by-side instead (resulting shape should be (2, 6)), 
simulating a case where batch2 contains additional features for the same samples.

````python
batch1 = np.array([[1, 2, 3], [4, 5, 6]])
batch2 = np.array([[7, 8, 9], [10, 11, 12]])

stacked = np.vstack((batch1, batch2))
print(stacked)
print(stacked.shape)    

side_by_side = np.hstack((batch1, batch2))
print(side_by_side)
print(side_by_side.shape)   
````

##### <mark>Exercise 3 — Ensemble Model Predictions</mark>
Three different models produce probability predictions for the same 5 samples:
```text
model1_preds = np.array([0.2, 0.8, 0.6, 0.3, 0.9])
model2_preds = np.array([0.3, 0.7, 0.5, 0.4, 0.85])
model3_preds = np.array([0.25, 0.75, 0.55, 0.35, 0.95])

Combine these three arrays into a single 2D array of shape (3, 5), 
then compute the average prediction per sample (averaging across the models, i.e., along axis=0)
to get the final ensemble prediction for the 5 samples.

````python
model1_preds = np.array([0.2, 0.8, 0.6, 0.3, 0.9])
model2_preds = np.array([0.3, 0.7, 0.5, 0.4, 0.85])
model3_preds = np.array([0.25, 0.75, 0.55, 0.35, 0.95])

all_preds = np.vstack((model1_preds, model2_preds, model3_preds))
print(all_preds.shape)  

ensemble = all_preds.mean(axis=0)
print(ensemble)            
print(ensemble.shape)
````
