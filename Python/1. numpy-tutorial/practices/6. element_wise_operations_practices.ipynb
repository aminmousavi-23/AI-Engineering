##### <mark>Exercise 1 — Feature Scaling</mark>

Create:

```text
features = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
], dtype=np.float32)

Perform:

1. Add 10 to every element.
2. Subtract 5 from every element.
3. Multiply every element by 2.
4. Divide every element by 10.

Print each result.

````python
features = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
], dtype=np.float32)

print(features + 10)
print(features - 5)
print(features * 2)
print(features / 10)
````

##### <mark>Exercise 2 — Image Normalization</mark>

Create a synthetic grayscale image:

```text
image = np.array([
    [0, 64, 128],
    [192, 224, 255]
], dtype=np.float32)

Normalize the image to approximately [0, 1] using element-wise operations.

Requirements:

1. Divide all values by 255.0.
2. Print the normalized image.
3. Print its dtype.
4. Verify that the maximum value is 1.0.

````python
image = np.array([
    [0, 64, 128],
    [192, 224, 255]
], dtype=np.float32)

image = image / 255.0
print(image)
print(image.dtype)
````

##### <mark>Exercise 3 — Prediction Error</mark>

Suppose a model produces these predictions:

predictions = np.array([0.8, 0.6, 0.9, 0.4])

and the actual values are:

targets = np.array([1.0, 0.5, 1.0, 0.2])

Use element-wise operations to:

1. Calculate the prediction error.
2. Calculate the absolute error.
3. Calculate the squared error.
4. Print all three results.

Hint:

error = predictions - targets

````python
predictions = np.array([0.8, 0.6, 0.9, 0.4])
targets = np.array([1.0, 0.5, 1.0, 0.2])

error = predictions - targets
absolute_error = np.abs(error)
squared_error = error ** 2

print(error)
print(absolute_error)
print(squared_error)
````
