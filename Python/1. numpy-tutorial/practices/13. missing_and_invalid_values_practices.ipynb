##### <mark>Exercise 1 — Cleaning a Dataset with Missing Values</mark>
```text
Create an array simulating sensor readings with missing values:

data = np.array([23.5, np.nan, 21.0, np.nan, 25.3, 19.8, np.nan, 22.1])

1. Count how many missing values exist, 
2. Compute the mean of the valid (non-missing) values using two different approaches:
    (1) filtering with Boolean Indexing 
    (2) using np.nanmean()

Confirm both give the same result.

````python
data = np.array([23.5, np.nan, 21.0, np.nan, 25.3, 19.8, np.nan, 22.1])

mask = np.isnan(data)
n_missing = mask.sum()        
print("Missing values:", n_missing)               

valid = data[~mask]             
mean_bool = valid.mean()
print("Mean (boolean indexing):", mean_bool)      

mean_nan = np.nanmean(data)
print("Mean (np.nanmean):", mean_nan)               

print("Same result?", np.isclose(mean_bool, mean_nan)) 
````

##### <mark>Exercise 3 — Detecting Invalid Model Outputs</mark>
Simulate the output of an unstable training process:
```text
losses = np.array([0.45, 0.38, np.nan, 0.31, np.inf, 0.28, np.nan, 0.25])

Write code that detects whether the training run contains any invalid values (nan or inf),
reports at which indices they occurred,
and computes the "clean" average loss ignoring those invalid entries.

````python
losses = np.array([0.45, 0.38, np.nan, 0.31, np.inf, 0.28, np.nan, 0.25])

invalid_mask = np.isnan(losses) | np.isinf(losses)
has_invalid = np.any(invalid_mask)

print("Contains invalid values:", has_invalid)

invalid_indices = np.where(invalid_mask)[0]
print("Invalid indices:", invalid_indices)

clean_avg = np.mean(losses[~invalid_mask])
print("Clean average loss:", clean_avg)
````
