**Concat Practices**

```
# Vertical
pd.concat([df1, df2])

# Vertical + new index
pd.concat([df1, df2], ignore_index=True)

# Horizontal
pd.concat([df1, df2], axis=1)

# Only common labels on other axis
pd.concat([df1, df2], join="inner")

# Track source
pd.concat([df1, df2], keys=["A", "B"])

# Check duplicate index
pd.concat([df1, df2], verify_integrity=True)
```

##### Exercise 1 — Combine Training Data Batches
```text
Create two DataFrames representing two batches of an image-classification dataset:

batch_1:
image_id, label, confidence

batch_2:
image_id, label, confidence

Perform the following:

1. Concatenate the two DataFrames vertically.
2. Reset the index using ignore_index=True.
3. Add a third batch with the same columns.
4. Concatenate all three batches together.
5. Use keys to keep track of which batch each row came from.
6. Print the final DataFrame and its index.

````python
batch_1 = pd.DataFrame({
    "image_id": ["img_001", "img_002", "img_003"],
    "label": ["cat", "dog", "bird"],
    "confidence": [0.92, 0.85, 0.76]
})

batch_2 = pd.DataFrame({
    "image_id": ["img_004", "img_005", "img_006"],
    "label": ["cat", "dog", "bird"],
    "confidence": [0.88, 0.91, 0.79]
})


# 1. Concatenate two DataFrames vertically
combined = pd.concat([batch_1, batch_2])

print(combined)


# 2. Reset the index
combined = pd.concat(
    [batch_1, batch_2],
    ignore_index=True
)

print(combined)


# 3. Add a third batch
batch_3 = pd.DataFrame({
    "image_id": ["img_007", "img_008", "img_009"],
    "label": ["cat", "dog", "bird"],
    "confidence": [0.95, 0.73, 0.89]
})


# 4. Concatenate all three batches
all_batches = pd.concat(
    [batch_1, batch_2, batch_3],
    ignore_index=True
)

print(all_batches)


# 5. Use keys to track the source batch
all_batches_with_keys = pd.concat(
    [batch_1, batch_2, batch_3],
    keys=["A", "B", "C"]
)

# 6. Print final DataFrame and its index
print(all_batches_with_keys)
print(all_batches_with_keys.index)
````
