**Apply and Map Practices**

```
# Element by element
df["score"].map(lambda x: x * 100)

# Column-wise
df.apply(lambda col: col.mean())

# Row-wise
df.apply(lambda row: ..., axis=1)

# Every cell
df.map(lambda x: ...)
```

| Method              | معمولاً روی چه چیزی؟ | کاربرد                       |
| ------------------- | -------------------- | ---------------------------- |
| `Series.map()`      | هر عنصر Series       | mapping / تبدیل element-wise |
| `Series.apply()`    | هر عنصر Series       | اجرای function               |
| `DataFrame.apply()` | row یا column        | محاسبات روی یک ردیف/ستون     |
| `DataFrame.map()`   | هر cell              | element-wise transformation  |

##### Exercise 1 — Analyze Model Performance
```text
Create:

results = pd.DataFrame({
    "model": ["A", "A", "B", "B"],
    "accuracy": [0.91, 0.87, 0.94, 0.89],
    "loss": [0.12, 0.18, 0.09, 0.15]
})

Perform the following:

1. Use apply() to calculate the mean of every numeric column.
2. Use apply(axis=1) to create a performance_score column using:
    performance_score = accuracy - loss
3. Use map() to convert the model names:
    A → "Baseline"
    B → "Improved"
4. Create a new model_type column.
5. Print the final DataFrame and the calculated column means.

````python
results = pd.DataFrame({
    "model": ["A", "A", "B", "B"],
    "accuracy": [0.91, 0.87, 0.94, 0.89],
    "loss": [0.12, 0.18, 0.09, 0.15]
})


# 1. Use apply() to calculate the mean of every numeric column
column_means = results[["accuracy", "loss"]].apply(
    lambda column: column.mean()
)

print(column_means)


# 2. Use apply(axis=1) to create performance_score
results["performance_score"] = results.apply(
    lambda row: row["accuracy"] - row["loss"],
    axis=1
)


# 3. Use map() to convert model names
results["model_type"] = results["model"].map({
    "A": "Baseline",
    "B": "Improved"
})


# 4. model_type has already been created above


# 5. Print final DataFrame and calculated means
print(results)
print(column_means)
````

##### Exercise 2 — Transform Model Predictions
```text
Create:

predictions = pd.DataFrame({
    "image_id": ["img_001", "img_002", "img_003", "img_004"],
    "label": ["cat", "dog", "bird", "cat"],
    "confidence": [0.92, 0.67, 0.81, 0.45]
})

Perform the following:

1. Use map() to convert label into numeric IDs:
    cat → 0
    dog → 1
    bird → 2
2. Create a confidence_percent column using map().
3. Create a new column called confidence_level.
4. Use apply() with a function:
    "high" if confidence >= 0.80
    "medium" if confidence >= 0.60
    "low" otherwise.
5. Print the final DataFrame.

````python
predictions = pd.DataFrame({
    "image_id": ["img_001", "img_002", "img_003", "img_004"],
    "label": ["cat", "dog", "bird", "cat"],
    "confidence": [0.92, 0.67, 0.81, 0.45]
})


# 1. Convert labels into numeric IDs
predictions["label"] = predictions["label"].map({
    "cat": 0,
    "dog": 1,
    "bird": 2
})


# 2. Create confidence_percent
predictions["confidence_percent"] = predictions["confidence"].map(
    lambda x: x * 100
)


# 3 & 4. Create confidence_level
predictions["confidence_level"] = predictions["confidence"].apply(
    lambda x: (
        "high" if x >= 0.80
        else "medium" if x >= 0.60
        else "low"
    )
)


# 5. Print final DataFrame
print(predictions)
````
