##### Exercise 1 — Employee Duplicates
```text
Create:

df = pd.DataFrame({
    "employee_id": [101, 102, 103, 101, 104, 102, 105],
    "name": ["Alice", "Bob", "Charlie", "Alice", "Diana", "Bob", "Ethan"],
    "department": ["AI", "Backend", "AI", "AI", "DevOps", "Backend", "AI"],
    "salary": [5500, 7200, 6800, 5500, 9000, 7200, 6400]
})

Perform the following:

1. Check which complete rows are duplicated.
2. Count the number of duplicate rows.
3. Display only the duplicate rows.
4. Check duplicates based only on employee_id.
5. Remove duplicate employees based on employee_id.
6. Create another DataFrame where all occurrences of duplicated employee_ids are removed.
7. Compare the number of rows before and after removing duplicates.

````python
df = pd.DataFrame({
    "employee_id": [101, 102, 103, 101, 104, 102, 105],
    "name": ["Alice", "Bob", "Charlie", "Alice", "Diana", "Bob", "Ethan"],
    "department": ["AI", "Backend", "AI", "AI", "DevOps", "Backend", "AI"],
    "salary": [5500, 7200, 6800, 5500, 9000, 7200, 6400]
})

# 1. Check which complete rows are duplicated
print(df.duplicated())

# 2. Count the number of duplicate rows
print(df.duplicated().sum())

# 3. Display only the duplicate rows
print(df[df.duplicated()])

# 4. Check duplicates based only on employee_id
print(df.duplicated(subset=["employee_id"]))

# 5. Remove duplicate employees based on employee_id
new_df = df.drop_duplicates(subset=["employee_id"])

print(new_df)

# 6. Remove ALL occurrences of duplicated employee_ids
unique_employees = df[~df.duplicated(
    subset=["employee_id"],
    keep=False
)]

print(unique_employees)

# 7. Compare number of rows before and after removing duplicates
print("Before:", len(df))
print("After:", len(new_df))
print("Rows removed:", len(df) - len(new_df))
````

##### Exercise 2 — ML Prediction Duplicates
```text
Create:

predictions = pd.DataFrame({
    "image_id": [
        "img_001", "img_002", "img_003",
        "img_001", "img_004", "img_002"
    ],
    "label": [
        "cat", "dog", "car",
        "cat", "bird", "dog"
    ],
    "confidence": [
        0.82, 0.91, 0.76,
        0.95, 0.88, 0.85
    ]
})

Perform the following:

1. Find duplicate image_ids.
2. Display all rows belonging to duplicated image_ids.
3. Count the number of duplicated image_ids.
4. Sort the DataFrame by confidence in descending order.
5. Remove duplicate image_ids while keeping the highest-confidence prediction.
6. Verify that every image_id is now unique.
7. Print the final DataFrame.

````python
predictions = pd.DataFrame({
    "image_id": [
        "img_001", "img_002", "img_003",
        "img_001", "img_004", "img_002"
    ],
    "label": [
        "cat", "dog", "car",
        "cat", "bird", "dog"
    ],
    "confidence": [
        0.82, 0.91, 0.76,
        0.95, 0.88, 0.85
    ]
})

# 1. Find duplicate image_ids
print(predictions.duplicated(subset=["image_id"]))

# 2. Display all rows belonging to duplicated image_ids
print(
    predictions[
        predictions.duplicated(subset=["image_id"], keep=False)
    ]
)

# 3. Count the number of duplicated image_ids
print(
    predictions.duplicated(subset=["image_id"]).sum()
)

# 4. Sort by confidence in descending order
predictions = predictions.sort_values(
    by="confidence",
    ascending=False
)

print(predictions)

# 5. Remove duplicate image_ids,
#    keeping the highest-confidence prediction
predictions = predictions.drop_duplicates(
    subset=["image_id"]
)

print(predictions)

# 6. Verify that every image_id is unique
print(predictions["image_id"].is_unique)
````
