````python
experiments = pd.DataFrame({
    "experiment_id": [
        "exp_003",
        "exp_001",
        "exp_002",
        "exp_004"
    ],
    "model": [
        "ResNet",
        "CNN",
        "ViT",
        "CNN"
    ],
    "dataset": [
        "CIFAR10",
        "MNIST",
        "CIFAR10",
        "MNIST"
    ],
    "accuracy": [
        0.94,
        0.91,
        0.96,
        0.93
    ]
})

experiments = experiments.set_index("experiment_id")

experiments
````

````python
experiments.index
````

````python
experiments.index.is_unique
````

````python
experiments = experiments.sort_index()

experiments
````

````python
new_order = [
    "exp_004",
    "exp_002",
    "exp_001",
    "exp_003"
]

experiments = experiments.reindex(new_order)

experiments
````

````python
experiments = experiments.reset_index()

experiments
````

````python
experiments = experiments.set_index(["model", "dataset"])

experiments
````

````python
experiments = experiments.sort_index()

experiments
````

````python
experiments = experiments.reset_index()

experiments
````

````python
# set_index()
# Converts an existing column into the DataFrame's index.

# reindex()
# Changes the existing index to a specified set/order of labels.
````
