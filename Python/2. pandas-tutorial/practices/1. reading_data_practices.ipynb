##### Exercise 1 — Reading a Machine Learning Dataset
```text
You have a CSV file named model_predictions.csv with the following content:

image_id,label,confidence,model_version
IMG001,cat,0.95,v1
IMG002,dog,0.87,v1
IMG003,cat,0.91,v2
IMG004,bird,0.78,v2
IMG005,dog,0.93,v2

Write Python code to:

1. Read the CSV file into a Pandas DataFrame.
2. Print the DataFrame.
3. Use image_id as the DataFrame index while reading the file.
4. Load only the label and confidence columns into another DataFrame.
5. Print the columns and index of both DataFrames.
6. Calculate the average confidence of the loaded predictions.

````python
file_path = r"E:\AI-Engineering\Python\2. pandas-tutorial\data\model_predictions.csv"

# 1-3. Read the complete CSV and use image_id as index
df = pd.read_csv(
    file_path,
    index_col="image_id"
)

# 2. Print the DataFrame
print("Full DataFrame:")
print(df)

# 4. Load only label and confidence
df_predictions = pd.read_csv(
    file_path,
    index_col="image_id",
    usecols=["image_id", "label", "confidence"]
)

print("\nPredictions DataFrame:")
print(df_predictions)

# 5. Print columns and index of both DataFrames
print("\nFull DataFrame columns:")
print(df.columns)

print("Full DataFrame index:")
print(df.index)

print("\nPredictions DataFrame columns:")
print(df_predictions.columns)

print("Predictions DataFrame index:")
print(df_predictions.index)

# 6. Calculate average confidence
average_confidence = df_predictions["confidence"].mean()

print("\nAverage confidence:", average_confidence)
````

##### Exercise 2 — Reading Different CSV Formats
```text
You receive a file named students.csv with the following content:

Ali;25;85
Sara;28;92
Reza;23;76
Mina;31;95

The file:

has no header
uses ; as the separator

Write Python code to:

1. Read the file using pd.read_csv().
2. Correctly specify the separator.
3. Tell Pandas that the file has no header.
4. Assign these column names:
    name
    age
    score
5. Print the resulting DataFrame.
6. Print its dtypes.
7. Calculate the average score.
8. Create another DataFrame containing only the name and score columns.

````python
file_path = r"..\data\students.csv"

df = pd.read_csv(file_path, header=None, sep=";", names=["name", "age", "score"])

df2 = pd.read_csv(file_path, header=None, sep=";", names=["name", "age", "score"], usecols=["name", "score"])

print(df)
print("---------------")
print(df.dtypes)
print("---------------")
print(df["score"].mean())
print("---------------")
print(df2)
````
