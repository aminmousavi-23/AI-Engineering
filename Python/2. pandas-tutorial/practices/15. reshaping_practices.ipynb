**Reshaping Practices**

| ابزار           | کاربرد                      |
| --------------- | --------------------------- |
| `pivot()`       | Long → Wide                 |
| `pivot_table()` | Long → Wide + Aggregation   |
| `melt()`        | Wide → Long                 |
| `stack()`       | Columns → Index             |
| `unstack()`     | Index → Columns             |
| Reshaping       | مفهوم کلی تغییر ساختار داده |

````python
results = pd.DataFrame({
    "image_id": [
        "img_001", "img_001",
        "img_002", "img_002",
        "img_003", "img_003",
        "img_003"
    ],
    "model": [
        "CNN", "ResNet",
        "CNN", "ResNet",
        "CNN", "CNN",
        "ResNet"
    ],
    "confidence": [
        0.91, 0.95,
        0.82, 0.88,
        0.79, 0.85,
        0.92
    ]
})

print(results)


# ============================================================
# Part 1 — Pivot
# ============================================================

try:
    wide_results = results.pivot(
        index="image_id",
        columns="model",
        values="confidence"
    )

    print(wide_results)

except ValueError as e:
    print("pivot() failed:")
    print(e)

# pivot() cannot handle duplicate combinations of:
# image_id + model
#
# img_003 + CNN appears twice:
# 0.79
# 0.85
#
# Therefore, pivot() does not know which value to place
# in the single img_003 / CNN cell.


# ============================================================
# Part 2 — Pivot Table
# ============================================================

wide_results_mean = results.pivot_table(
    index="image_id",
    columns="model",
    values="confidence",
    aggfunc="mean"
)

print("\nMean:")
print(wide_results_mean)


wide_results_max = results.pivot_table(
    index="image_id",
    columns="model",
    values="confidence",
    aggfunc="max"
)

print("\nMax:")
print(wide_results_max)


# Compare mean and max:
#
# For img_003 + CNN:
# mean = (0.79 + 0.85) / 2 = 0.82
# max  = 0.85


# ============================================================
# Part 3 — Melt
# ============================================================

long_result_mean = (
    wide_results_mean
    .reset_index()
    .melt(
        id_vars="image_id",
        var_name="model",
        value_name="confidence"
    )
)

print("\nMelted:")
print(long_result_mean)


# ============================================================
# Part 4 — Stack / Unstack
# ============================================================

stacked = wide_results_mean.stack()

print("\nStacked:")
print(stacked)

print("\nStacked index:")
print(stacked.index)

print("\nStacked type:")
print(type(stacked))


unstacked = stacked.unstack()

print("\nUnstacked:")
print(unstacked)

print("\nUnstacked index:")
print(unstacked.index)

print("\nUnstacked columns:")
print(unstacked.columns)


# ============================================================
# Part 5 — Final Understanding
# ============================================================

# 1. What is the difference between pivot() and pivot_table()?
# pivot() requires each index + column combination to be unique.
# pivot_table() can handle duplicate combinations by aggregating
# their values using functions such as mean, max, min, sum, etc.


# 2. When would you use melt()?
# Use melt() when you want to convert Wide data into Long data.
# It is useful when several columns represent values of the same
# variable and you want those columns to become rows.


# 3. What is the difference between Long and Wide data?
# Long data stores observations in rows.
# Wide data spreads different categories/variables across columns.


# 4. Why did pivot() fail on this dataset?
# Because img_003 + CNN appears more than once.
# There are two confidence values for this combination:
# 0.79 and 0.85
#
# pivot() cannot decide which value should be placed in
# the img_003 / CNN cell.


# 5. Which operation converts Wide → Long?
# melt()


# 6. Which operation converts Long → Wide?
# pivot() or pivot_table()
````
