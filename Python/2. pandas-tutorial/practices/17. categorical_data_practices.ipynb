````python
results = pd.DataFrame({
    "model": [
        "CNN", "ResNet", "CNN", "ViT",
        "ResNet", "CNN", "ViT", "ResNet"
    ],
    "status": [
        "success", "success", "failed", "success",
        "success", "failed", "success", "success"
    ],
    "priority": [
        "high", "medium", "low", "high",
        "medium", "low", "high", "medium"
    ],
    "confidence": [
        0.91, 0.95, 0.42, 0.93,
        0.88, 0.51, 0.97, 0.89
    ]
})

results
````

````python
results["model"] = results["model"].astype("category")
results["status"] = results["status"].astype("category")
results["priority"] = results["priority"].astype("category")

results.dtypes
````

````python
print("Model categories:")
print(results["model"].cat.categories)

print("\nStatus categories:")
print(results["status"].cat.categories)

print("\nPriority categories:")
print(results["priority"].cat.categories)
````

````python
results["model"] = results["model"].cat.add_categories(["Transformer"])

results["model"].cat.categories
````

````python
results["model"] = results["model"].cat.remove_categories(["Transformer"])

results["model"].cat.categories
````

````python
results["model"] = results["model"].cat.remove_unused_categories()
results["status"] = results["status"].cat.remove_unused_categories()
results["priority"] = results["priority"].cat.remove_unused_categories()

results
````

````python
results["priority"] = results["priority"].cat.reorder_categories(
    ["low", "medium", "high"],
    ordered=True
)

results["priority"].cat.categories
````

````python
sorted_results = results.sort_values("priority")

sorted_results
````

````python
results["status"] = results["status"].cat.rename_categories({
    "success": "completed",
    "failed": "error"
})

results["status"].cat.categories
````

````python
model_category_count = results["model"].cat.categories.size

print(model_category_count)
````

````python
print(results["model"].dtype)
print(results["status"].dtype)
print(results["priority"].dtype)
````
