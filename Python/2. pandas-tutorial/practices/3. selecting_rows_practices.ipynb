iloc → Position
loc  → Label

# One row
df.iloc[0]

# Rows 0, 1, 2
df.iloc[0:3]

# Specific rows
df.iloc[[0, 3]]

# Last row
df.iloc[-1]

# Last 3 rows
df.iloc[-3:]

# First 2 rows

##### Exercise 1 — Selecting Rows from a Dataset
```text
Create the following DataFrame:

df = pd.DataFrame({
    "employee": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Frank"],
    "department": ["AI", "Backend", "AI", "DevOps", "Backend", "AI"],
    "experience": [2, 5, 3, 7, 4, 6],
    "salary": [5000, 7500, 6200, 9000, 7000, 8500]
})

Perform the following:

1. Select the first row.
2. Select the last row.
3. Select the first three rows.
4. Select rows at positions 1, 3, and 5.
5. Select the last three rows.
6. Print the type of the result for:
    df.iloc[0]
    and:
    df.iloc[[0]]
7. Print the shape of:
    df.iloc[0:3]

````python
import pandas as pd

df = pd.DataFrame({
    "employee": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Frank"],
    "department": ["AI", "Backend", "AI", "DevOps", "Backend", "AI"],
    "experience": [2, 5, 3, 7, 4, 6],
    "salary": [5000, 7500, 6200, 9000, 7000, 8500]
})

# 1. Select the first row
print(df.iloc[0])

# 2. Select the last row
print(df.iloc[-1])

# 3. Select the first three rows
print(df.iloc[0:3])

# 4. Select rows at positions 1, 3, and 5
print(df.iloc[[1, 3, 5]])

# 5. Select the last three rows
print(df.iloc[-3:])

# 6. Compare the types
print(type(df.iloc[0]))
print(type(df.iloc[[0]]))

# 7. Shape of the first three rows
print(df.iloc[0:3].shape)
````

##### Exercise 2 — Image Dataset
```text
Create:

images = pd.DataFrame({
    "image_id": ["img_001", "img_002", "img_003", "img_004", "img_005", "img_006", "img_007"],
    "label": ["cat", "dog", "car", "cat", "bird", "dog", "car"],
    "width": [128, 256, 128, 512, 256, 128, 512],
    "height": [128, 256, 128, 512, 256, 128, 512],
    "confidence": [0.91, 0.87, 0.76, 0.95, 0.68, 0.89, 0.73]
})

Perform the following:

1. Select the first image.
2. Select the images at positions 1, 4, and 6.
3. Select images from position 2 up to, but not including, position 6.
4. Select the final two images.
6. Compare the result and shape of:
    images.iloc[0]
    with:
    images.iloc[[0]]
7. Store the first five rows in a variable called sample.
8. Print sample.shape.

````python
images = pd.DataFrame({
    "image_id": [
        "img_001", "img_002", "img_003",
        "img_004", "img_005", "img_006", "img_007"
    ],
    "label": ["cat", "dog", "car", "cat", "bird", "dog", "car"],
    "width": [128, 256, 128, 512, 256, 128, 512],
    "height": [128, 256, 128, 512, 256, 128, 512],
    "confidence": [0.91, 0.87, 0.76, 0.95, 0.68, 0.89, 0.73]
})

# 1. Select the first image
print(images.iloc[0])

# 2. Select images at positions 1, 4, and 6
print(images.iloc[[1, 4, 6]])

# 3. Select positions 2 up to, but not including, 6
print(images.iloc[2:6])

# 4. Select the final two images
print(images.iloc[-2:])

# 6. Compare result and shape
print(images.iloc[0])
print(type(images.iloc[0]))
print(images.iloc[0].shape)

print(images.iloc[[0]])
print(type(images.iloc[[0]]))
print(images.iloc[[0]].shape)

# 7. Store the first five rows
sample = images.iloc[:5]

# 8. Print sample shape
print(sample.shape)
````
