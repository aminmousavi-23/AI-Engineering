##### Exercise 1 — Model Confidence Analysis
```text
Create a Pandas Series named confidence containing the following model confidence scores:

[0.91, 0.42, 0.78, 0.95, 0.63, 0.87]

Use the following labels as the Series index:

["img_01", "img_02", "img_03", "img_04", "img_05", "img_06"]

Perform the following tasks:

1. Print the entire Series.
2. Print the Series name, index, values, and dtype.
3. Access the confidence score of "img_04" using its label.
4. Calculate the mean, min, max, and sum of the confidence scores.
5. Create a boolean condition for scores greater than or equal to 0.80.
6. Use that condition to create a new Series containing only high-confidence predictions.
7. Multiply all confidence scores by 100 to convert them into percentages.
8. Print the final high-confidence results.

````python
import pandas as pd

confidence = pd.Series(
    data=[0.91, 0.42, 0.78, 0.95, 0.63, 0.87],
    index=["img_01", "img_02", "img_03", "img_04", "img_05", "img_06"],
    name="confidence",
)

# 1. Print the entire Series
print(confidence)

# 2. Series information
print(confidence.name)
print(confidence.index)
print(confidence.values)
print(confidence.dtype)

# 3. Access a value using its label
print(confidence["img_04"])

# 4. Aggregation
print(confidence.mean())
print(confidence.min())
print(confidence.max())
print(confidence.sum())

# 5. Create boolean condition
greater_than_or_equal = confidence >= 0.80

print(greater_than_or_equal)

# 6. Filter high-confidence predictions
high_confidence = confidence[greater_than_or_equal]

print(high_confidence)

# 7. Convert confidence scores to percentages
confidence_percentage = confidence * 100

print(confidence_percentage)

# 8. Final high-confidence results
high_confidence_percentage = high_confidence * 100

print(high_confidence_percentage)
````

##### Exercise 2 — Student Score Analysis
```text
Create a Pandas Series named scores using the following data:

{
    "Ali": 85,
    "Sara": 92,
    "Reza": 67,
    "Mina": 78,
    "Amir": 95,
    "Neda": 54
}

Perform the following tasks:

1. Print the Series.
2. Print its name, index, values, and dtype.
3. Access Sara's score using her index label.
4. Access Amir's score using positional indexing.
5. Calculate:
    Mean score
    Minimum score
    Maximum score
    Total score
6. Create a boolean condition for students with scores greater than 80.
7. Use the condition to create a new Series containing only students who scored above 80.
8. Add 5 bonus points to every student's score.
9. Create a new Series containing only students whose original score was below 70.
10. Print the final results.

````python
scores = pd.Series(
    data={
        "Ali": 85,
        "Sara": 92,
        "Reza": 67,
        "Mina": 78,
        "Amir": 95,
        "Neda": 54
    },
    name="scores"
)

# 1. Print the Series
print(scores)

# 2. Print name, index, values, and dtype
print(scores.name)
print(scores.index)
print(scores.values)
print(scores.dtype)

# 3. Access Sara's score using her label
print(scores["Sara"])

# 4. Access Amir's score using positional indexing
print(scores.iloc[4])

# 5. Calculate statistics
print("Mean:", scores.mean())
print("Min:", scores.min())
print("Max:", scores.max())
print("Total:", scores.sum())

# 6. Boolean condition for scores greater than 80
above_80_condition = scores > 80

print(above_80_condition)

# 7. Filter students who scored above 80
above_80 = scores[above_80_condition]

print(above_80)

# 8. Add 5 bonus points to every student
bonus_scores = scores + 5

print(bonus_scores)

# 9. Students whose original score was below 70
below_70 = scores[scores < 70]

print(below_70)

# 10. Final results
print("\n--- Final Results ---")
print("Above 80:")
print(above_80)

print("\nScores after bonus:")
print(bonus_scores)

print("\nOriginal scores below 70:")
print(below_70)
````

##### Exercise 1 — Image Dataset Indexing
```text
Create the following DataFrame:

data = {
    "image": ["cat.jpg", "dog.jpg", "bird.jpg", "car.jpg"],
    "label": ["cat", "dog", "bird", "car"],
    "confidence": [0.95, 0.87, 0.91, 0.78]
}

Perform the following tasks:

1. Create the DataFrame with the default index.
2. Print the current index.
3. Replace the default index with:
    ["IMG001", "IMG002", "IMG003", "IMG004"]
4. Print the DataFrame and its index.
5. Check whether the index is unique using is_unique.
6. Iterate over the index and print every image ID.
7. Create a new DataFrame where "image" is used as the index with set_index().
8. Use reset_index() to convert the index back into a column.

````python
import pandas as pd

data = {
    "image": ["cat.jpg", "dog.jpg", "bird.jpg", "car.jpg"],
    "label": ["cat", "dog", "bird", "car"],
    "confidence": [0.95, 0.87, 0.91, 0.78]
}

# 1. Create DataFrame with default index
df = pd.DataFrame(data)

# 2. Print current index
print(df.index)

# 3. Replace default index
df.index = ["IMG001", "IMG002", "IMG003", "IMG004"]

# 4. Print DataFrame and index
print(df)
print(df.index)

# 5. Check whether index is unique
print(df.index.is_unique)

# 6. Iterate over the index
for image_id in df.index:
    print(image_id)

# 7. Use "image" as the index
df_by_image = df.set_index("image")

print(df_by_image)

# 8. Convert index back into a column
df_reset = df_by_image.reset_index()

print(df_reset)
````
