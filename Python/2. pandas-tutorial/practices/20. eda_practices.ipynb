# EDA — Exploratory Data Analysis

**EDA (Exploratory Data Analysis)** یا **تحلیل اکتشافی داده‌ها**، فرایند بررسی و شناخت یک **Dataset** قبل از انجام تحلیل‌های اصلی یا ساخت مدل **Machine Learning** است.

## هدف EDA

هدف EDA این است که بفهمیم:

* داده‌ها چه ساختاری دارند؟
* چه نوع متغیرهایی داریم؟
* داده‌ها چقدر کامل و باکیفیت هستند؟
* آیا **Missing Value** یا **Duplicate** داریم؟
* مقدارها چه **Distribution**ای دارند؟
* آیا **Outlier** یا مقدار غیرعادی وجود دارد؟
* بین متغیرها چه **Relationship**هایی وجود دارد؟
* آیا داده برای مرحله بعدی، مثل **Machine Learning**، آماده است؟

### به‌صورت ساده

```text
Raw Data
   ↓
Explore
   ↓
Understand
   ↓
Detect Problems
   ↓
Find Patterns
   ↓
Prepare for Analysis / ML
```

## EDA در Machine Learning

در **Machine Learning** معمولاً با این ساختار مواجه می‌شویم:

```text
Dataset
 ├── Features (X)
 └── Target / Label (y)
```

مثلاً در یک Dataset مربوط به **تشخیص تصویر**:

| image_id | model   | confidence | label  |
| -------- | ------- | ---------- | ------ |
| Sample   | Feature | Feature    | Target |

* `image_id` → شناسه نمونه
* `model` → یک Feature
* `confidence` → یک Feature
* `label` → Target / Label

EDA کمک می‌کند قبل از **Training** بفهمیم `X` و `y` چه وضعیتی دارند و آیا مشکلی در داده وجود دارد یا نه.

````python
import pandas as pd
````

````python
# Question:
# Load the large_predictions.csv dataset into a DataFrame.

file_path = "../data/large_predictions.csv"

df = pd.read_csv(file_path)

df.head()
````

````python
# Question:
# Find:
# 1. Number of rows and columns
# 2. First 5 rows
# 3. Last 5 rows
# 4. Column names
# 5. Data types

# Answer:

print("Shape:")
print(df.shape)

print("\nFirst 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nColumns:")
print(df.columns)

print("\nData types:")
print(df.dtypes)
````

````python
# Question:
# Generate descriptive statistics for:
# confidence, inference_time_ms, label, and prediction.

# Answer:

df[
    ["confidence", "inference_time_ms", "label", "prediction"]
].describe()
````

````python
# Question:
# Find the number of missing values in each column.

# Answer:

df.isna().sum()
````

````python
# Question:
# Find the number of duplicated rows.

# Answer:

df.duplicated().sum()
````

````python
# Question:
# Find:
# 1. The number of predictions for each model.
# 2. The percentage of predictions for each model.

# Answer:

print("Counts:")
print(df["model"].value_counts())

print("\nPercentages:")
print(df["model"].value_counts(normalize=True) * 100)
````

````python
# Question:
# Create a "correct" column that indicates whether
# prediction is equal to label.
# Then calculate the overall accuracy.

# Answer:

df["correct"] = df["prediction"] == df["label"]

accuracy = df["correct"].mean()

print(f"Accuracy: {accuracy:.2%}")
````

````python
# Question:
# For each model, calculate:
# 1. Number of predictions
# 2. Mean confidence
# 3. Accuracy
# 4. Mean inference time

# Answer:

model_analysis = df.groupby("model").agg(
    predictions=("model", "size"),
    mean_confidence=("confidence", "mean"),
    accuracy=("correct", "mean"),
    mean_inference_time_ms=("inference_time_ms", "mean")
)

model_analysis["accuracy"] *= 100

model_analysis
````

````python
# Question:
# Find the minimum, maximum, mean, median,
# and standard deviation of confidence.

# Answer:

confidence_stats = df["confidence"].agg([
    "min",
    "max",
    "mean",
    "median",
    "std"
])

confidence_stats
````

````python
# Question:
# Find all predictions with confidence below 0.60.
# Then calculate their count and percentage.

# Answer:

low_confidence = df[df["confidence"] < 0.60]

print("Count:")
print(len(low_confidence))

print("\nPercentage:")
print(f"{len(low_confidence) / len(df):.2%}")
````

````python
# Question:
# Calculate the average inference time for each model
# and sort the result from lowest to highest.

# Answer:

avg_inference_time = (
    df.groupby("model")["inference_time_ms"]
    .mean()
    .sort_values()
)

avg_inference_time
````

````python
# Question:
# Calculate the correlation between:
# confidence and inference_time_ms.

# Answer:

correlation = df["confidence"].corr(
    df["inference_time_ms"]
)

print(f"Correlation: {correlation:.4f}")
````

````python
# Question:
# Calculate the accuracy for each dataset.

# Answer:

dataset_accuracy = (
    df.groupby("dataset")["correct"]
    .mean()
    .mul(100)
)

dataset_accuracy
````

````python
# Question:
# Calculate accuracy for every combination of
# model and dataset.

# Answer:

model_dataset_accuracy = (
    df.groupby(["model", "dataset"])["correct"]
    .mean()
    .mul(100)
)

model_dataset_accuracy
````

````python
# Question:
# Create a final summary containing:
# - Number of rows
# - Number of columns
# - Missing values
# - Duplicate rows
# - Number of models
# - Overall accuracy
# - Average confidence
# - Average inference time

# Answer:

eda_summary = {
    "rows": len(df),
    "columns": df.shape[1],
    "missing_values": df.isna().sum().sum(),
    "duplicate_rows": df.duplicated().sum(),
    "number_of_models": df["model"].nunique(),
    "overall_accuracy": df["correct"].mean(),
    "average_confidence": df["confidence"].mean(),
    "average_inference_time_ms": df["inference_time_ms"].mean()
}

eda_summary
````
